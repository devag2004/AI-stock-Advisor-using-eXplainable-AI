```python
import pandas as pd
import pandas_ta as ta

reliance_data = pd.read_csv("reliance_data.csv", index_col=0, parse_dates=True)
nifty50_data = pd.read_csv("nifty50_data.csv", index_col=0, parse_dates=True)

def add_indicators(df):
    df = df.copy()

    # Trend Indicators
    df["SMA_20"] = ta.sma(df["Close"], length=20)
    df["EMA_20"] = ta.ema(df["Close"], length=20)
    df["EMA_50"] = ta.ema(df["Close"], length=50)
    macd = ta.macd(df["Close"])
    df = df.join(macd)

    # Momentum Indicators
    df["RSI_14"] = ta.rsi(df["Close"], length=14)
    stoch = ta.stoch(df["High"], df["Low"], df["Close"])
    df = df.join(stoch)

    # Volatility Indicators
    bbands = ta.bbands(df["Close"], length=20, std=2)
    df = df.join(bbands)
    df["ATR_14"] = ta.atr(df["High"], df["Low"], df["Close"], length=14)

    # Volume Indicator
    df["OBV"] = ta.obv(df["Close"], df["Volume"])

    return df

# Apply indicators
reliance_with_indicators = add_indicators(reliance_data)
nifty50_with_indicators = add_indicators(nifty50_data)

# Save to CSV
reliance_with_indicators.to_csv("reliance_with_indicators.csv")
nifty50_with_indicators.to_csv("nifty50_with_indicators.csv")

print("Technical indicators added and saved.")
print("\nReliance sample with indicators:")
print(reliance_with_indicators.tail(5))
```

```python
import os
import random
import math
from datetime import timedelta
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


CSV_PATH = "reliance_with_indicators.csv"
DATE_COL = None
START_DATE = "2019-01-01"
SEQ_LEN = 60
BATCH_SIZE = 64
HIDDEN_SIZE = 128
NUM_LAYERS = 2
DROPOUT = 0.2
LR = 1e-3
WEIGHT_DECAY = 1e-5
EPOCHS = 100
PATIENCE = 8
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
TEST_DAYS = 30

print("Device:", DEVICE)
```

```python
df = pd.read_csv(CSV_PATH, index_col=0, parse_dates=True)
df.sort_index(inplace=True)

df_timezone = df.index.tz

start_date_aware = pd.to_datetime(START_DATE).tz_localize(df_timezone)

df = df[df.index >= start_date_aware].copy()

df = df.ffill().bfill()
df.dropna(inplace=True)

print("Data range:", df.index.min(), "to", df.index.max())
print("Shape after cleaning:", df.shape)

raw_df = df.copy()

feature_cols = df.columns.tolist()   # all columns
target_cols = ["Open", "Close"]

for t in target_cols:
    if t not in df.columns:
        raise ValueError(f"Target column {t} not found in CSV columns.")

if TEST_DAYS >= len(df):
    raise ValueError("TEST_DAYS is too large for dataset length.")

train_val_df = df.iloc[:-TEST_DAYS]
test_df = df.iloc[-TEST_DAYS:].copy()

print("Train+Val rows:", len(train_val_df), "Test rows:", len(test_df))

val_frac = 0.1
val_size = int(len(train_val_df) * val_frac)
train_df = train_val_df.iloc[:-val_size].copy()
val_df = train_val_df.iloc[-val_size:].copy()

print("Train rows:", len(train_df), "Val rows:", len(val_df))

scaler = MinMaxScaler(feature_range=(0, 1))
scaler.fit(train_df[feature_cols].values)

train_scaled = pd.DataFrame(scaler.transform(train_df[feature_cols]), index=train_df.index, columns=feature_cols)
val_scaled = pd.DataFrame(scaler.transform(val_df[feature_cols]), index=val_df.index, columns=feature_cols)
test_scaled = pd.DataFrame(scaler.transform(test_df[feature_cols]), index=test_df.index, columns=feature_cols)

def inverse_targets(scaled_arr, feature_cols=feature_cols, target_cols=target_cols, scaler=scaler):
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)

    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(feature_cols)))

    for i, col in enumerate(feature_cols):
        if col in target_cols:
            idx = target_cols.index(col)
            filler[:, feature_cols.index(col)] = scaled_arr[:, idx]

    inv = scaler.inverse_transform(filler)
    return inv[:, [feature_cols.index(c) for c in target_cols]]

class TimeSeriesDataset(Dataset):
    def __init__(self, df_scaled, seq_len=SEQ_LEN, feature_cols=feature_cols, target_cols=target_cols):
        self.X = df_scaled.values.astype(np.float32)
        self.feature_cols = feature_cols
        self.target_cols = target_cols
        self.seq_len = seq_len
        self.indices = []
        for i in range(seq_len - 1, len(self.X) - 1):
            self.indices.append(i)
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, idx):
        i = self.indices[idx]
        seq = self.X[i - (self.seq_len - 1) : i + 1]
        target_row = self.X[i + 1]
        target = np.array([target_row[self.feature_cols.index(c)] for c in self.target_cols], dtype=np.float32)
        return torch.from_numpy(seq), torch.from_numpy(target)

train_ds = TimeSeriesDataset(train_scaled, seq_len=SEQ_LEN)
val_ds = TimeSeriesDataset(val_scaled, seq_len=SEQ_LEN)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, drop_last=False)

print("Train batches:", len(train_loader), "Val batches:", len(val_loader))
input_size = len(feature_cols)

class LSTMModel(nn.Module):
    def __init__(self, input_size, hidden_size=HIDDEN_SIZE, num_layers=NUM_LAYERS, output_size=2, dropout=DROPOUT):
        super().__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        self.lstm = nn.LSTM(input_size=input_size,
                            hidden_size=hidden_size,
                            num_layers=num_layers,
                            dropout=dropout,
                            batch_first=True)
        self.fc = nn.Sequential(
            nn.Linear(hidden_size, hidden_size//2),
            nn.ReLU(),
            nn.Linear(hidden_size//2, output_size)
        )
    def forward(self, x):
        out, _ = self.lstm(x)
        out = out[:, -1, :]
        out = self.fc(out)
        return out

model = LSTMModel(input_size=input_size).to(DEVICE)
print(model)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)

best_val_loss = float("inf")
counter = 0

for epoch in range(1, EPOCHS + 1):
    model.train()
    train_losses = []
    for xb, yb in train_loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)
        preds = model(xb)
        loss = criterion(preds, yb)
        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        train_losses.append(loss.item())
    avg_train = np.mean(train_losses)

    model.eval()
    val_losses = []
    with torch.no_grad():
        for xb, yb in val_loader:
            xb = xb.to(DEVICE)
            yb = yb.to(DEVICE)
            preds = model(xb)
            loss = criterion(preds, yb)
            val_losses.append(loss.item())
    avg_val = np.mean(val_losses)

    scheduler.step(avg_val)

    print(f"Epoch {epoch:03d} | Train Loss: {avg_train:.6f} | Val Loss: {avg_val:.6f}")

    if avg_val < best_val_loss - 1e-6:
        best_val_loss = avg_val
        counter = 0
        torch.save({
            "model_state_dict": model.state_dict(),
            "scaler": scaler,
            "feature_cols": feature_cols,
            "target_cols": target_cols
        }, "lstm_reliance_checkpoint.pth")
        print("  -> saved best model")
    else:
        counter += 1
        if counter >= PATIENCE:
            print("Early stopping triggered. Stopping training.")
            break

checkpoint = torch.load("lstm_reliance_checkpoint.pth", map_location=DEVICE, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
scaler = checkpoint["scaler"]


full_scaled = pd.DataFrame(scaler.transform(df[feature_cols].values), index=df.index, columns=feature_cols)
test_inputs = []
test_targets = []
test_dates = []
for i in range(len(df) - TEST_DAYS, len(df)):
    seq_end = i - 1
    seq_start = seq_end - (SEQ_LEN - 1)
    if seq_start < 0:
        continue
    seq = full_scaled.iloc[seq_start: seq_end + 1].values.astype(np.float32)
    target_row = full_scaled.iloc[i].values.astype(np.float32)
    target = np.array([target_row[feature_cols.index(c)] for c in target_cols], dtype=np.float32)
    test_inputs.append(seq)
    test_targets.append(target)
    test_dates.append(df.index[i])

test_inputs = np.stack(test_inputs)
test_targets = np.stack(test_targets)

model.eval()
with torch.no_grad():
    X_test_t = torch.from_numpy(test_inputs).to(DEVICE)
    preds_scaled = model(X_test_t).cpu().numpy()

preds_inv = inverse_targets(preds_scaled)
targets_inv = inverse_targets(test_targets)

rmse_open = math.sqrt(mean_squared_error(targets_inv[:,0], preds_inv[:,0]))
rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
mae_open = mean_absolute_error(targets_inv[:,0], preds_inv[:,0])
mae_close = mean_absolute_error(targets_inv[:,1], preds_inv[:,1])

print(f"Test RMSE Open: {rmse_open:.4f} | Close: {rmse_close:.4f}")
print(f"Test MAE  Open: {mae_open:.4f} | Close: {mae_close:.4f}")

plt.figure(figsize=(12,5))
plt.plot(test_dates, targets_inv[:,0], label="Actual Open")
plt.plot(test_dates, preds_inv[:,0], label="Pred Open", linestyle="--")
plt.title("Test: Open price - Actual vs Predicted")
plt.legend()
plt.show()

plt.figure(figsize=(12,5))
plt.plot(test_dates, targets_inv[:,1], label="Actual Close")
plt.plot(test_dates, preds_inv[:,1], label="Pred Close", linestyle="--")
plt.title("Test: Close price - Actual vs Predicted")
plt.legend()
plt.show()

roll_start_idx = len(df) - TEST_DAYS - 1
seq_start = roll_start_idx - (SEQ_LEN - 1)
if seq_start < 0:
    raise ValueError("Not enough history to build the initial sequence for iterative forecasting. Increase dataset or reduce SEQ_LEN.")
rolling_seq = full_scaled.iloc[seq_start: roll_start_idx + 1].values.astype(np.float32)

iter_preds_scaled = []
iter_dates = []
current_seq = rolling_seq.copy()

for step in range(TEST_DAYS):
    input_tensor = torch.from_numpy(current_seq[np.newaxis, :, :]).to(DEVICE)
    with torch.no_grad():
        pred_scaled = model(input_tensor).cpu().numpy().reshape(-1)
    iter_preds_scaled.append(pred_scaled.copy())

    last_row = current_seq[-1].copy()
    for j, col in enumerate(feature_cols):
        if col in target_cols:
            last_row[feature_cols.index(col)] = pred_scaled[target_cols.index(col)]
    current_seq = np.vstack([current_seq[1:], last_row])

    last_date = df.index[roll_start_idx + step]
    if roll_start_idx + step + 1 < len(df):
        pred_date = df.index[roll_start_idx + step + 1]
    else:
        pred_date = last_date + pd.Timedelta(days=1)
    iter_dates.append(pred_date)

iter_preds_scaled = np.stack(iter_preds_scaled)
iter_preds_inv = inverse_targets(iter_preds_scaled)

plt.figure(figsize=(12,5))
plt.plot(iter_dates, iter_preds_inv[:,0], label="Forecast Open (30d)")
plt.title("30-day Iterative Forecast - Open")
plt.xticks(rotation=45)
plt.legend()
plt.show()

plt.figure(figsize=(12,5))
plt.plot(iter_dates, iter_preds_inv[:,1], label="Forecast Close (30d)")
plt.title("30-day Iterative Forecast - Close")
plt.xticks(rotation=45)
plt.legend()
plt.show()

pred_df = pd.DataFrame(iter_preds_inv, columns=["Pred_Open", "Pred_Close"], index=iter_dates)
print("30-day forecast (open & close):")
display(pred_df)

torch.save({
    "model_state_dict": model.state_dict(),
    "scaler": scaler,
    "feature_cols": feature_cols,
    "target_cols": target_cols
}, "lstm_reliance_final.pth")

print("Done. Model and forecasts saved to disk.")
```

```python
# === Simplified + Fixed XAI suite (No DeepLift, SHAP fixed) ===
from captum.attr import IntegratedGradients, Saliency
import shap
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math
from sklearn.metrics import mean_squared_error
sns.set()

device = DEVICE if "DEVICE" in globals() else (torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu"))
model.to(device)
model.eval()

# Helper: predict unscaled
def predict_unscaled(X_np):
    with torch.no_grad():
        t = torch.from_numpy(X_np.astype(np.float32)).to(device)
        out = model(t)
        if isinstance(out, (tuple, list)):
            out = out[0]
        preds_scaled = out.cpu().numpy()
    preds_unscaled = inverse_targets(preds_scaled, feature_cols, target_cols, scaler)
    return preds_unscaled

# Base RMSE
base_preds = predict_unscaled(test_inputs)
base_rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], base_preds[:,1]))
print(f"Base test RMSE (Close): {base_rmse_close:.4f}\n")

# Safe run wrapper for Captum RNNs
def run_in_train_mode(fn, *args, **kwargs):
    prev_mode = model.training
    try:
        model.train()
        return fn(*args, **kwargs)
    finally:
        if not prev_mode:
            model.eval()

# -----------------------
# 1️⃣ Integrated Gradients (Local explanation)
# -----------------------
def ig_local_explanation(sample_idx=0, target_idx=1, n_steps=50):
    class Wrapper(torch.nn.Module):
        def __init__(self, base, target_idx):
            super().__init__(); self.base = base; self.target_idx = target_idx
        def forward(self, x):
            out = self.base(x)
            if isinstance(out, (tuple, list)): out = out[0]
            return out[:, self.target_idx]

    wmod = Wrapper(model, target_idx).to(device)
    ig = IntegratedGradients(wmod)

    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).to(device).float()
    baseline = torch.zeros_like(inp).to(device)

    def _compute_ig():
        attrs, delta = ig.attribute(inputs=inp, baselines=baseline, n_steps=n_steps, return_convergence_delta=True)
        return attrs.cpu().detach().numpy()[0], float(delta)

    at_np, delta = run_in_train_mode(_compute_ig)

    feat_scores = np.sum(np.abs(at_np), axis=0)
    total = feat_scores.sum() if feat_scores.sum() != 0 else 1e-9
    pct = 100 * feat_scores / total
    df_top = pd.DataFrame({"feature": feature_cols, "score": feat_scores, "pct": pct}).sort_values("score", ascending=False)

    signed = np.sum(at_np, axis=0)
    top3 = df_top.head(3)
    summary = "; ".join([f"{f} ({'↑' if np.sum(at_np[:, feature_cols.index(f)])>0 else '↓'}, {p:.1f}%)" for f, p in zip(top3['feature'], top3['pct'])])
    sentence = f"Top IG contributors: {summary}."

    plt.figure(figsize=(10,6))
    sns.heatmap(at_np.T, cmap="bwr", center=0, yticklabels=feature_cols)
    plt.title(f"Integrated Gradients (sample {sample_idx}) | delta={delta:.2e}")
    plt.xlabel("Time step (older → newer)")
    plt.tight_layout(); plt.show()

    return df_top, at_np, sentence

ig_df, ig_attr, ig_sentence = ig_local_explanation(sample_idx=0, target_idx=1)
print(ig_sentence)
display(ig_df.head(10))

# -----------------------
# 2️⃣ Saliency (grad × input)
# -----------------------
def saliency_local(sample_idx=0, target_idx=1):
    class Wrapper(torch.nn.Module):
        def __init__(self, base, target_idx):
            super().__init__(); self.base = base; self.target_idx = target_idx
        def forward(self, x):
            out = self.base(x)
            if isinstance(out, (tuple, list)): out = out[0]
            return out[:, self.target_idx]

    wmod = Wrapper(model, target_idx).to(device)
    sal = Saliency(wmod)

    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).to(device).float()
    inp.requires_grad = True

    def _compute_sal():
        attr = sal.attribute(inp)
        return attr.cpu().detach().numpy()[0]

    sal_np = run_in_train_mode(_compute_sal)
    grad_x_input = sal_np * test_inputs[sample_idx]

    feat_scores = np.sum(np.abs(grad_x_input), axis=0)
    df = pd.DataFrame({"feature": feature_cols, "score": feat_scores}).sort_values("score", ascending=False)

    plt.figure(figsize=(10,6))
    sns.heatmap(grad_x_input.T, cmap="bwr", center=0, yticklabels=feature_cols)
    plt.title("Saliency Map (grad × input)")
    plt.xlabel("Time step (older → newer)")
    plt.tight_layout(); plt.show()
    return df, grad_x_input

sal_df, sal_map = saliency_local(sample_idx=0, target_idx=1)
print("Top Saliency features:")
display(sal_df.head(8))

# -----------------------
# 3️⃣ Permutation Feature Importance (Global)
# -----------------------
def permutation_feature_importance(X_np, y_true, feature_names, target_idx=1, n_repeats=3):
    with torch.no_grad():
        preds_unscaled = predict_unscaled(X_np)
    base_rmse = math.sqrt(mean_squared_error(y_true, preds_unscaled[:, target_idx]))
    print(f"Base RMSE (target {target_idx}): {base_rmse:.4f}")

    importances = []
    N, T, F = X_np.shape
    for f_idx, f_name in enumerate(feature_names):
        rmses = []
        for _ in range(n_repeats):
            Xp = X_np.copy()
            perm = Xp[:,:,f_idx].flatten()
            np.random.shuffle(perm)
            Xp[:,:,f_idx] = perm.reshape(N, T)
            preds_perm = predict_unscaled(Xp)
            rmse = math.sqrt(mean_squared_error(y_true, preds_perm[:, target_idx]))
            rmses.append(rmse)
        delta = np.mean(rmses) - base_rmse
        importances.append((f_name, delta))
    df = pd.DataFrame(importances, columns=["feature", "delta_RMSE"]).sort_values("delta_RMSE", ascending=False)

    plt.figure(figsize=(8,5))
    sns.barplot(y="feature", x="delta_RMSE", data=df.head(10), palette="viridis")
    plt.title("Permutation Feature Importance (Global)")
    plt.xlabel("Increase in RMSE when permuted")
    plt.tight_layout(); plt.show()
    return df

perm_df = permutation_feature_importance(test_inputs, targets_inv[:,1], feature_cols, target_idx=1, n_repeats=3)
display(perm_df.head(10))

# -----------------------
# 4️⃣ SHAP (Fixed version using average window)
# -----------------------
def shap_time_series_fixed(n_background=40, explain_n=3, nsamples=200):
    X_avg = np.mean(test_inputs, axis=1)  # average last 60 timesteps → shape (N, features)
    bg_idx = np.random.choice(len(X_avg), min(n_background, len(X_avg)), replace=False)
    background = X_avg[bg_idx]

    def predict_flat(x_flat):
        # expand each flat vector back to a constant sequence
        x_seq = np.repeat(x_flat[:, np.newaxis, :], SEQ_LEN, axis=1).astype(np.float32)
        preds = predict_unscaled(x_seq)
        return preds[:, 1]  # Close

    explainer = shap.KernelExplainer(predict_flat, background)
    shap_vals = explainer.shap_values(X_avg[:explain_n], nsamples=nsamples)

    shap_vals = np.array(shap_vals)
    for i in range(min(explain_n, len(shap_vals))):
        df_sh = pd.DataFrame({"feature": feature_cols, "shap": shap_vals[i]})
        df_sh["abs"] = np.abs(df_sh["shap"])
        df_sh = df_sh.sort_values("abs", ascending=False)
        print(f"\nSHAP explanation for test sample {i}:")
        display(df_sh.head(10))
    shap.summary_plot(shap_vals, X_avg[:explain_n], feature_names=feature_cols, show=True)

print("\nRunning SHAP (this may take 1–2 mins)...")
shap_time_series_fixed(n_background=40, explain_n=2, nsamples=200)

print("\n✅ XAI complete: IG, Saliency, Permutation, and SHAP done.")
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math

# Set seeds and device
SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Data loading and preprocessing same as before
df = pd.read_csv("reliance_with_indicators.csv", index_col=0, parse_dates=True).ffill().bfill().dropna().sort_index()
TARGET_COLS = ["Open", "Close"]
FEATURE_COLS = df.columns.tolist()

train_df = df.loc["2019-01-01":"2024-12-31"]
test_df = df.loc["2025-01-01":]

scaler = MinMaxScaler()
scaler.fit(train_df[FEATURE_COLS])
train_scaled = pd.DataFrame(scaler.transform(train_df[FEATURE_COLS]), columns=FEATURE_COLS, index=train_df.index)
test_scaled = pd.DataFrame(scaler.transform(test_df[FEATURE_COLS]), columns=FEATURE_COLS, index=test_df.index)

SEQ_LEN = 60
LOWER_WINDOW = 5

class HierarchicalDataset(Dataset):
    def __init__(self, df_scaled, seq_len, lower_window):
        self.X = df_scaled.values.astype(np.float32)
        self.seq_len = seq_len
        self.lower_window = lower_window
        self.n_features = df_scaled.shape[1]

    def __len__(self):
        return len(self.X) - self.seq_len

    def __getitem__(self, idx):
        seq = self.X[idx: idx + self.seq_len]
        target = self.X[idx + self.seq_len, [FEATURE_COLS.index(c) for c in TARGET_COLS]]
        return torch.from_numpy(seq), torch.from_numpy(target)

train_ds = HierarchicalDataset(train_scaled, SEQ_LEN, LOWER_WINDOW)
test_ds = HierarchicalDataset(test_scaled, SEQ_LEN, LOWER_WINDOW)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, drop_last=True, num_workers=0)
test_loader = DataLoader(test_ds, batch_size=64, shuffle=False, drop_last=False, num_workers=0)

# Temporal Attention Module
class TemporalAttention(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.attn = nn.Sequential(
            nn.Linear(input_dim, input_dim//2),
            nn.Tanh(),
            nn.Linear(input_dim//2, 1)
        )

    def forward(self, x):
        # x: (batch, seq_len, features)
        scores = self.attn(x).squeeze(-1)  # (batch, seq_len)
        weights = torch.softmax(scores, dim=1).unsqueeze(-1)  # (batch, seq_len, 1)
        weighted_sum = (x * weights).sum(dim=1)  # (batch, features)
        return weighted_sum, weights

# Main Model: CNN + Hierarchical LSTM + Temporal + Multihead Attention
class HierarchicalCNNLSTMAttn(nn.Module):
    def __init__(self, n_features, lower_window=5, conv_channels=32, hidden_lower=64, hidden_higher=64, n_heads=4, target_dim=2):
        super().__init__()
        self.n_features = n_features
        self.lower_window = lower_window

        # CNN layer for lower window feature extraction
        self.conv1d = nn.Conv1d(n_features, conv_channels, kernel_size=3, padding=1)
        self.relu = nn.ReLU()

        # Lower-level LSTM (operates on CNN features)
        self.lower_lstm = nn.LSTM(conv_channels, hidden_lower, batch_first=True)

        # Number of lower-level windows in full sequence
        self.num_lower_windows = SEQ_LEN // lower_window

        # Temporal Attention on lower LSTM outputs
        self.temporal_attn = TemporalAttention(hidden_lower)

        # Higher-level LSTM on weighted lower outputs
        self.higher_lstm = nn.LSTM(hidden_lower, hidden_higher, batch_first=True)

        # Multi-head self-attention on higher LSTM outputs
        self.multihead_attn = nn.MultiheadAttention(embed_dim=hidden_higher, num_heads=n_heads, batch_first=True)

        # Final MLP for prediction
        self.fc = nn.Sequential(
            nn.Linear(hidden_higher, 32),
            nn.ReLU(),
            nn.Linear(32, target_dim)
        )

    def forward(self, x):
        batch_size = x.size(0)
        # Create lower windows
        x = x.unfold(dimension=1, size=self.lower_window, step=self.lower_window)  # (batch, num_windows, lower_window, n_features)
        x = x.contiguous().view(batch_size * self.num_lower_windows, self.lower_window, self.n_features)  # flatten batch & windows

        # CNN expects (batch, channels, width)
        x = x.permute(0, 2, 1)  # (batch*num_windows, n_features, lower_window)
        x = self.conv1d(x)
        x = self.relu(x)
        x = x.permute(0, 2, 1)  # (batch*num_windows, lower_window, conv_channels)

        # Lower LSTM on CNN features
        lower_out, _ = self.lower_lstm(x)
        lower_out = lower_out[:, -1, :]  # last hidden state

        # Reshape back to (batch, num_windows, hidden_lower)
        lower_out = lower_out.view(batch_size, self.num_lower_windows, -1)

        # Temporal attention on lower LSTM outputs
        weighted_lower, attn_weights = self.temporal_attn(lower_out)  # (batch, hidden_lower), attn_weights (batch, num_windows, 1)

        # Higher LSTM
        higher_out, _ = self.higher_lstm(lower_out)

        # Multi-head self-attention
        attn_out, _ = self.multihead_attn(higher_out, higher_out, higher_out)

        # Pool output by taking last timestep
        out = attn_out[:, -1, :]  # (batch, hidden_higher)

        # Final prediction
        preds = self.fc(out)
        return preds, attn_weights

# Instantiate model
model = HierarchicalCNNLSTMAttn(len(FEATURE_COLS), lower_window=LOWER_WINDOW).to(DEVICE)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

# Training loop
EPOCHS = 60
for epoch in range(1, EPOCHS+1):
    model.train()
    epoch_losses = []
    for xb, yb in train_loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        optimizer.zero_grad()
        preds, _ = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        optimizer.step()
        epoch_losses.append(loss.item())
    print(f"Epoch {epoch} | Train Loss: {np.mean(epoch_losses):.6f}")

# Evaluation
model.eval()
all_preds = []
all_targets = []
all_attn_weights = []
with torch.no_grad():
    for xb, yb in test_loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        preds, attn_weights = model(xb)
        all_preds.append(preds.cpu().numpy())
        all_targets.append(yb.cpu().numpy())
        all_attn_weights.append(attn_weights.cpu().numpy())

all_preds = np.vstack(all_preds)
all_targets = np.vstack(all_targets)
all_attn_weights = np.vstack(all_attn_weights)  # shape (num_batches * batch_size, num_windows, 1)

# Inverse scaling helper
def inverse_targets(scaled_arr):
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    filler = np.zeros((scaled_arr.shape[0], len(FEATURE_COLS)))
    for i, col in enumerate(FEATURE_COLS):
        if col in TARGET_COLS:
            filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

preds_inv = inverse_targets(all_preds)
targets_inv = inverse_targets(all_targets)

# Compute metrics
def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / (y_true + 1e-8))) * 100

rmse_open = math.sqrt(mean_squared_error(targets_inv[:, 0], preds_inv[:, 0]))
rmse_close = math.sqrt(mean_squared_error(targets_inv[:, 1], preds_inv[:, 1]))
mae_open = mean_absolute_error(targets_inv[:, 0], preds_inv[:, 0])
mae_close = mean_absolute_error(targets_inv[:, 1], preds_inv[:, 1])
mape_open = mape(targets_inv[:, 0], preds_inv[:, 0])
mape_close = mape(targets_inv[:, 1], preds_inv[:, 1])

print(f"Test RMSE Open: {rmse_open:.4f} | Close: {rmse_close:.4f}")
print(f"Test MAE  Open: {mae_open:.4f} | Close: {mae_close:.4f}")
print(f"Test MAPE Open: {mape_open:.2f}% | Close: {mape_close:.2f}%")

# Plot Close price actual vs predicted
dates = test_df.index[SEQ_LEN:]
plt.figure(figsize=(12,5))
plt.plot(dates, targets_inv[:, 1], label="Actual Close")
plt.plot(dates, preds_inv[:, 1], label="Predicted Close", linestyle="--")
plt.title("Hierarchical CNN+LSTM with Temporal & Self-Attention – Close Price Forecast")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python

# AB=CD Strategy Agent

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math
sns.set()

if 'preds_inv' not in globals() or 'targets_inv' not in globals():
    raise RuntimeError("preds_inv and targets_inv not found. Run the forecasting cell before this cell.")

forecast_close = np.asarray(preds_inv[:, 1], dtype=float)   # predicted Close (agent uses this to decide)
actual_close   = np.asarray(targets_inv[:, 1], dtype=float) # actual Close (execution happens at these prices)

if len(forecast_close) != len(actual_close):
    raise RuntimeError(f"Length mismatch: forecast_close ({len(forecast_close)}) vs actual_close ({len(actual_close)}).")

n = len(forecast_close)
if n == 0:
    raise RuntimeError("No test predictions found (length 0).")

if 'test_df' in globals():
    try:
        if len(test_df) >= (SEQ_LEN + n):
            test_dates = test_df.index[SEQ_LEN: SEQ_LEN + n]
        else:
            test_dates = test_df.index[-n:]
    except Exception:
        test_dates = np.arange(n)
else:
    test_dates = np.arange(n)

def abcd_signal_from_series(close_series, skip_loop=4, ma_window=7):
    """
    close_series: 1D array-like (forecasted closes)
    skip_loop: stride in nested search (keeps algorithm fast)
    ma_window: moving average smoothing window (int)
    """
    arr = np.asarray(close_series, dtype=float)
    if len(arr) < ma_window + 4:
        # too short to compute AB=CD
        return np.zeros(len(arr), dtype=float)
    ma = pd.Series(arr).rolling(ma_window, min_periods=1).mean().values
    x = []
    L = len(ma)
    # nested loops (a < b < c < d) with stride skip_loop
    for a in range(0, L):
        for b in range(a + 1, L, skip_loop):
            for c in range(b + 1, L, skip_loop):
                for d in range(c + 1, L, skip_loop):
                    # AB up, BC retrace between A and B, CD up above B (simple pattern)
                    if (ma[b] > ma[a]) and (ma[c] < ma[b] and ma[c] > ma[a]) and (ma[d] > ma[b]):
                        x.append((a, b, c, d))
    if len(x) == 0:
        return np.zeros(len(arr), dtype=float)
    x_np = np.array(x)
    ac = x_np[:, 0].tolist() + x_np[:, 2].tolist()
    bd = x_np[:, 1].tolist() + x_np[:, 3].tolist()
    ac_set = set(ac)
    bd_set = set(bd)
    signal = np.zeros(len(arr), dtype=float)
    buy_idxs = sorted(list(ac_set - bd_set))
    sell_idxs = sorted(list(bd_set - ac_set))
    # Ensure indices within bounds
    buy_idxs = [i for i in buy_idxs if 0 <= i < len(arr)]
    sell_idxs = [i for i in sell_idxs if 0 <= i < len(arr)]
    signal[buy_idxs] = 1.0
    signal[sell_idxs] = -1.0
    return signal

def execute_trades(real_prices, signal, initial_money=10000.0, max_buy=1, max_sell=1):
    """
    real_prices: actual execution prices (np array)
    signal: signals array (1 buy, -1 sell, 0 hold) computed over forecast series length
    returns buys, sells indices lists, final_cash, total_gains, percent_return, cash_history
    """
    rp = np.asarray(real_prices, dtype=float)
    sig = np.asarray(signal, dtype=float)
    if len(rp) != len(sig):
        raise ValueError("real_prices and signal must have same length.")
    cash = float(initial_money)
    inventory = []
    buys = []
    sells = []
    cash_history = []
    for i in range(len(rp)):
        s = sig[i]
        price = rp[i]
        if s == 1.0:
            if price <= 0:
                # skip invalid prices
                cash_history.append(cash)
                continue
            affordable = int(cash // price)
            if affordable < 1:
                # cannot buy
                cash_history.append(cash)
                continue
            buy_units = min(max_buy, affordable)
            cash -= buy_units * price
            for _ in range(buy_units):
                inventory.append(price)
            buys.append(i)
            print(f"day {i}: BUY {buy_units} units at {price:.4f}, cash left {cash:.4f}")
        elif s == -1.0:  # sell up to max_sell units (FIFO)
            if len(inventory) == 0:

                cash_history.append(cash)

                continue
            sell_units = min(max_sell, len(inventory))
            # execute sell at current price
            for _ in range(sell_units):
                bought_price = inventory.pop(0)
                # profit can be computed if desired
            cash += sell_units * price
            sells.append(i)
            try:
                last_buy_idx = buys[-1] if len(buys) else None
                invest_pct = ((price - (bought_price if 'bought_price' in locals() else price)) / (bought_price if 'bought_price' in locals() else price)) * 100
            except:
                invest_pct = 0.0
            print(f"day {i}: SELL {sell_units} units at {price:.4f}, cash now {cash:.4f} (recent trade gain% ~= {invest_pct:.3f})")
        cash_history.append(cash)
    if len(inventory) > 0:
        last_price = rp[-1]
        while inventory:
            inventory.pop(0)
            cash += last_price
    total_gains = cash - initial_money
    pct_return = (total_gains / initial_money) * 100.0
    return buys, sells, cash, total_gains, pct_return, cash_history

signal = abcd_signal_from_series(forecast_close, skip_loop=4, ma_window=7)
print("Signals generated on forecasted closes:", np.unique(signal, return_counts=True))

INITIAL_MONEY = 10000.0
MAX_BUY = 1
MAX_SELL = 1

buys, sells, final_cash, total_gains, pct_return, cash_hist = execute_trades(
    real_prices=actual_close,
    signal=signal,
    initial_money=INITIAL_MONEY,
    max_buy=MAX_BUY,
    max_sell=MAX_SELL
)

print("\n=== AB=CD Strategy Summary ===")
print(f"Trades -> buys: {len(buys)} sells: {len(sells)}")
print(f"Final cash: {final_cash:.2f} | Total gains: {total_gains:.2f} | Return: {pct_return:.2f}%")

if len(actual_close) > 0:
    start_price = actual_close[0]
    if start_price > 0:
        units = int(INITIAL_MONEY // start_price)
        leftover = INITIAL_MONEY - units * start_price
        final_bh = leftover + units * actual_close[-1]
        bh_return_pct = ((final_bh - INITIAL_MONEY) / INITIAL_MONEY) * 100.0
    else:
        final_bh = INITIAL_MONEY
        bh_return_pct = 0.0
else:
    final_bh = INITIAL_MONEY
    bh_return_pct = 0.0

print(f"Buy-and-hold baseline -> final cash: {final_bh:.2f} | return: {bh_return_pct:.2f}%")

plt.figure(figsize=(14,5))
plt.plot(test_dates, actual_close, label='Actual Close (execution)', lw=2)
plt.plot(test_dates, forecast_close, label='Forecast Close (agent view)', lw=1, linestyle='--')
if len(buys):
    plt.scatter([test_dates[i] for i in buys], actual_close[buys], marker='^', color='g', s=80, label='Buy')
if len(sells):
    plt.scatter([test_dates[i] for i in sells], actual_close[sells], marker='v', color='r', s=80, label='Sell')
plt.title(f"AB=CD trades on test (return {pct_return:.2f}%, B&H {bh_return_pct:.2f}%)")
plt.legend(); plt.xticks(rotation=45); plt.show()

plt.figure(figsize=(12,4))
plt.plot(test_dates, cash_hist, label='Cash over time')
plt.title("Cash history while trading")
plt.legend(); plt.xticks(rotation=45); plt.show()
```

```python
#ES agent
import numpy as np
import pandas as pd
import time
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()



if 'df' not in globals():
    raise RuntimeError("df not found. Run the model/training/evaluation block producing df first.")
if 'preds_inv' not in globals() or 'targets_inv' not in globals():
    raise RuntimeError("preds_inv or targets_inv not found. Run forecasting block first to produce preds_inv and targets_inv.")

actual_close_full = df['Close'].values

if 'TEST_DAYS' in globals():
    test_days = TEST_DAYS
else:
    test_days = len(preds_inv)  # fallback

if test_days <= 0 or test_days > len(actual_close_full):
    raise RuntimeError(f"TEST_DAYS={test_days} seems invalid for df length {len(actual_close_full)}")

train_actual_series = actual_close_full[:-test_days]
test_actual_series  = actual_close_full[-test_days:]
forecast_close = np.asarray(preds_inv[:, 1])
# ensure shapes
assert len(test_actual_series) == len(forecast_close) == len(preds_inv), "Mismatch between forecast and test actual lengths"

print(f"Train length (for ES): {len(train_actual_series)}, Test length (forecast window): {len(test_actual_series)}")

class Deep_Evolution_Strategy:
    def __init__(self, weights, reward_function, population_size=15, sigma=0.1, learning_rate=0.03):
        self.weights = weights
        self.reward_function = reward_function
        self.population_size = population_size
        self.sigma = sigma
        self.learning_rate = learning_rate

    def _get_weight_from_population(self, weights, population_member):
        weights_population = []
        for w, noise in zip(weights, population_member):
            weights_population.append(w + self.sigma * noise)
        return weights_population

    def get_weights(self):
        return self.weights

    def train(self, epoch=100, print_every=1):
        lasttime = time.time()
        for i in range(epoch):
            population = []
            rewards = np.zeros(self.population_size)
            # sample population noises
            for k in range(self.population_size):
                member = []
                for w in self.weights:
                    member.append(np.random.randn(*w.shape))
                population.append(member)
            # evaluate population
            for k in range(self.population_size):
                weights_k = self._get_weight_from_population(self.weights, population[k])
                rewards[k] = self.reward_function(weights_k)
            # normalize rewards
            rewards = (rewards - np.mean(rewards)) / (np.std(rewards) + 1e-7)
            # update weights
            for idx, w in enumerate(self.weights):
                A = np.array([p[idx] for p in population])
                grad = np.tensordot(rewards, A, axes=(0,0)) / (self.population_size * self.sigma)
                self.weights[idx] = w + self.learning_rate * grad
            if (i + 1) % print_every == 0:
                score = self.reward_function(self.weights)
                print(f'iter {i+1:4d}. reward: {score:8.4f}')
        print('time taken to train:', time.time() - lasttime, 'seconds')

class Model:
    def __init__(self, input_size, hidden_size, output_size):
        self.weights = [
            np.random.randn(input_size, hidden_size) * 0.1,
            np.random.randn(hidden_size, output_size) * 0.1,
            np.random.randn(1, hidden_size) * 0.1
        ]

    def predict(self, inputs):
        W1, W2, b1 = self.weights
        h = np.dot(inputs, W1) + b1
        h = np.maximum(h, 0)  # ReLU
        out = np.dot(h, W2)
        return out

    def get_weights(self):
        return self.weights

    def set_weights(self, weights):
        self.weights = weights

class Agent:
    POPULATION_SIZE = 15
    SIGMA = 0.1
    LEARNING_RATE = 0.03

    def __init__(self, model, window_size, forecast_series, actual_series, skip=1, initial_money=10000.0,
                 pop_size=None, sigma=None, lr=None):
        self.model = model
        self.window_size = window_size
        self.forecast = np.asarray(forecast_series)
        self.actual = np.asarray(actual_series)
        self.skip = skip
        self.initial_money = float(initial_money)
        if pop_size is not None: self.POPULATION_SIZE = pop_size
        if sigma is not None: self.SIGMA = sigma
        if lr is not None: self.LEARNING_RATE = lr

        # ES object
        self.es = Deep_Evolution_Strategy(self.model.get_weights(), self.get_reward,
                                         population_size=self.POPULATION_SIZE,
                                         sigma=self.SIGMA,
                                         learning_rate=self.LEARNING_RATE)

    def act(self, sequence):
        seq = np.array(sequence).reshape(1, -1)
        decision = self.model.predict(seq)
        return int(np.argmax(decision[0]))

    def get_state(self, t):
        window = self.window_size
        d = t - window + 1
        if d >= 0:
            block = self.forecast[d : t + 1]
        else:
            pad = np.full(-d, self.forecast[0])
            block = np.concatenate([pad, self.forecast[0 : t + 1]])
        # use deltas of window values (window-1)
        res = [(block[i+1] - block[i]) for i in range(window - 1)]
        return np.array([res], dtype=np.float32)

    def get_reward(self, weights):
        self.model.set_weights(weights)
        money = self.initial_money
        inventory = []
        L = len(self.forecast)
        state = self.get_state(0)
        for t in range(0, L - 1, self.skip):
            action = self.act(state)
            price_exec = float(self.actual[t])  # execute at real close
            # BUY
            if action == 1 and money >= price_exec:
                inventory.append(price_exec)
                money -= price_exec
            # SELL
            elif action == 2 and len(inventory) > 0:
                _b = inventory.pop(0)
                money += price_exec
            next_t = min(t + 1, L - 1)
            state = self.get_state(next_t)
        # liquidate remaining inventory at last price
        while len(inventory) > 0:
            _b = inventory.pop(0)
            money += float(self.actual[-1])
        return ((money - self.initial_money) / self.initial_money) * 100.0

    def fit(self, iterations=200, checkpoint=20):
        self.es.train(iterations, print_every=checkpoint)

    def trade(self):
        weights = self.es.get_weights()
        self.model.set_weights(weights)
        money = self.initial_money
        inventory = []
        states_buy = []
        states_sell = []
        L = len(self.forecast)
        state = self.get_state(0)
        for t in range(0, L - 1, self.skip):
            action = self.act(state)
            price_exec = float(self.actual[t])
            if action == 1 and money >= price_exec:
                inventory.append(price_exec)
                money -= price_exec
                states_buy.append(t)
            elif action == 2 and len(inventory) > 0:
                bought_price = inventory.pop(0)
                money += price_exec
                states_sell.append(t)
            next_t = min(t + 1, L - 1)
            state = self.get_state(next_t)
        # liquidate remaining inventory at last actual price
        while len(inventory) > 0:
            _b = inventory.pop(0)
            money += float(self.actual[-1])
        total_gains = money - self.initial_money
        invest_pct = (total_gains / self.initial_money) * 100.0
        return states_buy, states_sell, total_gains, invest_pct

window_size = 30
if window_size >= len(train_actual_series):
    window_size = max(5, len(train_actual_series) // 2)

input_size = window_size - 1
hidden_size = 64
output_size = 3

print(f"ES Training on actual historical closes (length {len(train_actual_series)}), window_size={window_size}")

model_train = Model(input_size=input_size, hidden_size=hidden_size, output_size=output_size)
agent_train = Agent(model_train, window_size=window_size,
                    forecast_series=train_actual_series,    # observations = actual closes during train
                    actual_series=train_actual_series,      # execution using actual closes
                    skip=1, initial_money=10000.0,
                    pop_size=15, sigma=0.1, lr=0.03)

# training iterations (adjust as required)
train_iterations = 200
start_time = time.time()
agent_train.fit(iterations=train_iterations, checkpoint=max(10, train_iterations//10))
print("Training finished in {:.1f}s".format(time.time() - start_time))

# obtain learned weights
trained_weights = agent_train.es.get_weights()

model_deploy = Model(input_size=input_size, hidden_size=hidden_size, output_size=output_size)
model_deploy.set_weights(trained_weights)

agent_deploy = Agent(model_deploy, window_size=window_size,
                     forecast_series=forecast_close,
                     actual_series=test_actual_series,
                     skip=1, initial_money=10000.0,
                     pop_size=15, sigma=0.1, lr=0.03)

states_buy, states_sell, total_gains, invest_pct = agent_deploy.trade()

print("\n--- Deployment (on forecasted observations, executed on real actual closes) ---")
print(f"Test window length: {len(forecast_close)}")
print(f"Total gains: {total_gains:.2f}, Return: {invest_pct:.2f}%")
print(f"Number of buys: {len(states_buy)}, Number of sells: {len(states_sell)}")

if 'test_dates' in globals():
    dates = list(test_dates[:len(forecast_close)])
else:
    # fallback: use df index for last TEST_DAYS
    dates = list(df.index[-test_days:])

action_df = pd.DataFrame({
    "date": dates,
    "pred_close": forecast_close,
    "actual_close": test_actual_series
})
action_df["buy"] = False
action_df["sell"] = False
action_df.loc[states_buy, "buy"] = True
action_df.loc[states_sell, "sell"] = True
action_df.set_index("date", inplace=True)

print("\nSample action dataframe (head):")
display(action_df.head(20))

plt.figure(figsize=(14,5))
plt.plot(action_df.index, action_df["actual_close"], label="Actual Close", lw=2)
plt.scatter(action_df.index[action_df["buy"]], action_df["actual_close"][action_df["buy"]], marker="^", color="g", s=80, label="Buy")
plt.scatter(action_df.index[action_df["sell"]], action_df["actual_close"][action_df["sell"]], marker="v", color="r", s=80, label="Sell")
plt.title(f"ES agent trades (trained on actual) — Gains: {total_gains:.2f}, Return: {invest_pct:.2f}%")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python

# Neuro-Evolution Agent

import numpy as np
import random
import matplotlib.pyplot as plt
import math

POPULATION_SIZE = 60
GENERATIONS = 60
MUTATION_RATE = 0.12
MUTATION_SCALE = 0.5
WINDOW_SIZE = 30         # state window length (differences)
SKIP = 1                 # step through timeline
INITIAL_MONEY = 10000.0
HIDDEN_SIZE = 128
SEED = 42

np.random.seed(SEED)
random.seed(SEED)

if 'preds_inv' not in globals() or 'targets_inv' not in globals():
    raise RuntimeError("preds_inv and targets_inv not found. Run the forecasting cell before this cell.")

# Determine forecast_close & actual_close
forecast_close = np.asarray(preds_inv[:, 1], dtype=float)
actual_close = np.asarray(targets_inv[:, 1], dtype=float)
if 'test_df' in globals():
    n_preds = len(forecast_close)
    if len(test_df) >= SEQ_LEN + n_preds:
        test_dates = test_df.index[SEQ_LEN: SEQ_LEN + n_preds]
    else:
        test_dates = test_df.index[-n_preds:]
else:
    test_dates = np.arange(len(forecast_close))

# final length checks
if len(forecast_close) != len(actual_close):
    raise RuntimeError(f"Length mismatch: forecast_close ({len(forecast_close)}) vs actual_close ({len(actual_close)}).")

if len(forecast_close) < WINDOW_SIZE + 2:
    raise RuntimeError(f"Not enough test steps ({len(forecast_close)}) for WINDOW_SIZE={WINDOW_SIZE}. Reduce WINDOW_SIZE or provide more test days.")

def relu(x): return np.maximum(0, x)
def softmax(x):
    e = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e / (np.sum(e, axis=-1, keepdims=True) + 1e-12)

class NNIndividual:
    def __init__(self, id_, input_size=WINDOW_SIZE, hidden_size=HIDDEN_SIZE, output_size=3):
        self.W1 = np.random.randn(input_size, hidden_size) / np.sqrt(input_size)
        self.b1 = np.zeros((1, hidden_size))
        self.W2 = np.random.randn(hidden_size, output_size) / np.sqrt(hidden_size)
        self.b2 = np.zeros((1, output_size))
        self.fitness = -1e9
        self.id = id_
    def predict(self, X):
        a1 = X.dot(self.W1) + self.b1
        z1 = relu(a1)
        a2 = z1.dot(self.W2) + self.b2
        return softmax(a2)

class NeuroEvolution:
    def __init__(self, population_size, mutation_rate, mutation_scale,
                 model_generator, window_size, train_series, skip=1, initial_money=10000.0):
        self.population_size = population_size
        self.mutation_rate = mutation_rate
        self.mutation_scale = mutation_scale
        self.model_generator = model_generator
        self.window_size = window_size
        self.train_series = np.asarray(train_series, dtype=float)
        self.skip = skip
        self.initial_money = initial_money
        self.population = []
    def _initialize_population(self):
        self.population = [self.model_generator(i) for i in range(self.population_size)]
    def mutate(self, individual, scale=None):
        if scale is None: scale = self.mutation_scale
        mask = np.random.binomial(1, self.mutation_rate, individual.W1.shape)
        individual.W1 += np.random.normal(0, scale, size=individual.W1.shape) * mask
        mask = np.random.binomial(1, self.mutation_rate, individual.b1.shape)
        individual.b1 += np.random.normal(0, scale, size=individual.b1.shape) * mask
        mask = np.random.binomial(1, self.mutation_rate, individual.W2.shape)
        individual.W2 += np.random.normal(0, scale, size=individual.W2.shape) * mask
        mask = np.random.binomial(1, self.mutation_rate, individual.b2.shape)
        individual.b2 += np.random.normal(0, scale, size=individual.b2.shape) * mask
        return individual
    def crossover(self, p1, p2):
        child = self.model_generator((p1.id + p2.id) % 100000)
        h = child.W1.shape[1]
        cut = np.random.randint(1, h)
        child.W1[:, :cut] = p1.W1[:, :cut].copy()
        child.W1[:, cut:] = p2.W1[:, cut:].copy()
        child.b1[:, :cut] = p1.b1[:, :cut].copy()
        child.b1[:, cut:] = p2.b1[:, cut:].copy()
        child.W2[:cut, :] = p1.W2[:cut, :].copy()
        child.W2[cut:, :] = p2.W2[cut:, :].copy()
        child.b2 = (p1.b2 + p2.b2) / 2.0
        return child
    def get_state(self, series, t):
        wplus1 = self.window_size + 1
        d = t - wplus1 + 1
        if d >= 0:
            block = series[d : t + 1]
        else:
            pad = np.repeat(series[0], -d)
            block = np.concatenate([pad, series[0 : t+1]])
        res = []
        for i in range(wplus1 - 1):
            res.append(block[i+1] - block[i])
        return np.array([res], dtype=float)
    def act(self, individual, state):
        probs = individual.predict(state)
        return int(np.argmax(probs[0]))  # 0 hold,1 buy,2 sell
    def evaluate_on_series(self, individual, series_forecast, series_actual):
        cash = float(self.initial_money)
        inventory = []
        n = len(series_forecast)
        state = self.get_state(series_forecast, 0)
        for t in range(0, n-1, self.skip):
            action = self.act(individual, state)
            price_exec = float(series_actual[t])
            if action == 1 and cash >= price_exec:  # BUY one unit
                inventory.append(price_exec)
                cash -= price_exec
            elif action == 2 and len(inventory):    # SELL FIFO one unit
                inventory.pop(0)
                cash += price_exec
            if t+1 < n:
                state = self.get_state(series_forecast, t+1)
        if len(inventory) > 0:
            last_price = float(series_actual[-1])
            while inventory:
                inventory.pop(0)
                cash += last_price
        profit_pct = ((cash - self.initial_money) / self.initial_money) * 100.0
        return profit_pct, cash
    def calculate_population_fitness(self):
        for ind in self.population:
            p, _ = self.evaluate_on_series(ind, self.train_series, self.train_series)
            ind.fitness = p
    def evolve(self, generations=20, checkpoint=5):
        self._initialize_population()
        for gen in range(1, generations+1):
            self.calculate_population_fitness()
            self.population.sort(key=lambda x: x.fitness, reverse=True)
            best = self.population[0]
            if gen % checkpoint == 0 or gen == 1:
                print(f"Gen {gen:03d} | Best fitness (train %) = {best.fitness:.4f}")
            n_elite = max(1, int(0.30 * self.population_size))
            next_pop = [self.population[i] for i in range(n_elite)]
            fitness_vals = np.array([max(0.0, p.fitness) for p in self.population]) + 1e-8
            probs = fitness_vals / fitness_vals.sum()
            while len(next_pop) < self.population_size:
                parents = np.random.choice(self.population, size=2, p=probs, replace=False)
                child = self.crossover(parents[0], parents[1])
                child = self.mutate(child, scale=self.mutation_scale)
                next_pop.append(child)
            self.population = next_pop
        self.population.sort(key=lambda x: x.fitness, reverse=True)
        return self.population[0]

if 'train_df' not in globals():
    raise RuntimeError("train_df not found. Run your training code that created train_df before this cell.")
train_close = train_df['Close'].values.astype(float)

def model_gen(i):
    return NNIndividual(id_=i, input_size=WINDOW_SIZE, hidden_size=HIDDEN_SIZE)

ne = NeuroEvolution(
    population_size=POPULATION_SIZE,
    mutation_rate=MUTATION_RATE,
    mutation_scale=MUTATION_SCALE,
    model_generator=model_gen,
    window_size=WINDOW_SIZE,
    train_series=train_close,
    skip=SKIP,
    initial_money=INITIAL_MONEY
)

print("Evolving on historical actual Close prices (training)...")
best_ind = ne.evolve(generations=GENERATIONS, checkpoint=5)
print("Evolution finished. Best training fitness (%):", best_ind.fitness)

print("\nEvaluating best evolved agent on test (decisions from forecasted closes, executed at actual closes)...")
test_profit_pct, test_cash = ne.evaluate_on_series(best_ind, forecast_close, actual_close)
print(f"Test trading -> final cash: {test_cash:.2f} | return: {test_profit_pct:.2f}% (initial {INITIAL_MONEY})")

# collect trades for plotting
def simulate_and_collect(ind, series_forecast, series_actual):
    cash = float(INITIAL_MONEY)
    inventory = []
    buys, sells = [], []
    n = len(series_forecast)
    state = ne.get_state(series_forecast, 0)
    for t in range(0, n-1, SKIP):
        action = ne.act(ind, state)
        price_exec = float(series_actual[t])
        if action == 1 and cash >= price_exec:
            inventory.append(price_exec); cash -= price_exec; buys.append(t)
        elif action == 2 and len(inventory):
            inventory.pop(0); cash += price_exec; sells.append(t)
        if t+1 < n:
            state = ne.get_state(series_forecast, t+1)
    if len(inventory) > 0:
        last_price = float(series_actual[-1])
        while inventory:
            inventory.pop(0); cash += last_price
    return buys, sells, cash

states_buy, states_sell, final_cash = simulate_and_collect(best_ind, forecast_close, actual_close)
total_gain = final_cash - INITIAL_MONEY
total_gain_pct = (total_gain / INITIAL_MONEY) * 100.0
print(f"Trades executed: buys={len(states_buy)} sells={len(states_sell)} final_cash={final_cash:.2f} gain={total_gain:.2f} ({total_gain_pct:.2f}%)")

# baseline buy-and-hold (1 unit) on test actual closes
if len(actual_close) > 0:
    start_price = actual_close[0]
    if INITIAL_MONEY >= start_price:
        units = int(INITIAL_MONEY // start_price)
        leftover = INITIAL_MONEY - units * start_price
        final_bah = leftover + units * actual_close[-1]
        bah_return_pct = ((final_bah - INITIAL_MONEY) / INITIAL_MONEY) * 100.0
    else:
        final_bah = INITIAL_MONEY; bah_return_pct = 0.0
else:
    final_bah = INITIAL_MONEY; bah_return_pct = 0.0

print(f"Buy-and-hold baseline -> final cash: {final_bah:.2f} | return: {bah_return_pct:.2f}%")

plt.figure(figsize=(14,5))
plt.plot(test_dates, actual_close, label='Actual Close (execution)', lw=2)
plt.plot(test_dates, forecast_close, label='Forecast Close (agent view)', lw=1, linestyle='--')
if len(states_buy)>0:
    plt.scatter(test_dates[states_buy], actual_close[states_buy], marker='^', color='g', s=90, label='Buy')
if len(states_sell)>0:
    plt.scatter(test_dates[states_sell], actual_close[states_sell], marker='v', color='r', s=90, label='Sell')
plt.title(f"Neuro-Evolution trades on test (gain={total_gain_pct:.2f}%, B&H={bah_return_pct:.2f}%)")
plt.legend(); plt.xticks(rotation=45)
plt.show()

best_weights = {"W1": best_ind.W1, "b1": best_ind.b1, "W2": best_ind.W2, "b2": best_ind.b2}
print("Best agent weights stored in dict `best_weights`.")
```

```python
import os
import random
import math
from datetime import timedelta
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Seeds and Device
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")


# Hyperparameters (enhanced)
CSV_PATH = "reliance_with_indicators.csv"
START_DATE = "2019-01-01"
SEQ_LEN = 60
BATCH_SIZE = 64
LR_GEN = 5e-4
LR_DISC = 5e-4
LR_REFINER = 1e-4
EPOCHS_GAN = 60
EPOCHS_REFINER = 50
TEST_DAYS = 30
HIDDEN_SIZE_GEN = 256
HIDDEN_SIZE_DISC = 256
HIDDEN_SIZE_REFINER = 256
NOISE_DIM = 32
DROPOUT = 0.3
GRAD_CLIP = 1.0


df = pd.read_csv(CSV_PATH, index_col=0, parse_dates=True)
df.sort_index(inplace=True)
df_timezone = df.index.tz
start_date_aware = pd.to_datetime(START_DATE).tz_localize(df_timezone)
df = df[df.index >= start_date_aware].copy()
df = df.ffill().bfill()
df.dropna(inplace=True)

feature_cols = df.columns.tolist()
target_cols = ["Open", "Close"]

train_val_df = df.iloc[:-TEST_DAYS]
test_df = df.iloc[-TEST_DAYS:].copy()

val_frac = 0.1
val_size = int(len(train_val_df) * val_frac)
train_df = train_val_df.iloc[:-val_size].copy()
val_df = train_val_df.iloc[-val_size:].copy()

scaler = MinMaxScaler(feature_range=(0, 1))
scaler.fit(train_df[feature_cols].values)

train_scaled = pd.DataFrame(scaler.transform(train_df[feature_cols]), index=train_df.index, columns=feature_cols)
val_scaled = pd.DataFrame(scaler.transform(val_df[feature_cols]), index=val_df.index, columns=feature_cols)
test_scaled = pd.DataFrame(scaler.transform(test_df[feature_cols]), index=test_df.index, columns=feature_cols)


def inverse_targets(scaled_arr, feature_cols=feature_cols, target_cols=target_cols, scaler=scaler):
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(feature_cols)))
    for i, col in enumerate(feature_cols):
        if col in target_cols:
            idx = target_cols.index(col)
            filler[:, feature_cols.index(col)] = scaled_arr[:, idx]
    inv = scaler.inverse_transform(filler)
    return inv[:, [feature_cols.index(col) for col in target_cols]]


class TimeSeriesDataset(Dataset):
    def __init__(self, df_scaled, seq_len=SEQ_LEN):
        self.X = df_scaled.values.astype(np.float32)
        self.seq_len = seq_len
        self.indices = list(range(seq_len-1, len(self.X)-1))
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, idx):
        i = self.indices[idx]
        seq = self.X[i-(self.seq_len-1):i+1]
        target_row = self.X[i+1]
        target = np.array([target_row[feature_cols.index(c)] for c in target_cols], dtype=np.float32)
        return torch.from_numpy(seq), torch.from_numpy(target)


train_ds = TimeSeriesDataset(train_scaled)
val_ds = TimeSeriesDataset(val_scaled)
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, drop_last=False)


class Generator(nn.Module):
    def __init__(self, seq_len, input_size, noise_dim, output_size):
        super().__init__()
        self.seq_len = seq_len
        self.lstm = nn.LSTM(input_size + noise_dim, HIDDEN_SIZE_GEN, num_layers=3, batch_first=True, dropout=DROPOUT)
        self.bn = nn.BatchNorm1d(HIDDEN_SIZE_GEN)
        self.fc = nn.Sequential(nn.Linear(HIDDEN_SIZE_GEN, 128), nn.ReLU(), nn.Linear(128, output_size))
    def forward(self, seq, noise):
        noise_exp = noise.unsqueeze(1).repeat(1, seq.size(1), 1)
        x = torch.cat([seq, noise_exp], dim=-1)
        out, _ = self.lstm(x)
        # Batch norm requires (batch, features, seq_length), so transpose
        out_bn = self.bn(out.permute(0, 2, 1)).permute(0, 2, 1)
        out = out_bn[:, -1, :]
        return self.fc(out)


class Discriminator(nn.Module):
    def __init__(self, seq_len, input_size, output_size):
        super().__init__()
        self.lstm = nn.LSTM(input_size + output_size, HIDDEN_SIZE_DISC, num_layers=3, batch_first=True, dropout=DROPOUT)
        self.fc = nn.Sequential(nn.Linear(HIDDEN_SIZE_DISC, 128), nn.LeakyReLU(0.2), nn.Linear(128, 1), nn.Sigmoid())
    def forward(self, seq, next_step):
        next_step_exp = next_step.unsqueeze(1).repeat(1, seq.shape[1], 1)
        x = torch.cat([seq, next_step_exp], dim=-1)
        out, _ = self.lstm(x)
        out = out[:, -1, :]
        return self.fc(out)


gen = Generator(SEQ_LEN, input_size=len(feature_cols), noise_dim=NOISE_DIM, output_size=len(target_cols)).to(DEVICE)
disc = Discriminator(SEQ_LEN, input_size=len(feature_cols), output_size=len(target_cols)).to(DEVICE)

opt_gen = torch.optim.Adam(gen.parameters(), lr=LR_GEN)
opt_disc = torch.optim.Adam(disc.parameters(), lr=LR_DISC)
criterion_gan = nn.BCELoss()

# Label smoothing and noisy labels
def smooth_labels(y, smooth=0.1):
    return y * (1 - smooth) + smooth * 0.5

# Training GAN with improvements
for epoch in range(EPOCHS_GAN):
    gen.train(); disc.train()
    losses_g, losses_d = [], []
    for xb, yb in train_loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        batch_size = xb.size(0)

        noise = torch.randn(batch_size, NOISE_DIM).to(DEVICE)
        fake_next = gen(xb, noise)

        real_labels = smooth_labels(torch.ones(batch_size,1).to(DEVICE), smooth=0.1)
        fake_labels = smooth_labels(torch.zeros(batch_size,1).to(DEVICE), smooth=0.1)

        disc_real = disc(xb, yb)
        disc_fake = disc(xb, fake_next.detach())

        loss_d = criterion_gan(disc_real, real_labels) + criterion_gan(disc_fake, fake_labels)
        opt_disc.zero_grad()
        loss_d.backward()
        nn.utils.clip_grad_norm_(disc.parameters(), GRAD_CLIP)
        opt_disc.step()
        losses_d.append(loss_d.item())

        disc_fake = disc(xb, fake_next)
        loss_g = criterion_gan(disc_fake, real_labels)
        opt_gen.zero_grad()
        loss_g.backward()
        nn.utils.clip_grad_norm_(gen.parameters(), GRAD_CLIP)
        opt_gen.step()
        losses_g.append(loss_g.item())

    print(f"Epoch {epoch+1} | Loss D: {np.mean(losses_d):.6f} | Loss G: {np.mean(losses_g):.6f}")

# Refiner: larger model, finer tuning
class LSTMRefiner(nn.Module):
    def __init__(self, input_size, hidden_size=HIDDEN_SIZE_REFINER, output_size=len(target_cols)):
        super().__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers=3, batch_first=True, dropout=DROPOUT)
        self.fc = nn.Sequential(nn.Linear(hidden_size, hidden_size//2), nn.ReLU(), nn.Linear(hidden_size//2, output_size))
    def forward(self, seq):
        out, _ = self.lstm(seq)
        out = out[:, -1, :]
        return self.fc(out)

lstm_input_size = len(feature_cols) + len(target_cols)
lstm_refiner = LSTMRefiner(lstm_input_size).to(DEVICE)
opt_refiner = torch.optim.Adam(lstm_refiner.parameters(), lr=LR_REFINER)
criterion_refiner = nn.MSELoss()

# Refiner Dataset definition unchanged, reuse yours with generator gen

class RefinerDataset(Dataset):
    def __init__(self, df_scaled, generator, seq_len=SEQ_LEN):
        self.seq_len = seq_len
        self.df = df_scaled.values.astype(np.float32)
        self.generator = generator
        self.indices = [i for i in range(seq_len-1, len(self.df)-1)]
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, idx):
        i = self.indices[idx]
        seq = self.df[i-(self.seq_len-1):i+1]
        noise = torch.randn(1, NOISE_DIM).to(DEVICE)
        seq_t = torch.from_numpy(seq).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            gen_next = self.generator(seq_t, noise).cpu().numpy().reshape(-1)
        lstm_input = np.hstack([seq[-1], gen_next]).astype(np.float32)
        target_row = self.df[i+1]
        target = np.array([target_row[feature_cols.index(c)] for c in target_cols], dtype=np.float32)
        return torch.from_numpy(lstm_input), torch.from_numpy(target)

refiner_ds = RefinerDataset(train_scaled, gen)
refiner_loader = DataLoader(refiner_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)

for epoch in range(EPOCHS_REFINER):
    lstm_refiner.train()
    losses = []
    for xb, yb in refiner_loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        xb = xb.unsqueeze(1)  # seq_len=1 for LSTM input here
        pred = lstm_refiner(xb)
        loss = criterion_refiner(pred, yb)
        opt_refiner.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(lstm_refiner.parameters(), GRAD_CLIP)
        opt_refiner.step()
        losses.append(loss.item())
    print(f"Refiner Epoch {epoch+1} | Loss: {np.mean(losses):.6f}")

# Hybrid forecasting on test set
full_scaled = pd.DataFrame(scaler.transform(df[feature_cols].values), index=df.index, columns=feature_cols)
rolling_seq = full_scaled.iloc[-SEQ_LEN:].values.astype(np.float32)
hybrid_preds = []
iter_dates = []

for step in range(TEST_DAYS):
    seq_tensor = torch.from_numpy(rolling_seq[np.newaxis, :, :]).to(DEVICE)
    noise = torch.randn(1, NOISE_DIM).to(DEVICE)
    with torch.no_grad():
        gen_next = gen(seq_tensor, noise).cpu().numpy().reshape(-1)
    lstm_input = np.hstack([rolling_seq[-1], gen_next]).astype(np.float32)
    lstm_input_tensor = torch.from_numpy(lstm_input).unsqueeze(0).unsqueeze(1).to(DEVICE)
    with torch.no_grad():
        refined_next = lstm_refiner(lstm_input_tensor).cpu().numpy().reshape(-1)
    hybrid_preds.append(refined_next.copy())
    # Roll the window
    last_row = rolling_seq[-1].copy()
    for j, col in enumerate(feature_cols):
        if col in target_cols:
            last_row[feature_cols.index(col)] = refined_next[target_cols.index(col)]
    rolling_seq = np.vstack([rolling_seq[1:], last_row])
    last_date = df.index[-SEQ_LEN + step]
    iter_dates.append(last_date + timedelta(days=1))

hybrid_preds_inv = inverse_targets(np.stack(hybrid_preds))

# Plot forecasts
plt.figure(figsize=(12,5))
plt.plot(iter_dates, hybrid_preds_inv[:,0], label="Forecast Open")
plt.plot(iter_dates, hybrid_preds_inv[:,1], label="Forecast Close")
plt.title("Hybrid GAN + LSTM Forecast (Improved)")
plt.xticks(rotation=45)
plt.legend()
plt.show()

# Show forecast DataFrame
pred_df = pd.DataFrame(hybrid_preds_inv, columns=["Pred_Open", "Pred_Close"], index=iter_dates)
print("Improved Hybrid 30-day forecast (Open & Close):")
display(pred_df)

# Save improved models and scaler
torch.save({
    "generator_state_dict": gen.state_dict(),
    "discriminator_state_dict": disc.state_dict(),
    "lstm_refiner_state_dict": lstm_refiner.state_dict(),
    "scaler": scaler,
    "feature_cols": feature_cols,
    "target_cols": target_cols
}, "improved_hybrid_gan_lstm_reliance.pth")

print("Done. Improved Hybrid GAN + LSTM model and forecasts saved.")
```

```python
# ---------------------------
# XAI: IG, Saliency, Permutation & SHAP
# Run this immediately after your forecasting / plotting code above
# ---------------------------
import os
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
import torch
from captum.attr import IntegratedGradients, Saliency
import shap
from sklearn.metrics import mean_squared_error
from tqdm import tqdm

print("Starting XAI analysis...")

# ---------------------------
# Helpers & robust variable resolution
# ---------------------------
# Try to get FEATURE_COLS / TARGET_COLS from either uppercase or lowercase variable names used in your scripts
FEATURE_COLS = globals().get("FEATURE_COLS", globals().get("feature_cols", None))
TARGET_COLS  = globals().get("TARGET_COLS", globals().get("target_cols", None))
# Ensure target_cols used in your forecasting snippet is captured too
if TARGET_COLS is None:
    TARGET_COLS = globals().get("target_cols", None)
if FEATURE_COLS is None or TARGET_COLS is None:
    raise RuntimeError("FEATURE_COLS and TARGET_COLS not found in environment. Make sure they exist before running XAI code.")

# Model / scaler retrieval (assume 'model' and 'scaler' are in workspace - if not, try common filenames)
if "model" not in globals():
    # try to load one of saved checkpoints if available
    ck_candidates = ["tcn_attention_reliance_final.pth","convtransformer_reliance_final.pth",
                     "informer_like_reliance_final.pth","informer_like_reliance.pth","convtransformer_reliance.pth"]
    loaded = False
    for ck in ck_candidates:
        if os.path.exists(ck):
            print(f"Loading checkpoint {ck} ...")
            ckpt = torch.load(ck, map_location="cuda" if torch.cuda.is_available() else "cpu")
            # expect ckpt contains 'model_state_dict' and 'scaler' and 'feature_cols'/ 'target_cols'
            if "model_state_dict" in ckpt:
                # can't auto-instantiate model architecture generically; prefer your script to have model object.
                print(f"Found checkpoint {ck} but no `model` object in namespace. Please re-create the model in this session before running XAI.")
            if "scaler" in ckpt and "feature_cols" in ckpt:
                scaler = ckpt["scaler"]
            loaded = True
            break
    if not loaded:
        print("No external checkpoint loaded. Expecting `model` and `scaler` to already be present in workspace.")
else:
    print("Model found in workspace.")

if "scaler" not in globals():
    raise RuntimeError("`scaler` not found in workspace. It's required to reconstruct scaled sequences. Load or set scaler variable before running this XAI script.")

# required variables: df, SEQ_LEN, TEST_DAYS
if "df" not in globals():
    raise RuntimeError("`df` (original dataframe) not found in workspace.")
if "SEQ_LEN" not in globals():
    raise RuntimeError("`SEQ_LEN` not found in workspace.")
if "TEST_DAYS" not in globals():
    raise RuntimeError("`TEST_DAYS` not found in workspace.")

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(DEVICE)
model.eval()

# ---------------------------
# Reconstruct the test-input sequences used for the final TEST_DAYS (same logic as your forecasting)
# ---------------------------
full_scaled = pd.DataFrame(scaler.transform(df[FEATURE_COLS].values), index=df.index, columns=FEATURE_COLS)
test_inputs = []
test_targets = []
test_dates = []
for i in range(len(df) - TEST_DAYS, len(df)):
    seq_end = i - 1
    seq_start = seq_end - (SEQ_LEN - 1)
    if seq_start < 0:
        continue
    seq = full_scaled.iloc[seq_start: seq_end + 1].values.astype(np.float32)
    target_row = full_scaled.iloc[i].values.astype(np.float32)
    target = np.array([target_row[FEATURE_COLS.index(c)] for c in TARGET_COLS], dtype=np.float32)
    test_inputs.append(seq)
    test_targets.append(target)
    test_dates.append(df.index[i])

test_inputs = np.stack(test_inputs)   # (N_test, seq_len, n_features)
test_targets = np.stack(test_targets)

print(f"Prepared {len(test_inputs)} test sequences for XAI (SEQ_LEN={SEQ_LEN})")

# inverse helper (same as your code)
def inverse_targets(scaled_arr):
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for i, col in enumerate(FEATURE_COLS):
        if col in TARGET_COLS:
            # find index in scaled_arr corresponding to this target (assumes order of TARGET_COLS)
            filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

# prediction helper returns unscaled preds
def predict_unscaled_from_seq_np(X_np):
    model.eval()
    with torch.no_grad():
        x_t = torch.from_numpy(X_np.astype(np.float32)).to(DEVICE)
        preds_scaled, _ = model(x_t)
        preds_scaled = preds_scaled.cpu().numpy()
    return inverse_targets(preds_scaled)

# Quick baseline metric
preds_inv = predict_unscaled_from_seq_np(test_inputs)
targets_inv = inverse_targets(test_targets)
base_rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
print(f"Base RMSE (Close) on reconstructed TEST_DAYS set: {base_rmse_close:.4f}")

# ---------------------------
# 1) Integrated Gradients (local) - a few samples
# ---------------------------
print("\nRunning Integrated Gradients (local) ...")
ig_wrapper_class = None
class WrapperForTarget(torch.nn.Module):
    def __init__(self, model, target_idx=1):
        super().__init__()
        self.model = model
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.model(x)
        return preds[:, self.target_idx]

ig_wrapper = WrapperForTarget(model, target_idx=1).to(DEVICE)
ig = IntegratedGradients(ig_wrapper)

N_SAMPLES_IG = min(6, len(test_inputs))
ig_records = []
# disable cudnn for backward safety if needed
for si in range(N_SAMPLES_IG):
    inp_np = test_inputs[si:si+1]
    inp = torch.from_numpy(inp_np).to(DEVICE).float()
    baseline = torch.zeros_like(inp).to(DEVICE)
    # compute IG inside cudnn-disabled context to avoid RNN double-backward issues (safe for all models)
    with torch.backends.cudnn.flags(enabled=False):
        attributions, delta = ig.attribute(inputs=inp, baselines=baseline, n_steps=50, return_convergence_delta=True)
    at_np = attributions.squeeze(0).cpu().numpy()   # (seq_len, n_features)
    feat_scores = np.sum(np.abs(at_np), axis=0)
    df_ig = pd.DataFrame({"feature": FEATURE_COLS, "ig_score": feat_scores})
    df_ig["ig_pct"] = 100 * df_ig["ig_score"] / (df_ig["ig_score"].sum() + 1e-12)
    df_ig = df_ig.sort_values("ig_score", ascending=False).reset_index(drop=True)
    print(f"\nIG top-10 (sample {si}, date {test_dates[si].date()}):")
    display(df_ig.head(10))
    # heatmap
    plt.figure(figsize=(10,6))
    sns.heatmap(at_np.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"IG heatmap (sample {si} — {test_dates[si].date()})")
    plt.xlabel("time index (older -> newer)")
    plt.tight_layout(); plt.show()
    df_ig["sample_idx"] = si; df_ig["date"] = str(test_dates[si].date())
    ig_records.append(df_ig.head(50))

ig_all = pd.concat(ig_records, ignore_index=True)
ig_all.to_csv("xai_ig_local.csv", index=False)
print("Saved xai_ig_local.csv")

# ---------------------------
# 2) Saliency (Grad × Input) local
# ---------------------------
print("\nRunning Saliency (grad × input) ...")
sal = Saliency(ig_wrapper)
sal_records = []
for si in range(N_SAMPLES_IG):
    inp_np = test_inputs[si:si+1].astype(np.float32)
    inp = torch.from_numpy(inp_np).to(DEVICE)
    inp.requires_grad = True
    with torch.backends.cudnn.flags(enabled=False):
        grads = sal.attribute(inp)   # (1, seq_len, n_features)
    grads_np = grads.squeeze(0).cpu().numpy()
    grad_x_input = grads_np * test_inputs[si]   # elementwise (seq_len, n_features)
    feat_scores = np.sum(np.abs(grad_x_input), axis=0)
    df_sal = pd.DataFrame({"feature": FEATURE_COLS, "saliency_score": feat_scores})
    df_sal["saliency_pct"] = 100 * df_sal["saliency_score"] / (df_sal["saliency_score"].sum() + 1e-12)
    df_sal = df_sal.sort_values("saliency_score", ascending=False).reset_index(drop=True)
    print(f"\nSaliency top-10 (sample {si}, date {test_dates[si].date()}):")
    display(df_sal.head(10))
    plt.figure(figsize=(10,6))
    sns.heatmap(grad_x_input.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"Saliency (grad×input) heatmap (sample {si})")
    plt.tight_layout(); plt.show()
    df_sal["sample_idx"] = si; df_sal["date"] = str(test_dates[si].date())
    sal_records.append(df_sal.head(50))

sal_all = pd.concat(sal_records, ignore_index=True)
sal_all.to_csv("xai_saliency_local.csv", index=False)
print("Saved xai_saliency_local.csv")

# ---------------------------
# 3) Permutation Importance (global) — sequence-level permutation
# ---------------------------
print("\nRunning Permutation Importance (global) ...")
def permutation_importance_seq(X_seq, y_unscaled, features, target_idx=1, n_repeats=3):
    base_preds = predict_unscaled_from_seq_np(X_seq)
    base_rmse = math.sqrt(mean_squared_error(y_unscaled[:, target_idx], base_preds[:, target_idx]))
    print(f"Base RMSE (target index {target_idx}) = {base_rmse:.4f}")
    rows = []
    N = X_seq.shape[0]
    for fi, fname in enumerate(features):
        vals = []
        for r in range(n_repeats):
            Xp = X_seq.copy()
            perm = np.random.permutation(N)
            Xp[:, :, fi] = Xp[perm, :, fi]   # shuffle entire time-series column across samples
            preds_perm = predict_unscaled_from_seq_np(Xp)
            rmse = math.sqrt(mean_squared_error(y_unscaled[:, target_idx], preds_perm[:, target_idx]))
            vals.append(rmse - base_rmse)
        rows.append((fname, np.mean(vals), np.std(vals)))
    dfp = pd.DataFrame(rows, columns=["feature", "delta_rmse_mean", "delta_rmse_std"]).sort_values("delta_rmse_mean", ascending=False)
    return dfp

perm_df = permutation_importance_seq(test_inputs.copy(), targets_inv.copy(), FEATURE_COLS, target_idx=1, n_repeats=3)
display(perm_df.head(12))
perm_df.to_csv("xai_permutation_importance.csv", index=False)
print("Saved xai_permutation_importance.csv")

plt.figure(figsize=(8,6))
sns.barplot(x="delta_rmse_mean", y="feature", data=perm_df.head(12))
plt.title("Permutation importance (increase in RMSE when permuted) - Close")
plt.tight_layout(); plt.show()

# ---------------------------
# 4) SHAP (KernelExplainer) - flatten each sequence to per-feature summary (mean) to keep explainer tractable
# ---------------------------
print("\nRunning SHAP (KernelExplainer) — sequence collapsed to per-feature mean for speed.")
X_flat = test_inputs.mean(axis=1)   # (N, n_features) — you can change to median or last-step as an alternative
y_unscaled = targets_inv.copy()     # unscaled actuals

# background for kernel explainer
BG = min(40, len(X_flat))
BG_IDX = np.random.choice(len(X_flat), BG, replace=False)
background = X_flat[BG_IDX]

def shap_predict_flat(x2d):
    # x2d: (m, n_features) -> convert to sequence by repeating across time dimension
    seq_like = np.repeat(x2d[:, np.newaxis, :], SEQ_LEN, axis=1).astype(np.float32)
    preds = predict_unscaled_from_seq_np(seq_like)
    return preds[:, 1]   # return Close predictions (unscaled)

N_SHAP = min(6, len(X_flat))
NSHAP_SAMPLES = 200  # reduce/increase based on runtime
print("Initializing KernelExplainer (this can be slow)...")
explainer = shap.KernelExplainer(shap_predict_flat, background)
to_explain = X_flat[:N_SHAP]
shap_vals = explainer.shap_values(to_explain, nsamples=NSHAP_SAMPLES)

# shap_values may be list or array depending on shap version
shap_records = []
for i in range(len(to_explain)):
    vals = shap_vals[i] if isinstance(shap_vals, (list,tuple)) else shap_vals[i]
    df_sh = pd.DataFrame({"feature": FEATURE_COLS, "shap": np.array(vals)})
    df_sh["abs_shap"] = df_sh["shap"].abs()
    df_sh = df_sh.sort_values("abs_shap", ascending=False).reset_index(drop=True)
    print(f"\nSHAP top features sample {i} (date {test_dates[i].date()}):")
    display(df_sh.head(10))
    plt.figure(figsize=(8,4))
    sns.barplot(x="shap", y="feature", data=df_sh.head(10))
    plt.title(f"SHAP values (Close) - sample {i}")
    plt.tight_layout(); plt.show()
    df_sh["sample_idx"] = i; df_sh["date"] = str(test_dates[i].date())
    shap_records.append(df_sh.head(50))

shap_all = pd.concat(shap_records, ignore_index=True)
shap_all.to_csv("xai_shap_local.csv", index=False)
print("Saved xai_shap_local.csv")

# ---------------------------
# Aggregation summary
# ---------------------------
print("\nBuilding aggregated summary of XAI outputs...")
top_ig = ig_all.groupby("feature")["ig_pct"].mean().sort_values(ascending=False)
top_sal = sal_all.groupby("feature")["saliency_pct"].mean().sort_values(ascending=False)
top_perm = perm_df.set_index("feature")["delta_rmse_mean"]
top_shap = shap_all.groupby("feature")["abs_shap"].mean().sort_values(ascending=False)

agg = pd.DataFrame({
    "IG_pct": top_ig,
    "Saliency_pct": top_sal,
    "Perm_delta_RMSE": top_perm,
    "SHAP_abs_mean": top_shap
}).fillna(0).sort_values(by=["IG_pct","Saliency_pct"], ascending=False)

display(agg.head(20))
agg.to_csv("xai_aggregate_summary.csv")
print("Saved xai_aggregate_summary.csv")

print("\nAll XAI outputs saved:")
print("- xai_ig_local.csv")
print("- xai_saliency_local.csv")
print("- xai_permutation_importance.csv")
print("- xai_shap_local.csv")
print("- xai_aggregate_summary.csv")
print("Done.")
```

```python


import os
import random
from datetime import datetime
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import math
from captum.attr import IntegratedGradients

# ---------------------------
# Repro / device
# ---------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

# ---------------------------
# Config / Hyperparams
# ---------------------------
CSV_PATH = "reliance_with_indicators.csv"  # your preprocessed data
START_DATE = "2019-01-01"
SEQ_LEN = 60         # input window length (days)
BATCH_SIZE = 64
EPOCHS = 80
LR = 1e-4
D_MODEL = 128        # model hidden dim (embedding dim)
N_HEADS = 4
N_ENCODER_LAYERS = 3
N_DECODER_LAYERS = 2
FFN_DIM = 256
DROPOUT = 0.1
TEST_DAYS = 30
PATIENCE = 8
SAVE_PATH = "informer_like_reliance.pth"

# ---------------------------
# Load & preprocess
# ---------------------------
df = pd.read_csv(CSV_PATH, index_col=0, parse_dates=True)
df.sort_index(inplace=True)

# Get timezone from the DataFrame index
df_timezone = df.index.tz

# Convert START_DATE string to datetime and localize to the DataFrame's timezone
start_date_aware = pd.to_datetime(START_DATE).tz_localize(df_timezone)

# Filter the DataFrame using the timezone-aware start date
df = df[df.index >= start_date_aware]

df = df.ffill().bfill()
df.dropna(inplace=True)
print("Data range:", df.index.min(), "to", df.index.max())
print("Columns:", df.columns.tolist())

# Select features and targets
TARGET_COLS = ["Open", "Close"]
FEATURE_COLS = df.columns.tolist()  # use all columns (includes Open/Close)
# ensure targets exist
for t in TARGET_COLS:
    if t not in FEATURE_COLS:
        raise ValueError(f"Target {t} not in columns.")

# split test last TEST_DAYS
if TEST_DAYS >= len(df):
    raise ValueError("TEST_DAYS too large.")
train_val_df = df.iloc[:-TEST_DAYS]
test_df = df.iloc[-TEST_DAYS:].copy()

# further split train/val
val_frac = 0.1
val_size = int(len(train_val_df) * val_frac)
train_df = train_val_df.iloc[:-val_size].copy()
val_df = train_val_df.iloc[-val_size:].copy()

# Scaler (fit on train)
scaler = MinMaxScaler()
scaler.fit(train_df[FEATURE_COLS].values)

def scale_df(a_df):
    arr = scaler.transform(a_df[FEATURE_COLS].values)
    return pd.DataFrame(arr, index=a_df.index, columns=FEATURE_COLS)

train_scaled = scale_df(train_df)
val_scaled = scale_df(val_df)
test_scaled = scale_df(test_df)

# ---------------------------
# Dataset / DataLoader
# ---------------------------
class TimeSeriesSeqDataset(Dataset):
    def __init__(self, df_scaled, seq_len=SEQ_LEN, feature_cols=FEATURE_COLS, target_cols=TARGET_COLS):
        self.X = df_scaled.values.astype(np.float32)
        self.feature_cols = feature_cols
        self.target_cols = target_cols
        self.seq_len = seq_len
        self.indices = []
        # produce samples where input = seq_len days, target = next day target cols
        for i in range(seq_len - 1, len(self.X) - 1):
            self.indices.append(i)
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, idx):
        i = self.indices[idx]
        seq = self.X[i - (self.seq_len - 1): i + 1]  # shape (seq_len, n_features)
        target_row = self.X[i + 1]
        target = np.array([target_row[self.feature_cols.index(c)] for c in self.target_cols], dtype=np.float32)
        # transpose to (n_features, seq_len) for conv or keep (seq_len, n_features)
        return torch.from_numpy(seq), torch.from_numpy(target)

train_ds = TimeSeriesSeqDataset(train_scaled)
val_ds = TimeSeriesSeqDataset(val_scaled)
test_ds = TimeSeriesSeqDataset(pd.concat([train_scaled.tail(SEQ_LEN), test_scaled]))  # ensure enough history

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, drop_last=False)

print("Train samples:", len(train_ds), "Val samples:", len(val_ds))

# ---------------------------
# Positional encoding
# ---------------------------
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=1000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        pos = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(pos * div)
        pe[:, 1::2] = torch.cos(pos * div)
        pe = pe.unsqueeze(0)  # (1, max_len, d_model)
        self.register_buffer("pe", pe)
    def forward(self, x):
        # x: (batch, seq_len, d_model)
        seq_len = x.size(1)
        return x + self.pe[:, :seq_len, :]

# ---------------------------
# Informer-like encoder/decoder layers that expose attention weights
# ---------------------------
class MultiHeadAttentionWithWeights(nn.Module):
    def __init__(self, d_model, n_heads, dropout=0.0):
        super().__init__()
        self.mha = nn.MultiheadAttention(embed_dim=d_model, num_heads=n_heads, dropout=dropout, batch_first=True)
    def forward(self, q, k, v, attn_mask=None, key_padding_mask=None):
        # q,k,v shape: (batch, seq_len, d_model)
        out, attn_weights = self.mha(q, k, v, attn_mask=attn_mask, key_padding_mask=key_padding_mask, need_weights=True)
        # attn_weights: (batch, tgt_len, src_len) when batch_first=True
        return out, attn_weights

class EncoderLayer(nn.Module):
    def __init__(self, d_model, n_heads, ffn_dim, dropout=0.1):
        super().__init__()
        self.mha = MultiHeadAttentionWithWeights(d_model, n_heads, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(ffn_dim, d_model),
            nn.Dropout(dropout)
        )
        self.norm2 = nn.LayerNorm(d_model)
    def forward(self, x):
        # x: (batch, seq_len, d_model)
        attn_out, attn_w = self.mha(x, x, x)
        x = self.norm1(x + attn_out)
        f = self.ffn(x)
        x = self.norm2(x + f)
        return x, attn_w

class DecoderLayer(nn.Module):
    def __init__(self, d_model, n_heads, ffn_dim, dropout=0.1):
        super().__init__()
        self.self_mha = MultiHeadAttentionWithWeights(d_model, n_heads, dropout)
        self.cross_mha = MultiHeadAttentionWithWeights(d_model, n_heads, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(ffn_dim, d_model),
            nn.Dropout(dropout)
        )
        self.norm3 = nn.LayerNorm(d_model)
    def forward(self, tgt, memory, tgt_mask=None, memory_mask=None):
        # tgt: (batch, tgt_seq_len, d_model)
        # memory: (batch, mem_seq_len, d_model)
        self_attn_out, self_attn_w = self.self_mha(tgt, tgt, tgt, attn_mask=tgt_mask)
        tgt = self.norm1(tgt + self_attn_out)
        cross_attn_out, cross_attn_w = self.cross_mha(tgt, memory, memory, attn_mask=memory_mask)
        tgt = self.norm2(tgt + cross_attn_out)
        f = self.ffn(tgt)
        tgt = self.norm3(tgt + f)
        return tgt, self_attn_w, cross_attn_w

# ---------------------------
# Informer-like Model (Encoder-Decoder)
# ---------------------------
class InformerLike(nn.Module):
    def __init__(self, n_features, d_model=128, n_heads=4, enc_layers=3, dec_layers=2, ffn_dim=256, dropout=0.1, target_size=2):
        super().__init__()
        self.n_features = n_features
        self.d_model = d_model
        # input projection
        self.input_proj = nn.Linear(n_features, d_model)
        self.pos_enc = PositionalEncoding(d_model, max_len=2000)
        # encoder stack
        self.encoder_layers = nn.ModuleList([EncoderLayer(d_model, n_heads, ffn_dim, dropout) for _ in range(enc_layers)])
        # decoder uses a small learned query vector to decode next-step (like Informer decoders with queries)
        self.decoder_query = nn.Parameter(torch.randn(1, 1, d_model))  # single-step decode query (can be expanded)
        self.decoder_layers = nn.ModuleList([DecoderLayer(d_model, n_heads, ffn_dim, dropout) for _ in range(dec_layers)])
        # final MLP
        self.fc = nn.Sequential(
            nn.Linear(d_model, d_model//2),
            nn.ReLU(),
            nn.Linear(d_model//2, target_size)
        )
    def forward(self, src):
        """
        src: (batch, seq_len, n_features)
        returns: preds (batch, target_size), dict of attention weights
        """
        batch_size = src.size(0)
        x = self.input_proj(src)                # (batch, seq_len, d_model)
        x = self.pos_enc(x)
        attn_record = {"encoder": [], "decoder_self": [], "decoder_cross": []}
        mem = x
        # encoder
        for layer in self.encoder_layers:
            mem, attn_w = layer(mem)
            attn_record["encoder"].append(attn_w.detach().cpu().numpy())
        # prepare decoder input (repeat query for batch)
        tgt = self.decoder_query.repeat(batch_size, 1, 1)  # (batch, 1, d_model)
        # decoder
        for layer in self.decoder_layers:
            tgt, self_attn_w, cross_attn_w = layer(tgt, mem)
            attn_record["decoder_self"].append(self_attn_w.detach().cpu().numpy())
            attn_record["decoder_cross"].append(cross_attn_w.detach().cpu().numpy())
        # readout from decoder query
        out = tgt[:, -1, :]  # (batch, d_model)
        preds = self.fc(out) # (batch, target_size)
        return preds, attn_record

# ---------------------------
# Instantiate model
# ---------------------------
n_features = len(FEATURE_COLS)
model = InformerLike(n_features, d_model=D_MODEL, n_heads=N_HEADS,
                    enc_layers=N_ENCODER_LAYERS, dec_layers=N_DECODER_LAYERS,
                    ffn_dim=FFN_DIM, dropout=DROPOUT, target_size=len(TARGET_COLS)).to(DEVICE)
print(model)

# ---------------------------
# Training utils
# ---------------------------
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-6)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)

def train_one_epoch():
    model.train()
    losses = []
    for xb, yb in train_loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)
        optimizer.zero_grad()
        preds, _ = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        losses.append(loss.item())
    return np.mean(losses)

def eval_on_loader(loader):
    model.eval()
    losses = []
    preds_all = []
    trues_all = []
    attn_examples = None
    with torch.no_grad():
        for xb, yb in loader:
            xb = xb.to(DEVICE)
            yb = yb.to(DEVICE)
            preds, attn_rec = model(xb)
            loss = criterion(preds, yb)
            losses.append(loss.item())
            preds_all.append(preds.cpu().numpy())
            trues_all.append(yb.cpu().numpy())
            # keep attention from first batch for visualization
            if attn_examples is None:
                attn_examples = attn_rec
    if len(preds_all):
        preds_all = np.vstack(preds_all)
        trues_all = np.vstack(trues_all)
    else:
        preds_all = np.zeros((0, len(TARGET_COLS)))
        trues_all = np.zeros((0, len(TARGET_COLS)))
    return np.mean(losses), preds_all, trues_all, attn_examples

# ---------------------------
# Train loop with early stopping
# ---------------------------
best_val = 1e9
counter = 0

for epoch in range(1, EPOCHS + 1):
    train_loss = train_one_epoch()
    val_loss, _, _, _ = eval_on_loader(val_loader)
    scheduler.step(val_loss)
    print(f"Epoch {epoch:03d} | Train Loss {train_loss:.6f} | Val Loss {val_loss:.6f}")
    if val_loss < best_val - 1e-6:
        best_val = val_loss
        counter = 0
        torch.save({
            "model_state_dict": model.state_dict(),
            "scaler": scaler,
            "feature_cols": FEATURE_COLS,
            "target_cols": TARGET_COLS
        }, SAVE_PATH)
        print("  -> Saved best model.")
    else:
        counter += 1
        if counter >= PATIENCE:
            print("Early stopping triggered.")
            break

# load best
checkpoint = torch.load(SAVE_PATH, map_location=DEVICE, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
scaler = checkpoint["scaler"]

# ---------------------------
# Test set evaluation (construct test inputs & targets same way as training)
# ---------------------------
# Build test sequences and targets (pure numpy) using full scaled df to match earlier code logic
full_scaled = pd.DataFrame(scaler.transform(df[FEATURE_COLS].values), index=df.index, columns=FEATURE_COLS)

test_inputs = []
test_targets = []
test_dates = []
for i in range(len(df) - TEST_DAYS, len(df)):
    seq_end = i - 1
    seq_start = seq_end - (SEQ_LEN - 1)
    if seq_start < 0:
        continue
    seq = full_scaled.iloc[seq_start: seq_end + 1].values.astype(np.float32)
    target_row = full_scaled.iloc[i].values.astype(np.float32)
    target = np.array([target_row[FEATURE_COLS.index(c)] for c in TARGET_COLS], dtype=np.float32)
    test_inputs.append(seq)
    test_targets.append(target)
    test_dates.append(df.index[i])

test_inputs = np.stack(test_inputs)  # (N_test, seq_len, n_features)
test_targets = np.stack(test_targets)

model.eval()
with torch.no_grad():
    X_test_t = torch.from_numpy(test_inputs).to(DEVICE)
    preds_scaled, attn_test = model(X_test_t)
    preds_scaled = preds_scaled.cpu().numpy()

# inverse transform predictions and targets (helper)
def inverse_targets(scaled_arr):
    # scaled_arr shape: (n, n_targets)
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for i, col in enumerate(FEATURE_COLS):
        if col in TARGET_COLS:
            filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

preds_inv = inverse_targets(preds_scaled)
targets_inv = inverse_targets(test_targets)

rmse_open = math.sqrt(mean_squared_error(targets_inv[:,0], preds_inv[:,0]))
rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
mae_open = mean_absolute_error(targets_inv[:,0], preds_inv[:,0])
mae_close = mean_absolute_error(targets_inv[:,1], preds_inv[:,1])

print(f"Test RMSE Open: {rmse_open:.4f} | Close: {rmse_close:.4f}")
print(f"Test MAE  Open: {mae_open:.4f} | Close: {mae_close:.4f}")

# Plot actual vs predicted
plt.figure(figsize=(12,5))
plt.plot(test_dates, targets_inv[:,1], label="Actual Close")
plt.plot(test_dates, preds_inv[:,1], label="Pred Close", linestyle="--")
plt.title("Test: Close price - Actual vs Predicted")
plt.xticks(rotation=45)
plt.legend()
plt.show()

# ---------------------------
# Attention visualization (example for first test sample)
# ---------------------------
# attn_test: dict with lists of attention arrays (encoder layers list each -> shape (batch, seq_len, seq_len))
print("Available attention keys:", attn_test.keys())
# we'll visualize encoder attention of layer 0 for the first sample if available
if attn_test and "encoder" in attn_test and len(attn_test["encoder"])>0:
    enc0 = attn_test["encoder"][0]  # shape (batch, seq_len, seq_len)
    if enc0.shape[0] > 0:
        sample_attn = enc0[0]  # first batch sample
        plt.figure(figsize=(8,6))
        sns.heatmap(sample_attn, cmap="viridis")
        plt.title("Encoder Attention (layer 0) for first batch sample")
        plt.xlabel("Source time index")
        plt.ylabel("Target time index")
        plt.show()
else:
    print("No encoder attention available to visualize.")

# ---------------------------
# Captum Integrated Gradients (IG) explanations for one example
# We'll compute IG for both targets (Open and Close) for a single validation sample.
# ---------------------------
# For Captum, we need a wrapper that maps an input tensor => scalar output (we'll use predicted Close)
model.eval()
# define wrapper
class ModelWrapper(nn.Module):
    def __init__(self, model, target_col_idx=1):  # index 1 -> Close
        super().__init__()
        self.model = model
        self.target_col_idx = target_col_idx
    def forward(self, x):
        # x: (batch, seq_len, n_features)
        preds, _ = self.model(x)
        # return the Close prediction (scalar per batch)
        return preds[:, self.target_col_idx]

wrapper = ModelWrapper(model, target_col_idx=1).to(DEVICE)
ig = IntegratedGradients(wrapper)

# pick one test sample
sample_idx = 0
input_tensor = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).to(DEVICE)  # (1, seq_len, n_features)
input_tensor.requires_grad = True

# baseline: zero (or mean)
baseline = torch.zeros_like(input_tensor).to(DEVICE)

# compute IG attributions (summed over timesteps or per feature)
attrs, delta = ig.attribute(inputs=input_tensor, baselines=baseline, return_convergence_delta=True)
# attrs shape: (1, seq_len, n_features)
attrs_np = attrs.detach().cpu().numpy()[0]  # (seq_len, n_features)

# Summarize attributions per feature by sum(abs) over time
feat_imp = np.sum(np.abs(attrs_np), axis=0)
feat_ranking = sorted(zip(FEATURE_COLS, feat_imp), key=lambda x: x[1], reverse=True)
print("Top feature importances (IG, summed over time) for sample 0 (Close prediction):")
for f, v in feat_ranking[:10]:
    print(f"{f}: {v:.6f}")

# Plot top-k feature attributions over time for top 5 features
top_feats = [f for f, _ in feat_ranking[:5]]
top_idx = [FEATURE_COLS.index(f) for f in top_feats]

plt.figure(figsize=(12,6))
for i, fi in enumerate(top_idx):
    # The input tensor corresponds to a sequence window ending at test_dates[sample_idx]
  seq_end_date = test_dates[sample_idx]
  seq_start_date = df.index[df.index.get_loc(seq_end_date) - SEQ_LEN + 1]  # inclusive
  date_range = df.loc[seq_start_date:seq_end_date].index

  plt.plot(date_range, attrs_np[:, fi], label=FEATURE_COLS[fi])

plt.legend()
plt.title("IG attributions over time (sample 0) for top 5 features")
plt.xticks(rotation=45)
plt.show()

# ---------------------------
# Save final artifacts
# ---------------------------
torch.save({
    "model_state_dict": model.state_dict(),
    "scaler": scaler,
    "feature_cols": FEATURE_COLS,
    "target_cols": TARGET_COLS
}, "informer_like_reliance_final.pth")

print("Done. Model saved to informer_like_reliance_final.pth")
```

```python
# ============================================================
# 📊 Explainable AI (XAI) for Informer-like Transformer model
# ============================================================
# Requirements:
# pip install captum shap seaborn --quiet

import torch
import numpy as np
import pandas as pd
import math
import shap
import matplotlib.pyplot as plt
import seaborn as sns
from captum.attr import IntegratedGradients, Saliency
from sklearn.metrics import mean_squared_error

sns.set()

print("\n========== Running Explainable AI (XAI) Analysis ==========")

# Check dependencies from your informer code
assert 'model' in globals(), "Error: model not found. Run the training cell first."
assert 'test_inputs' in globals(), "Error: test_inputs missing. Run evaluation cell first."
assert 'test_targets' in globals(), "Error: test_targets missing. Run evaluation cell first."

DEVICE = DEVICE if 'DEVICE' in globals() else torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(DEVICE)
model.eval()

# ---------------------------
# Helper functions
# ---------------------------
def inverse_targets_local(scaled_arr):
    """Inverse transform scaled Open/Close targets."""
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for col in TARGET_COLS:
        filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

def predict_unscaled(X_np):
    """Run model inference and inverse-scale outputs."""
    with torch.no_grad():
        x_t = torch.from_numpy(X_np.astype(np.float32)).to(DEVICE)
        preds, _ = model(x_t)
        preds_np = preds.cpu().numpy()
    return inverse_targets_local(preds_np)

# ---------------------------
# Baseline performance check
# ---------------------------
preds_inv = predict_unscaled(test_inputs)
targets_inv = inverse_targets_local(test_targets)
base_rmse = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
print(f"✅ Base Test RMSE (Close): {base_rmse:.4f}")

# ============================================================
# 1️⃣ Integrated Gradients (Local)
# ============================================================
print("\n--- Integrated Gradients ---")

class IGWrapper(torch.nn.Module):
    def __init__(self, base, target_idx=1):
        super().__init__()
        self.base = base
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.base(x)
        return preds[:, self.target_idx]

def compute_ig(sample_idx, steps=50):
    wrapper = IGWrapper(model, target_idx=1).to(DEVICE)
    ig = IntegratedGradients(wrapper)
    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).float().to(DEVICE)
    baseline = torch.zeros_like(inp).to(DEVICE)
    attrs, delta = ig.attribute(inputs=inp, baselines=baseline, n_steps=steps, return_convergence_delta=True)
    attrs_np = attrs.squeeze(0).detach().cpu().numpy()
    feat_score = np.sum(np.abs(attrs_np), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'score': feat_score})
    df['pct'] = 100 * df['score'] / df['score'].sum()
    df = df.sort_values('score', ascending=False)
    return df, attrs_np

IG_SAMPLES = min(6, len(test_inputs))
ig_rows = []

for i in range(IG_SAMPLES):
    df_ig, ig_map = compute_ig(i)
    print(f"\nSample {i} ({test_dates[i].date()}) top IG features:")
    display(df_ig.head(8))
    plt.figure(figsize=(10,4))
    sns.heatmap(ig_map.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"IG Heatmap - {test_dates[i].date()}")
    plt.tight_layout(); plt.show()
    for r in df_ig.head(10).itertuples():
        ig_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'score': r.score, 'pct': r.pct})
pd.DataFrame(ig_rows).to_csv("informer_ig_summary.csv", index=False)
print("✅ Saved informer_ig_summary.csv")

# ============================================================
# 2️⃣ Saliency (Grad × Input)
# ============================================================
print("\n--- Saliency ---")

class SalWrapper(torch.nn.Module):
    def __init__(self, base, target_idx=1):
        super().__init__()
        self.base = base
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.base(x)
        return preds[:, self.target_idx]

def compute_saliency(sample_idx):
    wrapper = SalWrapper(model, target_idx=1).to(DEVICE)
    sal = Saliency(wrapper)
    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).float().to(DEVICE)
    inp.requires_grad = True
    attr = sal.attribute(inp)
    attr_np = attr.squeeze(0).detach().cpu().numpy()
    grad_x_input = attr_np * test_inputs[sample_idx]
    feat_score = np.sum(np.abs(grad_x_input), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'score': feat_score})
    df['pct'] = 100 * df['score'] / df['score'].sum()
    df = df.sort_values('score', ascending=False)
    return df, grad_x_input

sal_rows = []
for i in range(IG_SAMPLES):
    df_sal, sal_map = compute_saliency(i)
    print(f"\nSample {i} ({test_dates[i].date()}) top Saliency features:")
    display(df_sal.head(8))
    plt.figure(figsize=(10,4))
    sns.heatmap(sal_map.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"Saliency Heatmap - {test_dates[i].date()}")
    plt.tight_layout(); plt.show()
    for r in df_sal.head(10).itertuples():
        sal_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'score': r.score, 'pct': r.pct})
pd.DataFrame(sal_rows).to_csv("informer_saliency_summary.csv", index=False)
print("✅ Saved informer_saliency_summary.csv")

# ============================================================
# 3️⃣ Permutation Feature Importance (Global)
# ============================================================
print("\n--- Permutation Feature Importance ---")

def permutation_importance(X_np, y_true_unscaled, features, target_idx=1, n_repeats=3):
    preds_base = predict_unscaled(X_np)
    base_rmse = math.sqrt(mean_squared_error(y_true_unscaled[:, target_idx], preds_base[:, target_idx]))
    results = []
    for i, f in enumerate(features):
        deltas = []
        for _ in range(n_repeats):
            Xp = X_np.copy()
            np.random.shuffle(Xp[:,:,i])
            preds_perm = predict_unscaled(Xp)
            rmse = math.sqrt(mean_squared_error(y_true_unscaled[:, target_idx], preds_perm[:, target_idx]))
            deltas.append(rmse - base_rmse)
        results.append((f, np.mean(deltas)))
    df = pd.DataFrame(results, columns=['feature','delta_RMSE']).sort_values('delta_RMSE', ascending=False)
    return df

perm_df = permutation_importance(test_inputs, targets_inv, FEATURE_COLS)
display(perm_df.head(12))
perm_df.to_csv("informer_permutation_importance.csv", index=False)
print("✅ Saved informer_permutation_importance.csv")

plt.figure(figsize=(8,6))
sns.barplot(y="feature", x="delta_RMSE", data=perm_df.head(12), palette="viridis")
plt.title("Permutation Feature Importance (Close)")
plt.tight_layout(); plt.show()

# ============================================================
# 4️⃣ SHAP (KernelExplainer - Average Window)
# ============================================================
print("\n--- SHAP Analysis ---")

SHAP_BG = min(50, len(test_inputs))
SHAP_EXPLAIN_N = min(5, len(test_inputs))
SHAP_NSAMPLES = 200

X_avg = np.mean(test_inputs, axis=1)
background = X_avg[np.random.choice(len(X_avg), SHAP_BG, replace=False)]

def predict_flat(x_flat):
    seq = np.repeat(x_flat[:, np.newaxis, :], SEQ_LEN, axis=1).astype(np.float32)
    preds = predict_unscaled(seq)
    return preds[:, 1]

explainer = shap.KernelExplainer(predict_flat, background)
to_explain = X_avg[:SHAP_EXPLAIN_N]
shap_vals = explainer.shap_values(to_explain, nsamples=SHAP_NSAMPLES)

shap_rows = []
for i, sv in enumerate(shap_vals):
    df_sh = pd.DataFrame({'feature': FEATURE_COLS, 'shap': sv})
    df_sh['abs'] = np.abs(df_sh['shap'])
    df_sh = df_sh.sort_values('abs', ascending=False)
    print(f"\nSample {i} ({test_dates[i].date()}) SHAP top features:")
    display(df_sh.head(8))
    plt.figure(figsize=(8,4))
    sns.barplot(x='shap', y='feature', data=df_sh.head(10), palette="coolwarm")
    plt.title(f"SHAP Feature Contributions ({test_dates[i].date()})")
    plt.tight_layout(); plt.show()
    for r in df_sh.head(10).itertuples():
        shap_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'shap': r.shap, 'abs': r.abs})
pd.DataFrame(shap_rows).to_csv("informer_shap_summary.csv", index=False)
print("✅ Saved informer_shap_summary.csv")

# ============================================================
# ✅ Final Summary
# ============================================================
print("\nExplainability completed. Files saved:")
print("- informer_ig_summary.csv  (Integrated Gradients)")
print("- informer_saliency_summary.csv  (Saliency)")
print("- informer_permutation_importance.csv  (Global)")
print("- informer_shap_summary.csv  (SHAP)")
```

```python
# ============================================================
# 📊 Explainable AI (XAI) for 2025 Forecast Evaluation
# ============================================================
# Requirements:
# pip install captum shap seaborn --quiet

import torch
import numpy as np
import pandas as pd
import math
import shap
import matplotlib.pyplot as plt
import seaborn as sns
from captum.attr import IntegratedGradients, Saliency
from sklearn.metrics import mean_squared_error

sns.set()
print("\n========== Running Explainable AI (XAI) Analysis for 2025 ==========")

# ------------------------------------------------------
# Helpers
# ------------------------------------------------------
def inverse_targets_local(scaled_arr):
    """Inverse transform scaled Open/Close targets."""
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for col in TARGET_COLS:
        filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

def predict_unscaled(X_np):
    """Run model inference and inverse-scale outputs."""
    with torch.no_grad():
        x_t = torch.from_numpy(X_np.astype(np.float32)).to(DEVICE)
        preds, _ = model(x_t)
        preds_np = preds.cpu().numpy()
    return inverse_targets_local(preds_np)

# Reconfirm base performance (Close)
preds_inv = predict_unscaled(test_inputs)
targets_inv_local = inverse_targets_local(test_targets)
base_rmse = math.sqrt(mean_squared_error(targets_inv_local[:,1], preds_inv[:,1]))
print(f"✅ Base Test RMSE (Close): {base_rmse:.4f}")

# ============================================================
# 1️⃣ Integrated Gradients (Local explanations)
# ============================================================
print("\n--- Integrated Gradients ---")

class IGWrapper(torch.nn.Module):
    def __init__(self, base, target_idx=1):
        super().__init__()
        self.base = base
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.base(x)
        return preds[:, self.target_idx]

def compute_ig(sample_idx, steps=50):
    wrapper = IGWrapper(model, target_idx=1).to(DEVICE)
    ig = IntegratedGradients(wrapper)
    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).float().to(DEVICE)
    baseline = torch.zeros_like(inp).to(DEVICE)
    attrs, delta = ig.attribute(inputs=inp, baselines=baseline, n_steps=steps, return_convergence_delta=True)
    attrs_np = attrs.squeeze(0).detach().cpu().numpy()
    feat_score = np.sum(np.abs(attrs_np), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'score': feat_score})
    df['pct'] = 100 * df['score'] / df['score'].sum()
    df = df.sort_values('score', ascending=False)
    return df, attrs_np

IG_SAMPLES = min(6, len(test_inputs))
ig_rows = []

for i in range(IG_SAMPLES):
    df_ig, ig_map = compute_ig(i)
    print(f"\nSample {i} ({test_dates[i].date()}) top IG features:")
    display(df_ig.head(8))
    plt.figure(figsize=(10,4))
    sns.heatmap(ig_map.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"IG Heatmap - {test_dates[i].date()}")
    plt.tight_layout(); plt.show()
    for r in df_ig.head(10).itertuples():
        ig_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'score': r.score, 'pct': r.pct})
pd.DataFrame(ig_rows).to_csv("2025_informer_ig_summary.csv", index=False)
print("✅ Saved 2025_informer_ig_summary.csv")

# ============================================================
# 2️⃣ Saliency (Grad × Input)
# ============================================================
print("\n--- Saliency ---")

class SalWrapper(torch.nn.Module):
    def __init__(self, base, target_idx=1):
        super().__init__()
        self.base = base
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.base(x)
        return preds[:, self.target_idx]

def compute_saliency(sample_idx):
    wrapper = SalWrapper(model, target_idx=1).to(DEVICE)
    sal = Saliency(wrapper)
    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).float().to(DEVICE)
    inp.requires_grad = True
    attr = sal.attribute(inp)
    attr_np = attr.squeeze(0).detach().cpu().numpy()
    grad_x_input = attr_np * test_inputs[sample_idx]
    feat_score = np.sum(np.abs(grad_x_input), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'score': feat_score})
    df['pct'] = 100 * df['score'] / df['score'].sum()
    df = df.sort_values('score', ascending=False)
    return df, grad_x_input

sal_rows = []
for i in range(IG_SAMPLES):
    df_sal, sal_map = compute_saliency(i)
    print(f"\nSample {i} ({test_dates[i].date()}) top Saliency features:")
    display(df_sal.head(8))
    plt.figure(figsize=(10,4))
    sns.heatmap(sal_map.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"Saliency Heatmap - {test_dates[i].date()}")
    plt.tight_layout(); plt.show()
    for r in df_sal.head(10).itertuples():
        sal_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'score': r.score, 'pct': r.pct})
pd.DataFrame(sal_rows).to_csv("2025_informer_saliency_summary.csv", index=False)
print("✅ Saved 2025_informer_saliency_summary.csv")

# ============================================================
# 3️⃣ Permutation Feature Importance (Global)
# ============================================================
print("\n--- Permutation Feature Importance ---")

def permutation_importance(X_np, y_true_unscaled, features, target_idx=1, n_repeats=3):
    preds_base = predict_unscaled(X_np)
    base_rmse = math.sqrt(mean_squared_error(y_true_unscaled[:, target_idx], preds_base[:, target_idx]))
    results = []
    for i, f in enumerate(features):
        deltas = []
        for _ in range(n_repeats):
            Xp = X_np.copy()
            np.random.shuffle(Xp[:,:,i])
            preds_perm = predict_unscaled(Xp)
            rmse = math.sqrt(mean_squared_error(y_true_unscaled[:, target_idx], preds_perm[:, target_idx]))
            deltas.append(rmse - base_rmse)
        results.append((f, np.mean(deltas)))
    df = pd.DataFrame(results, columns=['feature','delta_RMSE']).sort_values('delta_RMSE', ascending=False)
    return df

perm_df = permutation_importance(test_inputs, targets_inv, FEATURE_COLS)
display(perm_df.head(12))
perm_df.to_csv("2025_informer_permutation_importance.csv", index=False)
print("✅ Saved 2025_informer_permutation_importance.csv")

plt.figure(figsize=(8,6))
sns.barplot(y="feature", x="delta_RMSE", data=perm_df.head(12), palette="viridis")
plt.title("Permutation Feature Importance (Close)")
plt.tight_layout(); plt.show()

# ============================================================
# 4️⃣ SHAP (KernelExplainer)
# ============================================================
print("\n--- SHAP Analysis ---")

SHAP_BG = min(50, len(test_inputs))
SHAP_EXPLAIN_N = min(5, len(test_inputs))
SHAP_NSAMPLES = 200

X_avg = np.mean(test_inputs, axis=1)
background = X_avg[np.random.choice(len(X_avg), SHAP_BG, replace=False)]

def predict_flat(x_flat):
    seq = np.repeat(x_flat[:, np.newaxis, :], SEQ_LEN, axis=1).astype(np.float32)
    preds = predict_unscaled(seq)
    return preds[:, 1]  # Close

explainer = shap.KernelExplainer(predict_flat, background)
to_explain = X_avg[:SHAP_EXPLAIN_N]
shap_vals = explainer.shap_values(to_explain, nsamples=SHAP_NSAMPLES)

shap_rows = []
for i, sv in enumerate(shap_vals):
    df_sh = pd.DataFrame({'feature': FEATURE_COLS, 'shap': sv})
    df_sh['abs'] = np.abs(df_sh['shap'])
    df_sh = df_sh.sort_values('abs', ascending=False)
    print(f"\nSample {i} ({test_dates[i].date()}) SHAP top features:")
    display(df_sh.head(8))
    plt.figure(figsize=(8,4))
    sns.barplot(x='shap', y='feature', data=df_sh.head(10), palette="coolwarm")
    plt.title(f"SHAP Feature Contributions ({test_dates[i].date()})")
    plt.tight_layout(); plt.show()
    for r in df_sh.head(10).itertuples():
        shap_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'shap': r.shap, 'abs': r.abs})
pd.DataFrame(shap_rows).to_csv("2025_informer_shap_summary.csv", index=False)
print("✅ Saved 2025_informer_shap_summary.csv")

# ============================================================
# ✅ Final Summary
# ============================================================
print("\n✅ Explainability complete for 2025 test data.")
print("Generated files:")
print("- 2025_informer_ig_summary.csv")
print("- 2025_informer_saliency_summary.csv")
print("- 2025_informer_permutation_importance.csv")
print("- 2025_informer_shap_summary.csv")
```

```python
# convtransformer_reliance.py
# ConvTransformer: Conv1D -> Transformer Encoder (attention exposed) -> MLP

# Dependencies: torch, numpy, pandas, scikit-learn, matplotlib, seaborn, captum, pandas_ta (for preprocessing)
# Ensure reliance_with_indicators.csv exists (from your preprocessing code)

import os
import random
import math
from datetime import datetime
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
from captum.attr import IntegratedGradients

# ---------------------------
# Repro / device
# ---------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

# ---------------------------
# Config / Hyperparams
# ---------------------------
CSV_PATH = "reliance_with_indicators.csv"
START_DATE = "2019-01-01"
SEQ_LEN = 60          # input window length (days)
BATCH_SIZE = 48
EPOCHS = 60
LR = 1e-4
D_MODEL = 128         # feature embedding dim after projection
N_HEADS = 4
ENC_LAYERS = 3
FFN_DIM = 256
DROPOUT = 0.1
TEST_DAYS = 30
PATIENCE = 8
SAVE_PATH = "convtransformer_reliance.pth"

# ---------------------------
# Load & preprocess
# ---------------------------
df = pd.read_csv(CSV_PATH, index_col=0, parse_dates=True)
df.sort_index(inplace=True)

# Get timezone from the DataFrame index
df_timezone = df.index.tz

# Convert START_DATE string to datetime and localize to the DataFrame's timezone
start_date_aware = pd.to_datetime(START_DATE).tz_localize(df_timezone)

# Filter the DataFrame using the timezone-aware start date
df = df[df.index >= start_date_aware]


df = df.ffill().bfill()
df.dropna(inplace=True)
print("Data range:", df.index.min(), "to", df.index.max())
print("Columns:", df.columns.tolist())

TARGET_COLS = ["Open", "Close"]
FEATURE_COLS = df.columns.tolist()
for t in TARGET_COLS:
    if t not in FEATURE_COLS:
        raise ValueError(f"Target {t} not in columns.")

# train/val/test split
if TEST_DAYS >= len(df):
    raise ValueError("TEST_DAYS too large.")
train_val_df = df.iloc[:-TEST_DAYS]
test_df = df.iloc[-TEST_DAYS:].copy()
val_frac = 0.1
val_size = int(len(train_val_df) * val_frac)
train_df = train_val_df.iloc[:-val_size].copy()
val_df = train_val_df.iloc[-val_size:].copy()

# scaler
scaler = MinMaxScaler()
scaler.fit(train_df[FEATURE_COLS].values)

def scale_df(a_df):
    arr = scaler.transform(a_df[FEATURE_COLS].values)
    return pd.DataFrame(arr, index=a_df.index, columns=FEATURE_COLS)

train_scaled = scale_df(train_df)
val_scaled = scale_df(val_df)
test_scaled = scale_df(test_df)

# ---------------------------
# Dataset & DataLoader
# ---------------------------
class TimeSeriesSeqDataset(Dataset):
    def __init__(self, df_scaled, seq_len=SEQ_LEN, feature_cols=FEATURE_COLS, target_cols=TARGET_COLS):
        self.X = df_scaled.values.astype(np.float32)
        self.feature_cols = feature_cols
        self.target_cols = target_cols
        self.seq_len = seq_len
        self.indices = []
        for i in range(seq_len - 1, len(self.X) - 1):
            self.indices.append(i)
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, idx):
        i = self.indices[idx]
        seq = self.X[i - (self.seq_len - 1): i + 1]  # (seq_len, n_features)
        target_row = self.X[i + 1]
        target = np.array([target_row[self.feature_cols.index(c)] for c in self.target_cols], dtype=np.float32)
        return torch.from_numpy(seq), torch.from_numpy(target)

train_ds = TimeSeriesSeqDataset(train_scaled)
val_ds = TimeSeriesSeqDataset(val_scaled)
# for tests create sequences similarly later from full-scaled

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, drop_last=False)
print("Train samples:", len(train_ds), "Val samples:", len(val_ds))

# ---------------------------
# Positional encoding
# ---------------------------
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=2000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        pos = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(pos * div)
        pe[:, 1::2] = torch.cos(pos * div)
        pe = pe.unsqueeze(0)  # (1, max_len, d_model)
        self.register_buffer("pe", pe)
    def forward(self, x):
        return x + self.pe[:, :x.size(1), :]

# ---------------------------
# Attention wrapper to obtain weights
# ---------------------------
class MultiHeadAttentionWithWeights(nn.Module):
    def __init__(self, d_model, n_heads, dropout=0.0):
        super().__init__()
        self.mha = nn.MultiheadAttention(embed_dim=d_model, num_heads=n_heads, dropout=dropout, batch_first=True)
    def forward(self, q, k, v, attn_mask=None, key_padding_mask=None):
        out, attn_weights = self.mha(q, k, v, attn_mask=attn_mask, key_padding_mask=key_padding_mask, need_weights=True)
        return out, attn_weights

class TransformerEncoderLayerWithWeights(nn.Module):
    def __init__(self, d_model, n_heads, ffn_dim, dropout=0.1):
        super().__init__()
        self.mha = MultiHeadAttentionWithWeights(d_model, n_heads, dropout)
        self.norm1 = nn.LayerNorm(d_model)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(ffn_dim, d_model),
            nn.Dropout(dropout)
        )
        self.norm2 = nn.LayerNorm(d_model)
    def forward(self, x):
        attn_out, attn_w = self.mha(x, x, x)
        x = self.norm1(x + attn_out)
        f = self.ffn(x)
        x = self.norm2(x + f)
        return x, attn_w

# ---------------------------
# ConvTransformer Model
# ---------------------------
class ConvTransformer(nn.Module):
    def __init__(self, n_features, conv_channels=[64, 96], kernel_size=3, d_model=128, n_heads=4, n_layers=3, ffn_dim=256, dropout=0.1, target_size=2):
        super().__init__()
        self.n_features = n_features
        # Conv stack: input shape (batch, seq_len, n_features) -> transpose -> (batch, n_features, seq_len)
        # We'll apply conv over time for each feature jointly by projecting input -> channels
        self.input_proj = nn.Conv1d(in_channels=n_features, out_channels=conv_channels[0], kernel_size=1)
        convs = []
        in_ch = conv_channels[0]
        for out_ch in conv_channels:
            convs.append(nn.Conv1d(in_channels=in_ch, out_channels=out_ch, kernel_size=kernel_size, padding=kernel_size//2))
            convs.append(nn.ReLU())
            convs.append(nn.BatchNorm1d(out_ch))
            in_ch = out_ch
        self.conv_stack = nn.Sequential(*convs)
        # project conv output channels -> d_model
        self.conv_to_model = nn.Conv1d(in_channels=in_ch, out_channels=d_model, kernel_size=1)
        self.pos_enc = PositionalEncoding(d_model, max_len=2000)
        # transformer encoder layers with exposed attention
        self.encoder_layers = nn.ModuleList([TransformerEncoderLayerWithWeights(d_model, n_heads, ffn_dim, dropout) for _ in range(n_layers)])
        # pooling and readout
        self.pool = nn.AdaptiveAvgPool1d(1)  # pool over time -> (batch, d_model, 1)
        self.fc = nn.Sequential(
            nn.Linear(d_model, d_model//2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(d_model//2, target_size)
        )

    def forward(self, src):
        """
        src: (batch, seq_len, n_features)
        """
        b, seq_len, nfeat = src.size()
        # conv expects (batch, channels, seq_len) where channels=input features after initial proj
        x = src.permute(0, 2, 1)  # (batch, n_features, seq_len)
        x = self.input_proj(x)    # (batch, conv_ch0, seq_len)
        x = self.conv_stack(x)    # (batch, conv_ch_last, seq_len)
        x = self.conv_to_model(x) # (batch, d_model, seq_len)
        x = x.permute(0, 2, 1)    # (batch, seq_len, d_model)
        x = self.pos_enc(x)
        attn_record = []
        mem = x
        for layer in self.encoder_layers:
            mem, attn_w = layer(mem)
            attn_record.append(attn_w.detach().cpu().numpy())  # (batch, seq_len, seq_len)
        # pool over time and predict
        pooled = self.pool(mem.permute(0,2,1)).squeeze(-1)  # (batch, d_model)
        preds = self.fc(pooled)
        return preds, attn_record

# instantiate
n_features = len(FEATURE_COLS)
model = ConvTransformer(n_features=n_features, conv_channels=[64,96], kernel_size=3,
                        d_model=D_MODEL, n_heads=N_HEADS, n_layers=ENC_LAYERS, ffn_dim=FFN_DIM,
                        dropout=DROPOUT, target_size=len(TARGET_COLS)).to(DEVICE)
print(model)

# ---------------------------
# Training utils
# ---------------------------
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-6)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3) # Removed verbose=True

def train_one_epoch():
    model.train()
    losses = []
    for xb, yb in train_loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)
        optimizer.zero_grad()
        preds, _ = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        losses.append(loss.item())
    return np.mean(losses)

def eval_on_loader(loader):
    model.eval()
    losses = []
    preds_all = []
    trues_all = []
    attn_examples = None
    with torch.no_grad():
        for xb, yb in loader:
            xb = xb.to(DEVICE)
            yb = yb.to(DEVICE)
            preds, attn_rec = model(xb)
            loss = criterion(preds, yb)
            losses.append(loss.item())
            preds_all.append(preds.cpu().numpy())
            trues_all.append(yb.cpu().numpy())
            if attn_examples is None:
                attn_examples = attn_rec
    if len(preds_all):
        preds_all = np.vstack(preds_all)
        trues_all = np.vstack(trues_all)
    else:
        preds_all = np.zeros((0, len(TARGET_COLS)))
        trues_all = np.zeros((0, len(TARGET_COLS)))
    return np.mean(losses), preds_all, trues_all, attn_examples

# ---------------------------
# Train loop with early stopping
# ---------------------------
best_val = 1e9
counter = 0

for epoch in range(1, EPOCHS + 1):
    train_loss = train_one_epoch()
    val_loss, _, _, _ = eval_on_loader(val_loader)
    scheduler.step(val_loss)
    print(f"Epoch {epoch:03d} | Train Loss {train_loss:.6f} | Val Loss {val_loss:.6f}")
    if val_loss < best_val - 1e-6:
        best_val = val_loss
        counter = 0
        torch.save({
            "model_state_dict": model.state_dict(),
            "scaler": scaler,
            "feature_cols": FEATURE_COLS,
            "target_cols": TARGET_COLS
        }, SAVE_PATH)
        print("  -> Saved best model.")
    else:
        counter += 1
        if counter >= PATIENCE:
            print("Early stopping triggered.")
            break

# load best
checkpoint = torch.load(SAVE_PATH, map_location=DEVICE, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
scaler = checkpoint["scaler"]

# ---------------------------
# Build test sequences (same logic as earlier)
# ---------------------------
full_scaled = pd.DataFrame(scaler.transform(df[FEATURE_COLS].values), index=df.index, columns=FEATURE_COLS)

test_inputs = []
test_targets = []
test_dates = []
for i in range(len(df) - TEST_DAYS, len(df)):
    seq_end = i - 1
    seq_start = seq_end - (SEQ_LEN - 1)
    if seq_start < 0:
        continue
    seq = full_scaled.iloc[seq_start: seq_end + 1].values.astype(np.float32)
    target_row = full_scaled.iloc[i].values.astype(np.float32)
    target = np.array([target_row[FEATURE_COLS.index(c)] for c in TARGET_COLS], dtype=np.float32)
    test_inputs.append(seq)
    test_targets.append(target)
    test_dates.append(df.index[i])

test_inputs = np.stack(test_inputs)
test_targets = np.stack(test_targets)

# run model on test inputs
model.eval()
with torch.no_grad():
    X_test_t = torch.from_numpy(test_inputs).to(DEVICE)
    preds_scaled, attn_test = model(X_test_t)
    preds_scaled = preds_scaled.cpu().numpy()

# inverse transform helper (same as earlier)
def inverse_targets(scaled_arr):
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for i, col in enumerate(FEATURE_COLS):
        if col in TARGET_COLS:
            filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

preds_inv = inverse_targets(preds_scaled)
targets_inv = inverse_targets(test_targets)

rmse_open = math.sqrt(mean_squared_error(targets_inv[:,0], preds_inv[:,0]))
rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
mae_open = mean_absolute_error(targets_inv[:,0], preds_inv[:,0])
mae_close = mean_absolute_error(targets_inv[:,1], preds_inv[:,1])

print(f"Test RMSE Open: {rmse_open:.4f} | Close: {rmse_close:.4f}")
print(f"Test MAE  Open: {mae_open:.4f} | Close: {mae_close:.4f}")

# plot actual vs predicted close
plt.figure(figsize=(12,5))
plt.plot(test_dates, targets_inv[:,1], label="Actual Close")
plt.plot(test_dates, preds_inv[:,1], label="Pred Close", linestyle="--")
plt.title("Test: Close price - Actual vs Predicted (ConvTransformer)")
plt.xticks(rotation=45)
plt.legend()
plt.show()

# ---------------------------
# Attention visualization (layer 0 first test sample)
# ---------------------------
if attn_test and len(attn_test)>0:
    # attn_test is a list of arrays per layer; each arr shape (batch, seq_len, seq_len)
    layer0 = attn_test[0]
    if layer0.shape[0] > 0:
        sample_attn = layer0[0]  # first sample
        plt.figure(figsize=(8,6))
        sns.heatmap(sample_attn, cmap="viridis")
        plt.title("Encoder Attention (layer 0) for first test sample")
        plt.xlabel("Source time index")
        plt.ylabel("Target time index")
        plt.show()
else:
    print("No attention recorded.")

# ---------------------------
# Captum Integrated Gradients for one test example (explain Close prediction)
# ---------------------------
class WrapperClose(nn.Module):
    def __init__(self, model, target_idx=1):
        super().__init__()
        self.model = model
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.model(x)
        return preds[:, self.target_idx]

wrapper = WrapperClose(model, target_idx=1).to(DEVICE)
ig = IntegratedGradients(wrapper)

sample_idx = 0
input_tensor = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).to(DEVICE)
input_tensor.requires_grad = True
baseline = torch.zeros_like(input_tensor).to(DEVICE)  # zero baseline

attrs, delta = ig.attribute(inputs=input_tensor, baselines=baseline, return_convergence_delta=True)
attrs_np = attrs.detach().cpu().numpy()[0]  # (seq_len, n_features)

# summarize attributions per feature
feat_imp = np.sum(np.abs(attrs_np), axis=0)
feat_ranking = sorted(zip(FEATURE_COLS, feat_imp), key=lambda x: x[1], reverse=True)
print("Top feature importances (IG, summed over time) for sample 0 (Close prediction):")
for f, v in feat_ranking[:10]:
    print(f"{f}: {v:.6f}")

# plot top 5 features attributions over time
top_feats = [f for f, _ in feat_ranking[:5]]
top_idx = [FEATURE_COLS.index(f) for f in top_feats]
plt.figure(figsize=(12,6))
# Corrected date slicing to match the sequence length
ig_dates = df.index[len(df) - TEST_DAYS + sample_idx - SEQ_LEN + 1 : len(df) - TEST_DAYS + sample_idx + 1]
plt.plot(ig_dates, attrs_np[:, fi], label=FEATURE_COLS[fi])

plt.legend()
plt.title("IG attributions over time (sample 0) for top 5 features")
plt.xticks(rotation=45)
plt.show()

# ---------------------------
# Save final artifacts
# ---------------------------
torch.save({
    "model_state_dict": model.state_dict(),
    "scaler": scaler,
    "feature_cols": FEATURE_COLS,
    "target_cols": TARGET_COLS
}, "convtransformer_reliance_final.pth")

print("Done. Model saved to convtransformer_reliance_final.pth")
```

```python
# ============================================================
# 📊 Explainable AI (XAI) for ConvTransformer model
# ============================================================
# Requirements:
# pip install captum shap seaborn --quiet

import torch
import numpy as np
import pandas as pd
import math
import shap
import matplotlib.pyplot as plt
import seaborn as sns
from captum.attr import IntegratedGradients, Saliency
from sklearn.metrics import mean_squared_error

sns.set()
print("\n========== Running Explainable AI (XAI) Analysis ==========")

# ------------------------------------------------------
# Helper functions
# ------------------------------------------------------
def inverse_targets_local(scaled_arr):
    """Inverse transform scaled Open/Close targets."""
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for col in TARGET_COLS:
        filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

def predict_unscaled(X_np):
    """Run model inference and inverse-scale outputs."""
    with torch.no_grad():
        x_t = torch.from_numpy(X_np.astype(np.float32)).to(DEVICE)
        preds, _ = model(x_t)
        preds_np = preds.cpu().numpy()
    return inverse_targets_local(preds_np)

# Check base RMSE
preds_inv = predict_unscaled(test_inputs)
targets_inv_local = inverse_targets_local(test_targets)
base_rmse = math.sqrt(mean_squared_error(targets_inv_local[:,1], preds_inv[:,1]))
print(f"✅ Base Test RMSE (Close): {base_rmse:.4f}")

# ============================================================
# 1️⃣ Integrated Gradients (Local explanations)
# ============================================================
print("\n--- Integrated Gradients ---")

class IGWrapper(torch.nn.Module):
    def __init__(self, base, target_idx=1):
        super().__init__()
        self.base = base
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.base(x)
        return preds[:, self.target_idx]

def compute_ig(sample_idx, steps=50):
    wrapper = IGWrapper(model, target_idx=1).to(DEVICE)
    ig = IntegratedGradients(wrapper)
    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).float().to(DEVICE)
    baseline = torch.zeros_like(inp).to(DEVICE)
    attrs, delta = ig.attribute(inputs=inp, baselines=baseline, n_steps=steps, return_convergence_delta=True)
    attrs_np = attrs.squeeze(0).detach().cpu().numpy()
    feat_score = np.sum(np.abs(attrs_np), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'score': feat_score})
    df['pct'] = 100 * df['score'] / df['score'].sum()
    df = df.sort_values('score', ascending=False)
    return df, attrs_np

IG_SAMPLES = min(5, len(test_inputs))
ig_rows = []

for i in range(IG_SAMPLES):
    df_ig, ig_map = compute_ig(i)
    print(f"\nSample {i} ({test_dates[i].date()}) top IG features:")
    display(df_ig.head(8))
    plt.figure(figsize=(10,4))
    sns.heatmap(ig_map.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"IG Heatmap - {test_dates[i].date()}")
    plt.tight_layout(); plt.show()
    for r in df_ig.head(10).itertuples():
        ig_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'score': r.score, 'pct': r.pct})
pd.DataFrame(ig_rows).to_csv("convtransformer_ig_summary.csv", index=False)
print("✅ Saved convtransformer_ig_summary.csv")

# ============================================================
# 2️⃣ Saliency (Grad × Input)
# ============================================================
print("\n--- Saliency ---")

class SalWrapper(torch.nn.Module):
    def __init__(self, base, target_idx=1):
        super().__init__()
        self.base = base
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.base(x)
        return preds[:, self.target_idx]

def compute_saliency(sample_idx):
    wrapper = SalWrapper(model, target_idx=1).to(DEVICE)
    sal = Saliency(wrapper)
    inp = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).float().to(DEVICE)
    inp.requires_grad = True
    attr = sal.attribute(inp)
    attr_np = attr.squeeze(0).detach().cpu().numpy()
    grad_x_input = attr_np * test_inputs[sample_idx]
    feat_score = np.sum(np.abs(grad_x_input), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'score': feat_score})
    df['pct'] = 100 * df['score'] / df['score'].sum()
    df = df.sort_values('score', ascending=False)
    return df, grad_x_input

sal_rows = []
for i in range(IG_SAMPLES):
    df_sal, sal_map = compute_saliency(i)
    print(f"\nSample {i} ({test_dates[i].date()}) top Saliency features:")
    display(df_sal.head(8))
    plt.figure(figsize=(10,4))
    sns.heatmap(sal_map.T, cmap="bwr", center=0, yticklabels=FEATURE_COLS)
    plt.title(f"Saliency Heatmap - {test_dates[i].date()}")
    plt.tight_layout(); plt.show()
    for r in df_sal.head(10).itertuples():
        sal_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'score': r.score, 'pct': r.pct})
pd.DataFrame(sal_rows).to_csv("convtransformer_saliency_summary.csv", index=False)
print("✅ Saved convtransformer_saliency_summary.csv")

# ============================================================
# 3️⃣ Permutation Feature Importance (Global)
# ============================================================
print("\n--- Permutation Feature Importance ---")

def permutation_importance(X_np, y_true_unscaled, features, target_idx=1, n_repeats=3):
    preds_base = predict_unscaled(X_np)
    base_rmse = math.sqrt(mean_squared_error(y_true_unscaled[:, target_idx], preds_base[:, target_idx]))
    results = []
    for i, f in enumerate(features):
        deltas = []
        for _ in range(n_repeats):
            Xp = X_np.copy()
            np.random.shuffle(Xp[:,:,i])
            preds_perm = predict_unscaled(Xp)
            rmse = math.sqrt(mean_squared_error(y_true_unscaled[:, target_idx], preds_perm[:, target_idx]))
            deltas.append(rmse - base_rmse)
        results.append((f, np.mean(deltas)))
    df = pd.DataFrame(results, columns=['feature','delta_RMSE']).sort_values('delta_RMSE', ascending=False)
    return df

perm_df = permutation_importance(test_inputs, targets_inv, FEATURE_COLS)
display(perm_df.head(12))
perm_df.to_csv("convtransformer_permutation_importance.csv", index=False)
print("✅ Saved convtransformer_permutation_importance.csv")

plt.figure(figsize=(8,6))
sns.barplot(y="feature", x="delta_RMSE", data=perm_df.head(12), palette="viridis")
plt.title("Permutation Feature Importance (Close)")
plt.tight_layout(); plt.show()

# ============================================================
# 4️⃣ SHAP (KernelExplainer)
# ============================================================
print("\n--- SHAP Analysis ---")

SHAP_BG = min(50, len(test_inputs))
SHAP_EXPLAIN_N = min(5, len(test_inputs))
SHAP_NSAMPLES = 200

X_avg = np.mean(test_inputs, axis=1)
background = X_avg[np.random.choice(len(X_avg), SHAP_BG, replace=False)]

def predict_flat(x_flat):
    seq = np.repeat(x_flat[:, np.newaxis, :], SEQ_LEN, axis=1).astype(np.float32)
    preds = predict_unscaled(seq)
    return preds[:, 1]  # Close prediction

explainer = shap.KernelExplainer(predict_flat, background)
to_explain = X_avg[:SHAP_EXPLAIN_N]
shap_vals = explainer.shap_values(to_explain, nsamples=SHAP_NSAMPLES)

shap_rows = []
for i, sv in enumerate(shap_vals):
    df_sh = pd.DataFrame({'feature': FEATURE_COLS, 'shap': sv})
    df_sh['abs'] = np.abs(df_sh['shap'])
    df_sh = df_sh.sort_values('abs', ascending=False)
    print(f"\nSample {i} ({test_dates[i].date()}) SHAP top features:")
    display(df_sh.head(8))
    plt.figure(figsize=(8,4))
    sns.barplot(x='shap', y='feature', data=df_sh.head(10), palette="coolwarm")
    plt.title(f"SHAP Feature Contributions ({test_dates[i].date()})")
    plt.tight_layout(); plt.show()
    for r in df_sh.head(10).itertuples():
        shap_rows.append({'sample_idx': i, 'date': str(test_dates[i].date()), 'feature': r.feature, 'shap': r.shap, 'abs': r.abs})
pd.DataFrame(shap_rows).to_csv("convtransformer_shap_summary.csv", index=False)
print("✅ Saved convtransformer_shap_summary.csv")

# ============================================================
# ✅ Final Summary
# ============================================================
print("\n✅ Explainability complete for ConvTransformer.")
print("Files generated:")
print("- convtransformer_ig_summary.csv")
print("- convtransformer_saliency_summary.csv")
print("- convtransformer_permutation_importance.csv")
print("- convtransformer_shap_summary.csv")
```

```python
import os
import random
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
from captum.attr import IntegratedGradients, Saliency
import shap

# ============================================================
# ⚙️ Config / Device
# ============================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

CSV_PATH = "reliance_with_indicators.csv"
TRAIN_START = "2019-01-01"
TRAIN_END = "2024-12-31"
TEST_START = "2025-01-01"
SEQ_LEN = 60
BATCH_SIZE = 48
EPOCHS = 60
LR = 1e-4
D_MODEL = 128
N_HEADS = 4
ENC_LAYERS = 3
FFN_DIM = 256
DROPOUT = 0.1
PATIENCE = 8
SAVE_PATH = "convtransformer_reliance_2025.pth"

# ============================================================
# 📥 Load & Preprocess
# ============================================================
df = pd.read_csv(CSV_PATH, index_col=0, parse_dates=True)
df.sort_index(inplace=True)
df = df.ffill().bfill()

TARGET_COLS = ["Open", "Close"]
FEATURE_COLS = df.columns.tolist()

train_df = df.loc[(df.index >= TRAIN_START) & (df.index <= TRAIN_END)].copy()
test_df = df.loc[df.index >= TEST_START].copy()

print(f"Train: {len(train_df)} rows ({train_df.index.min()} → {train_df.index.max()})")
print(f"Test : {len(test_df)} rows ({test_df.index.min()} → {test_df.index.max()})")

# Scale data
scaler = MinMaxScaler()
scaler.fit(train_df[FEATURE_COLS].values)
def scale_df(a_df): return pd.DataFrame(scaler.transform(a_df[FEATURE_COLS]), index=a_df.index, columns=FEATURE_COLS)
train_scaled, test_scaled = scale_df(train_df), scale_df(test_df)

# ============================================================
# 🧩 Dataset
# ============================================================
class TimeSeriesSeqDataset(Dataset):
    def __init__(self, df_scaled, seq_len, feature_cols, target_cols):
        self.X = df_scaled.values.astype(np.float32)
        self.feature_cols = feature_cols
        self.target_cols = target_cols
        self.seq_len = seq_len
        self.indices = [i for i in range(seq_len - 1, len(self.X) - 1)]
    def __len__(self): return len(self.indices)
    def __getitem__(self, idx):
        i = self.indices[idx]
        seq = self.X[i - (self.seq_len - 1): i + 1]
        target_row = self.X[i + 1]
        target = np.array([target_row[self.feature_cols.index(c)] for c in self.target_cols], dtype=np.float32)
        return torch.from_numpy(seq), torch.from_numpy(target)

train_ds = TimeSeriesSeqDataset(train_scaled, SEQ_LEN, FEATURE_COLS, TARGET_COLS)
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)

# ============================================================
# 🧠 ConvTransformer Model
# ============================================================
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=2000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        pos = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2], pe[:, 1::2] = torch.sin(pos * div), torch.cos(pos * div)
        pe = pe.unsqueeze(0)
        self.register_buffer("pe", pe)
    def forward(self, x): return x + self.pe[:, :x.size(1), :]

class MultiHeadAttentionWithWeights(nn.Module):
    def __init__(self, d_model, n_heads, dropout=0.0):
        super().__init__()
        self.mha = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
    def forward(self, q, k, v):
        out, w = self.mha(q, k, v, need_weights=True)
        return out, w

class TransformerEncoderLayerWithWeights(nn.Module):
    def __init__(self, d_model, n_heads, ffn_dim, dropout):
        super().__init__()
        self.mha = MultiHeadAttentionWithWeights(d_model, n_heads, dropout)
        self.norm1, self.norm2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(ffn_dim, d_model), nn.Dropout(dropout)
        )
    def forward(self, x):
        attn_out, attn_w = self.mha(x, x, x)
        x = self.norm1(x + attn_out)
        x = self.norm2(x + self.ffn(x))
        return x, attn_w

class ConvTransformer(nn.Module):
    def __init__(self, n_features, conv_channels=[64,96], kernel_size=3,
                 d_model=128, n_heads=4, n_layers=3, ffn_dim=256,
                 dropout=0.1, target_size=2):
        super().__init__()
        self.input_proj = nn.Conv1d(in_channels=n_features, out_channels=conv_channels[0], kernel_size=1)
        convs, in_ch = [], conv_channels[0]
        for out_ch in conv_channels:
            convs += [nn.Conv1d(in_ch, out_ch, kernel_size, padding=kernel_size//2), nn.ReLU(), nn.BatchNorm1d(out_ch)]
            in_ch = out_ch
        self.conv_stack = nn.Sequential(*convs)
        self.conv_to_model = nn.Conv1d(in_ch, d_model, kernel_size=1)
        self.pos_enc = PositionalEncoding(d_model)
        self.enc_layers = nn.ModuleList([TransformerEncoderLayerWithWeights(d_model, n_heads, ffn_dim, dropout) for _ in range(n_layers)])
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(nn.Linear(d_model, d_model//2), nn.ReLU(), nn.Linear(d_model//2, target_size))
    def forward(self, src):
        x = self.input_proj(src.permute(0,2,1))
        x = self.conv_stack(x)
        x = self.conv_to_model(x).permute(0,2,1)
        x = self.pos_enc(x)
        attns = []
        for layer in self.enc_layers:
            x, w = layer(x); attns.append(w.detach().cpu().numpy())
        pooled = self.pool(x.permute(0,2,1)).squeeze(-1)
        preds = self.fc(pooled)
        return preds, attns

model = ConvTransformer(len(FEATURE_COLS), d_model=D_MODEL, n_heads=N_HEADS,
                        n_layers=ENC_LAYERS, ffn_dim=FFN_DIM, dropout=DROPOUT,
                        target_size=len(TARGET_COLS)).to(DEVICE)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=3)

# ============================================================
# 🏋️‍♂️ Train Model
# ============================================================
best_val = 1e9; counter = 0
for epoch in range(1, EPOCHS+1):
    model.train()
    losses = []
    for xb, yb in train_loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        optimizer.zero_grad()
        preds, _ = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        losses.append(loss.item())
    avg_loss = np.mean(losses)
    scheduler.step(avg_loss)
    print(f"Epoch {epoch:03d} | Train Loss: {avg_loss:.6f}")
    if avg_loss < best_val - 1e-6:
        best_val, counter = avg_loss, 0
        torch.save({
            "model_state_dict": model.state_dict(),
            "scaler": scaler,
            "feature_cols": FEATURE_COLS,
            "target_cols": TARGET_COLS
        }, SAVE_PATH)
        print("  -> Saved best model.")
    else:
        counter += 1
        if counter >= PATIENCE:
            print("Early stopping.")
            break

# ============================================================
# 🔍 Safe checkpoint loading (PyTorch 2.6+ fix)
# ============================================================
import torch.serialization
from sklearn.preprocessing import MinMaxScaler
torch.serialization.add_safe_globals([MinMaxScaler])

checkpoint = torch.load(SAVE_PATH, map_location=DEVICE, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
scaler = checkpoint["scaler"]

# ============================================================
# 🔮 Evaluation on 2025–Present
# ============================================================
def inverse_targets(scaled_arr):
    if scaled_arr.ndim == 1: scaled_arr = scaled_arr.reshape(1,-1)
    filler = np.zeros((scaled_arr.shape[0], len(FEATURE_COLS)))
    for i, c in enumerate(TARGET_COLS):
        filler[:, FEATURE_COLS.index(c)] = scaled_arr[:, i]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

full_scaled = pd.DataFrame(scaler.transform(df[FEATURE_COLS]), index=df.index, columns=FEATURE_COLS)
test_inputs, test_targets, test_dates = [], [], []
for idx in test_df.index:
    loc = df.index.get_loc(idx)
    seq_start = loc - SEQ_LEN
    if seq_start < 0: continue
    seq = full_scaled.iloc[seq_start:loc].values.astype(np.float32)
    target_row = full_scaled.iloc[loc].values.astype(np.float32)
    target = np.array([target_row[FEATURE_COLS.index(c)] for c in TARGET_COLS], dtype=np.float32)
    test_inputs.append(seq); test_targets.append(target); test_dates.append(idx)
test_inputs, test_targets = np.stack(test_inputs), np.stack(test_targets)

model.eval()
with torch.no_grad():
    preds_scaled, _ = model(torch.from_numpy(test_inputs).to(DEVICE))
    preds_scaled = preds_scaled.cpu().numpy()

preds_inv, targets_inv = inverse_targets(preds_scaled), inverse_targets(test_targets)
rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
mae_close = mean_absolute_error(targets_inv[:,1], preds_inv[:,1])
print(f"\n2025 Test RMSE (Close): {rmse_close:.4f} | MAE: {mae_close:.4f}")

plt.figure(figsize=(12,5))
plt.plot(test_dates, targets_inv[:,1], label="Actual Close")
plt.plot(test_dates, preds_inv[:,1], label="Pred Close", linestyle="--")
plt.title("2025 Actual vs Predicted Close Prices (ConvTransformer)")
plt.legend(); plt.xticks(rotation=45); plt.show()
```

```python
# ---------------------------
# XAI: IG, Saliency, Permutation, SHAP
# ---------------------------
import os
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
import torch
from captum.attr import IntegratedGradients, Saliency
import shap
from sklearn.metrics import mean_squared_error

print("\nStarting XAI analysis...")

# small helpers (uses existing FEATURE_COLS, TARGET_COLS, scaler, model, DEVICE, test_inputs, test_targets, test_dates)
def inverse_targets_local(scaled_arr):
    """Inverse transform scaled Open/Close predictions or targets."""
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    # scaled_arr columns are [Open_scaled, Close_scaled] in order of TARGET_COLS
    for j, col in enumerate(TARGET_COLS):
        filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, j]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

def predict_unscaled_from_seq(X_np):
    """X_np shape: (N, seq_len, n_features). Returns inverse-scaled preds (N, 2)."""
    model.eval()
    with torch.no_grad():
        x_t = torch.from_numpy(X_np.astype(np.float32)).to(DEVICE)
        preds_scaled, _ = model(x_t)
        preds_scaled = preds_scaled.cpu().numpy()
    return inverse_targets_local(preds_scaled)

# Basic check: base RMSE (Close)
preds_inv = predict_unscaled_from_seq(test_inputs)
targets_inv = inverse_targets_local(test_targets)
base_rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
print(f"Base RMSE (Close) on test set: {base_rmse_close:.4f}")

# ---------------------------
# 1) Integrated Gradients (local)
# ---------------------------
print("\nRunning Integrated Gradients (IG) for first few test samples...")

class IGWrapper(torch.nn.Module):
    def __init__(self, model, target_idx=1):
        super().__init__()
        self.model = model
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.model(x)
        # return scalar (batch,) for the requested target index
        return preds[:, self.target_idx]

ig_wrapper = IGWrapper(model, target_idx=1).to(DEVICE)
ig_attr = IntegratedGradients(ig_wrapper)

IG_N = min(5, len(test_inputs))   # how many local samples to compute
ig_summary = []
for i in range(IG_N):
    inp = torch.from_numpy(test_inputs[i:i+1].astype(np.float32)).to(DEVICE)
    baseline = torch.zeros_like(inp).to(DEVICE)   # zero baseline (simple)
    atts, delta = ig_attr.attribute(inputs=inp, baselines=baseline, n_steps=50, return_convergence_delta=True)
    atts_np = atts.squeeze(0).cpu().numpy()      # (seq_len, n_features)
    # Summarize by summing absolute attribution across time per feature
    feat_scores = np.sum(np.abs(atts_np), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'ig_score': feat_scores})
    df['ig_pct'] = 100 * df['ig_score'] / df['ig_score'].sum()
    df = df.sort_values('ig_score', ascending=False).reset_index(drop=True)
    print(f"\nIG Top features for sample {i} (date {test_dates[i].date()}):")
    display(df.head(10))
    # heatmap of IG (features x time)
    plt.figure(figsize=(10,4))
    sns.heatmap(atts_np.T, cmap='bwr', center=0, yticklabels=FEATURE_COLS)
    plt.title(f"IG heatmap (sample {i} - {test_dates[i].date()})")
    plt.xlabel("time index (sequence)")
    plt.tight_layout()
    plt.show()
    # save top-k
    topk = df.head(10).copy()
    topk['sample_idx'] = i
    topk['date'] = str(test_dates[i].date())
    ig_summary.append(topk)

ig_df = pd.concat(ig_summary, ignore_index=True)
ig_df.to_csv("convtransformer_ig_local.csv", index=False)
print("Saved convtransformer_ig_local.csv")

# ---------------------------
# 2) Saliency (grad * input)
# ---------------------------
print("\nRunning Saliency (Grad × Input) for first few test samples...")
saliency_wrapper = IGWrapper(model, target_idx=1).to(DEVICE)  # same wrapper works for saliency
sal = Saliency(saliency_wrapper)

sal_summary = []
for i in range(IG_N):
    inp = torch.from_numpy(test_inputs[i:i+1].astype(np.float32)).to(DEVICE)
    inp.requires_grad = True
    grads = sal.attribute(inp)    # returns gradients shape (1, seq_len, n_features)
    grads_np = grads.squeeze(0).cpu().numpy()
    grad_x_input = grads_np * test_inputs[i]   # elementwise
    feat_scores = np.sum(np.abs(grad_x_input), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'saliency_score': feat_scores})
    df['saliency_pct'] = 100 * df['saliency_score'] / df['saliency_score'].sum()
    df = df.sort_values('saliency_score', ascending=False).reset_index(drop=True)
    print(f"\nSaliency Top features for sample {i} (date {test_dates[i].date()}):")
    display(df.head(10))
    plt.figure(figsize=(10,4))
    sns.heatmap(grad_x_input.T, cmap='bwr', center=0, yticklabels=FEATURE_COLS)
    plt.title(f"Saliency heatmap (sample {i} - {test_dates[i].date()})")
    plt.xlabel("time index (sequence)")
    plt.tight_layout()
    plt.show()
    topk = df.head(10).copy()
    topk['sample_idx'] = i
    topk['date'] = str(test_dates[i].date())
    sal_summary.append(topk)

sal_df = pd.concat(sal_summary, ignore_index=True)
sal_df.to_csv("convtransformer_saliency_local.csv", index=False)
print("Saved convtransformer_saliency_local.csv")

# ---------------------------
# 3) Permutation Feature Importance (global)
# ---------------------------
print("\nRunning Permutation Feature Importance (global). This may take a little time...")

def permutation_importance(X_np, y_unscaled, features, target_idx=1, n_repeats=5):
    """Shuffle feature sequences across samples (preserving within-sample time order)"""
    base_preds = predict_unscaled_from_seq(X_np)
    base_rmse = math.sqrt(mean_squared_error(y_unscaled[:, target_idx], base_preds[:, target_idx]))
    results = []
    N = X_np.shape[0]
    for f_i, f_name in enumerate(features):
        diffs = []
        for _ in range(n_repeats):
            Xp = X_np.copy()
            perm = np.random.permutation(N)
            # replace the entire time-series of feature f_i for each sample by that of a different sample
            Xp[:, :, f_i] = Xp[perm, :, f_i]
            preds_perm = predict_unscaled_from_seq(Xp)
            rmse = math.sqrt(mean_squared_error(y_unscaled[:, target_idx], preds_perm[:, target_idx]))
            diffs.append(rmse - base_rmse)
        results.append((f_name, np.mean(diffs), np.std(diffs)))
    df = pd.DataFrame(results, columns=['feature','delta_rmse','delta_std']).sort_values('delta_rmse', ascending=False)
    return df

perm_df = permutation_importance(test_inputs, targets_inv, FEATURE_COLS, target_idx=1, n_repeats=3)
print("\nTop features by permutation importance (Close):")
display(perm_df.head(12))
perm_df.to_csv("convtransformer_permutation_importance.csv", index=False)
print("Saved convtransformer_permutation_importance.csv")
plt.figure(figsize=(8,6))
sns.barplot(x="delta_rmse", y="feature", data=perm_df.head(12))
plt.title("Permutation importance (delta RMSE) - Close")
plt.tight_layout()
plt.show()

# ---------------------------
# 4) SHAP (KernelExplainer) — flattened features
# ---------------------------
print("\nRunning SHAP (KernelExplainer) with flattened-per-sample representation (mean over time).")
# KernelExplainer expects 2D input. We collapse each sequence to a single vector per sample.
# There are many choices; here we use per-feature mean across the sequence as input to SHAP.
X_flat = test_inputs.mean(axis=1)   # shape (N, n_features) - per-sample mean across time
y_true_unscaled = targets_inv      # actual targets unscaled

SHAP_BG = min(40, len(X_flat))
SHAP_EXPLAIN_N = min(6, len(X_flat))
SHAP_NSAMPLES = 200   # keep moderate to control runtime

# background: random small set of rows from X_flat
bg_idx = np.random.choice(len(X_flat), SHAP_BG, replace=False)
background = X_flat[bg_idx]

def predict_shap_flat(x2d):
    # x2d shape (m, n_features) where each row is per-feature summary (mean)
    # we must convert back to sequence shape expected by model: repeat the mean across SEQ_LEN
    seq_like = np.repeat(x2d[:, np.newaxis, :], SEQ_LEN, axis=1).astype(np.float32)
    preds = predict_unscaled_from_seq(seq_like)   # (m, 2)
    return preds[:, 1]   # return Close only

explainer = shap.KernelExplainer(predict_shap_flat, background)
to_explain = X_flat[:SHAP_EXPLAIN_N]
shap_vals = explainer.shap_values(to_explain, nsamples=SHAP_NSAMPLES)

# shap_vals is a list or array of shape (m,n_features) depending on shap version
shap_rows = []
for i in range(len(to_explain)):
    vals = shap_vals[i] if isinstance(shap_vals, list) else shap_vals[i]
    df_sh = pd.DataFrame({'feature': FEATURE_COLS, 'shap': vals})
    df_sh['abs_shap'] = np.abs(df_sh['shap'])
    df_sh = df_sh.sort_values('abs_shap', ascending=False).reset_index(drop=True)
    print(f"\nSHAP top features for sample {i} (date {test_dates[i].date()}):")
    display(df_sh.head(10))
    plt.figure(figsize=(8,4))
    sns.barplot(x='shap', y='feature', data=df_sh.head(10))
    plt.title(f"SHAP contributions (Close) - sample {i}")
    plt.tight_layout()
    plt.show()
    topk = df_sh.head(10).copy()
    topk['sample_idx'] = i
    topk['date'] = str(test_dates[i].date())
    shap_rows.append(topk)
shap_out = pd.concat(shap_rows, ignore_index=True)
shap_out.to_csv("convtransformer_shap_local.csv", index=False)
print("Saved convtransformer_shap_local.csv")

# ---------------------------
# Final summary prints
# ---------------------------
print("\nXAI complete — files written to working directory:")
print("- convtransformer_ig_local.csv")
print("- convtransformer_saliency_local.csv")
print("- convtransformer_permutation_importance.csv")
print("- convtransformer_shap_local.csv")

# Quick top-feature summary table (aggregate)
top_ig = ig_df.groupby('feature')['ig_pct'].mean().sort_values(ascending=False).head(10)
top_sal = sal_df.groupby('feature')['saliency_pct'].mean().sort_values(ascending=False).head(10)
top_perm = perm_df.head(10).set_index('feature')['delta_rmse']
top_shap = shap_out.groupby('feature')['abs_shap'].mean().sort_values(ascending=False).head(10)

print("\nAggregate top features (IG avg %, Saliency avg %, Perm delta RMSE, SHAP avg abs):\n")
agg = pd.DataFrame({
    'IG_pct': top_ig,
    'Saliency_pct': top_sal,
}).fillna(0)
agg['Perm_delta_RMSE'] = top_perm
agg['SHAP_abs_mean'] = top_shap
display(agg.fillna(0))

print("\nDone. Inspect CSVs and plots. If you want, I can also produce human-readable one-sentence explanations for each test day using SHAP+IG.")
```

```python
# tcn_attention_reliance.py
# Temporal Convolutional Network (TCN) + Attention for Reliance forecasting
# Explainability via Captum Integrated Gradients (IG)
# Dependencies: torch, numpy, pandas, scikit-learn, matplotlib, seaborn, captum, pandas_ta (preprocessing)
# Put reliance_with_indicators.csv in same folder (from your preprocessing)

import os
import random
import math
from datetime import datetime
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
from captum.attr import IntegratedGradients

# ---------------------------
# Repro / device
# ---------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

# ---------------------------
# Config / Hyperparams
# ---------------------------
CSV_PATH = "reliance_with_indicators.csv"
START_DATE = "2019-01-01"
SEQ_LEN = 60          # input window length (days)
BATCH_SIZE = 48
EPOCHS = 60
LR = 1e-4
CHANNELS = [32, 64, 96]   # channels per TCN stage (lower memory footprint)
KERNEL_SIZE = 3
N_LAYERS = len(CHANNELS)  # number of TCN blocks
D_MODEL = 128         # dimension for attention projection
N_HEADS = 4
FFN_DIM = 256
DROPOUT = 0.1
TEST_DAYS = 30
PATIENCE = 8
SAVE_PATH = "tcn_attention_reliance.pth"

# ---------------------------
# Load & preprocess
# ---------------------------
df = pd.read_csv(CSV_PATH, index_col=0, parse_dates=True)
df.sort_index(inplace=True)

# Get timezone from the DataFrame index
df_timezone = df.index.tz

# Convert START_DATE string to datetime and localize to the DataFrame's timezone
start_date_aware = pd.to_datetime(START_DATE).tz_localize(df_timezone)

# Filter the DataFrame using the timezone-aware start date
df = df[df.index >= start_date_aware]


df = df.ffill().bfill()
df.dropna(inplace=True)
print("Data range:", df.index.min(), "to", df.index.max())
print("Columns:", df.columns.tolist())

TARGET_COLS = ["Open", "Close"]
FEATURE_COLS = df.columns.tolist()
for t in TARGET_COLS:
    if t not in FEATURE_COLS:
        raise ValueError(f"Target {t} not in columns.")

# train/val/test split
if TEST_DAYS >= len(df):
    raise ValueError("TEST_DAYS too large.")
train_val_df = df.iloc[:-TEST_DAYS]
test_df = df.iloc[-TEST_DAYS:].copy()
val_frac = 0.1
val_size = int(len(train_val_df) * val_frac)
train_df = train_val_df.iloc[:-val_size].copy()
val_df = train_val_df.iloc[-val_size:].copy()

# scaler
scaler = MinMaxScaler()
scaler.fit(train_df[FEATURE_COLS].values)

def scale_df(a_df):
    arr = scaler.transform(a_df[FEATURE_COLS].values)
    return pd.DataFrame(arr, index=a_df.index, columns=FEATURE_COLS)

train_scaled = scale_df(train_df)
val_scaled = scale_df(val_df)
test_scaled = scale_df(test_df)

# ---------------------------
# Dataset & DataLoader
# ---------------------------
class TimeSeriesSeqDataset(Dataset):
    def __init__(self, df_scaled, seq_len=SEQ_LEN, feature_cols=FEATURE_COLS, target_cols=TARGET_COLS):
        self.X = df_scaled.values.astype(np.float32)
        self.feature_cols = feature_cols
        self.target_cols = target_cols
        self.seq_len = seq_len
        self.indices = []
        for i in range(seq_len - 1, len(self.X) - 1):
            self.indices.append(i)
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, idx):
        i = self.indices[idx]
        seq = self.X[i - (self.seq_len - 1): i + 1]  # (seq_len, n_features)
        target_row = self.X[i + 1]
        target = np.array([target_row[self.feature_cols.index(c)] for c in self.target_cols], dtype=np.float32)
        return torch.from_numpy(seq), torch.from_numpy(target)

train_ds = TimeSeriesSeqDataset(train_scaled)
val_ds = TimeSeriesSeqDataset(val_scaled)
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, drop_last=False)
print("Train samples:", len(train_ds), "Val samples:", len(val_ds))

# ---------------------------
# TCN building blocks (causal dilated conv, residual)
# ---------------------------
class Chomp1d(nn.Module):
    """Remove padding on the right to preserve causality."""
    def __init__(self, chomp_size):
        super().__init__()
        self.chomp_size = chomp_size
    def forward(self, x):
        # x shape (batch, channels, seq_len + padding)
        if self.chomp_size == 0:
            return x
        return x[:, :, :-self.chomp_size].contiguous()

class TemporalBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dilation, padding, dropout):
        super().__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, stride=1, padding=padding, dilation=dilation)
        self.chomp1 = Chomp1d(padding)
        self.relu1 = nn.ReLU()
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.dropout1 = nn.Dropout(dropout)

        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, stride=1, padding=padding, dilation=dilation)
        self.chomp2 = Chomp1d(padding)
        self.relu2 = nn.ReLU()
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.dropout2 = nn.Dropout(dropout)

        self.downsample = nn.Conv1d(in_channels, out_channels, 1) if in_channels != out_channels else None
        self.relu = nn.ReLU()

    def forward(self, x):
        # x: (batch, channels, seq_len)
        out = self.conv1(x)
        out = self.chomp1(out)
        out = self.relu1(out)
        out = self.bn1(out)
        out = self.dropout1(out)

        out = self.conv2(out)
        out = self.chomp2(out)
        out = self.relu2(out)
        out = self.bn2(out)
        out = self.dropout2(out)

        res = x if self.downsample is None else self.downsample(x)
        return self.relu(out + res)

class TCNEncoder(nn.Module):
    def __init__(self, num_inputs, channels, kernel_size=2, dropout=0.2):
        super().__init__()
        layers = []
        num_levels = len(channels)
        in_ch = num_inputs
        for i in range(num_levels):
            out_ch = channels[i]
            dilation_size = 2 ** i
            padding = (kernel_size - 1) * dilation_size
            layers += [TemporalBlock(in_ch, out_ch, kernel_size, dilation=dilation_size, padding=padding, dropout=dropout)]
            in_ch = out_ch
        self.network = nn.Sequential(*layers)

    def forward(self, x):
        # x: (batch, seq_len, n_features) => convert to (batch, channels, seq_len)
        x = x.permute(0, 2, 1)
        out = self.network(x)  # (batch, out_ch, seq_len)
        out = out.permute(0, 2, 1)  # (batch, seq_len, out_ch)
        return out

# ---------------------------
# Model: TCN -> Attention -> MLP
# ---------------------------
class TCNAttentionModel(nn.Module):
    def __init__(self, n_features, channels, kernel_size=3, d_model=128, n_heads=4, n_layers=1, ffn_dim=256, dropout=0.1, target_size=2):
        super().__init__()
        self.n_features = n_features
        self.tcn = TCNEncoder(num_inputs=n_features, channels=channels, kernel_size=kernel_size, dropout=dropout)
        last_ch = channels[-1]
        # project TCN channels -> d_model for attention
        self.proj = nn.Linear(last_ch, d_model)
        self.pos_enc = nn.Parameter(self._build_pos_enc(d_model, 2000), requires_grad=False)
        # attention
        self.attn = nn.MultiheadAttention(embed_dim=d_model, num_heads=n_heads, batch_first=True)
        # feed-forward after attention
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(ffn_dim, d_model),
            nn.Dropout(dropout)
        )
        # pooling and output
        self.pool = nn.AdaptiveAvgPool1d(1)
        self.fc = nn.Sequential(
            nn.Linear(d_model, d_model//2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(d_model//2, target_size)
        )

    def _build_pos_enc(self, d_model, max_len=2000):
        pe = torch.zeros(max_len, d_model)
        pos = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1)
        div = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(pos * div)
        pe[:, 1::2] = torch.cos(pos * div)
        pe = pe.unsqueeze(0)  # (1, max_len, d_model)
        return pe  # not registered as buffer, but as tensor param (non-trainable)

    def forward(self, x):
        """
        x: (batch, seq_len, n_features)
        returns preds (batch, target_size), attn_weights (batch, seq_len, seq_len)
        """
        b, seq_len, nfeat = x.size()
        tcn_out = self.tcn(x)  # (batch, seq_len, last_ch)
        proj = self.proj(tcn_out)  # (batch, seq_len, d_model)
        # add positional enc (slice to seq_len)
        pos = self.pos_enc[:, :seq_len, :].to(proj.device)
        proj = proj + pos
        # self-attention (query=key=value=proj)
        attn_out, attn_w = self.attn(proj, proj, proj, need_weights=True)  # attn_w: (batch, seq_len, seq_len)
        # residual + ffn
        x = proj + attn_out
        x = x + self.ffn(x)
        # pool across time: convert (batch, seq_len, d_model) -> (batch, d_model, 1)
        pooled = self.pool(x.permute(0,2,1)).squeeze(-1)
        preds = self.fc(pooled)
        return preds, attn_w

# instantiate
n_features = len(FEATURE_COLS)
model = TCNAttentionModel(n_features=n_features, channels=CHANNELS, kernel_size=KERNEL_SIZE,
                          d_model=D_MODEL, n_heads=N_HEADS, n_layers=1, ffn_dim=FFN_DIM,
                          dropout=DROPOUT, target_size=len(TARGET_COLS)).to(DEVICE)
print(model)

# ---------------------------
# Training utils
# ---------------------------
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-6)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)

def train_one_epoch():
    model.train()
    losses = []
    for xb, yb in train_loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)
        optimizer.zero_grad()
        preds, _ = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        losses.append(loss.item())
    return np.mean(losses)

def eval_on_loader(loader):
    model.eval()
    losses = []
    preds_all = []
    trues_all = []
    attn_examples = None
    with torch.no_grad():
        for xb, yb in loader:
            xb = xb.to(DEVICE)
            yb = yb.to(DEVICE)
            preds, attn_w = model(xb)
            loss = criterion(preds, yb)
            losses.append(loss.item())
            preds_all.append(preds.cpu().numpy())
            trues_all.append(yb.cpu().numpy())
            if attn_examples is None:
                attn_examples = attn_w.detach().cpu().numpy()
    if len(preds_all):
        preds_all = np.vstack(preds_all)
        trues_all = np.vstack(trues_all)
    else:
        preds_all = np.zeros((0, len(TARGET_COLS)))
        trues_all = np.zeros((0, len(TARGET_COLS)))
    return np.mean(losses), preds_all, trues_all, attn_examples

# ---------------------------
# Train loop with early stopping
# ---------------------------
best_val = 1e9
counter = 0

for epoch in range(1, EPOCHS + 1):
    train_loss = train_one_epoch()
    val_loss, _, _, _ = eval_on_loader(val_loader)
    scheduler.step(val_loss)
    print(f"Epoch {epoch:03d} | Train Loss {train_loss:.6f} | Val Loss {val_loss:.6f}")
    if val_loss < best_val - 1e-6:
        best_val = val_loss
        counter = 0
        torch.save({
            "model_state_dict": model.state_dict(),
            "scaler": scaler,
            "feature_cols": FEATURE_COLS,
            "target_cols": TARGET_COLS
        }, SAVE_PATH)
        print("  -> Saved best model.")
    else:
        counter += 1
        if counter >= PATIENCE:
            print("Early stopping triggered.")
            break

# load best
checkpoint = torch.load(SAVE_PATH, map_location=DEVICE, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
scaler = checkpoint["scaler"]

# ---------------------------
# Build test sequences
# ---------------------------
full_scaled = pd.DataFrame(scaler.transform(df[FEATURE_COLS].values), index=df.index, columns=FEATURE_COLS)

test_inputs = []
test_targets = []
test_dates = []
for i in range(len(df) - TEST_DAYS, len(df)):
    seq_end = i - 1
    seq_start = seq_end - (SEQ_LEN - 1)
    if seq_start < 0:
        continue
    seq = full_scaled.iloc[seq_start: seq_end + 1].values.astype(np.float32)
    target_row = full_scaled.iloc[i].values.astype(np.float32)
    target = np.array([target_row[FEATURE_COLS.index(c)] for c in TARGET_COLS], dtype=np.float32)
    test_inputs.append(seq)
    test_targets.append(target)
    test_dates.append(df.index[i])

test_inputs = np.stack(test_inputs)
test_targets = np.stack(test_targets)

# run model on test inputs
model.eval()
with torch.no_grad():
    X_test_t = torch.from_numpy(test_inputs).to(DEVICE)
    preds_scaled, attn_test = model(X_test_t)
    preds_scaled = preds_scaled.cpu().numpy()

# inverse transform helper (same as earlier)
def inverse_targets(scaled_arr):
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for i, col in enumerate(FEATURE_COLS):
        if col in TARGET_COLS:
            filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, TARGET_COLS.index(col)]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

preds_inv = inverse_targets(preds_scaled)
targets_inv = inverse_targets(test_targets)

rmse_open = math.sqrt(mean_squared_error(targets_inv[:,0], preds_inv[:,0]))
rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
mae_open = mean_absolute_error(targets_inv[:,0], preds_inv[:,0])
mae_close = mean_absolute_error(targets_inv[:,1], preds_inv[:,1])

print(f"Test RMSE Open: {rmse_open:.4f} | Close: {rmse_close:.4f}")
print(f"Test MAE  Open: {mae_open:.4f} | Close: {mae_close:.4f}")

# plot actual vs predicted close
plt.figure(figsize=(12,5))
plt.plot(test_dates, targets_inv[:,1], label="Actual Close")
plt.plot(test_dates, preds_inv[:,1], label="Pred Close", linestyle="--")
plt.title("Test: Close price - Actual vs Predicted (TCN-Attention)")
plt.xticks(rotation=45)
plt.legend()
plt.show()

# ---------------------------
# Attention visualization (first test sample)
# ---------------------------
if attn_test is not None:
    # attn_test shape: (batch, seq_len, seq_len)
    sample_attn = attn_test[0]
    plt.figure(figsize=(8,6))
    sns.heatmap(sample_attn.cpu().numpy(), cmap="viridis")
    plt.title("Self-Attention weights (first test batch sample)")
    plt.xlabel("Source time index")
    plt.ylabel("Target time index")
    plt.show()
else:
    print("No attention recorded.")

# ---------------------------
# Captum Integrated Gradients for one test example (explain Close prediction)
# ---------------------------
class WrapperClose(nn.Module):
    def __init__(self, model, target_idx=1):
        super().__init__()
        self.model = model
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.model(x)
        return preds[:, self.target_idx]

wrapper = WrapperClose(model, target_idx=1).to(DEVICE)
ig = IntegratedGradients(wrapper)

sample_idx = 0
input_tensor = torch.from_numpy(test_inputs[sample_idx:sample_idx+1]).to(DEVICE)
input_tensor.requires_grad = True
baseline = torch.zeros_like(input_tensor).to(DEVICE)

attrs, delta = ig.attribute(inputs=input_tensor, baselines=baseline, return_convergence_delta=True)
attrs_np = attrs.detach().cpu().numpy()[0]  # (seq_len, n_features)

# summarize attributions per feature
feat_imp = np.sum(np.abs(attrs_np), axis=0)
feat_ranking = sorted(zip(FEATURE_COLS, feat_imp), key=lambda x: x[1], reverse=True)
print("Top feature importances (IG, summed over time) for sample 0 (Close prediction):")
for f, v in feat_ranking[:10]:
    print(f"{f}: {v:.6f}")

# plot top 5 feature attributions over time
top_feats = [f for f, _ in feat_ranking[:5]]
top_idx = [FEATURE_COLS.index(f) for f in top_feats]
plt.figure(figsize=(12,6))
for fi in top_idx:
    # Get the dates for the current sample's input sequence
    sample_start_date = df.index[len(df) - TEST_DAYS + sample_idx - SEQ_LEN + 1]
    sample_end_date = df.index[len(df) - TEST_DAYS + sample_idx]
    ig_dates = df.loc[sample_start_date:sample_end_date].index

    plt.plot(ig_dates, attrs_np[:, fi], label=FEATURE_COLS[fi])
plt.legend()
plt.title("IG attributions over time (sample 0) for top 5 features")
plt.xticks(rotation=45)
plt.show()

# ---------------------------
# Save final artifacts
# ---------------------------
torch.save({
    "model_state_dict": model.state_dict(),
    "scaler": scaler,
    "feature_cols": FEATURE_COLS,
    "target_cols": TARGET_COLS
}, "tcn_attention_reliance_final.pth")

print("Done. Model saved to tcn_attention_reliance_final.pth")
```

```python
# ---------------------------
# XAI (IG, Saliency, Permutation, SHAP)
# Paste and run this immediately AFTER your tcn_attention_reliance.py script
# ---------------------------
import os
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
import torch
from captum.attr import IntegratedGradients, Saliency
import shap
from sklearn.metrics import mean_squared_error

print("\nStarting XAI analysis for TCN-Attention model...")

# ---------- helpers (use existing variables from your script) ----------
# expected to be defined already: model, scaler, FEATURE_COLS, TARGET_COLS, SEQ_LEN, DEVICE, test_inputs, test_targets, test_dates

def inverse_targets_scaled_to_unscaled(scaled_arr):
    """Inverse transform array of shape (N, n_targets) -> (N, n_targets) in original units."""
    if scaled_arr.ndim == 1:
        scaled_arr = scaled_arr.reshape(1, -1)
    n = scaled_arr.shape[0]
    filler = np.zeros((n, len(FEATURE_COLS)))
    for j, col in enumerate(TARGET_COLS):
        filler[:, FEATURE_COLS.index(col)] = scaled_arr[:, j]
    inv = scaler.inverse_transform(filler)
    return inv[:, [FEATURE_COLS.index(c) for c in TARGET_COLS]]

def predict_unscaled_from_seq_np(X_np):
    """X_np: (N, seq_len, n_features) scaled -> returns unscaled preds (N, len(TARGET_COLS))"""
    model.eval()
    with torch.no_grad():
        x_t = torch.from_numpy(X_np.astype(np.float32)).to(DEVICE)
        preds_scaled, _ = model(x_t)
        preds_scaled = preds_scaled.cpu().numpy()
    return inverse_targets_scaled_to_unscaled(preds_scaled)

# quick sanity (base RMSE close)
preds_inv = predict_unscaled_from_seq_np(test_inputs)
targets_inv = inverse_targets_scaled_to_unscaled(test_targets)
base_rmse_close = math.sqrt(mean_squared_error(targets_inv[:,1], preds_inv[:,1]))
print(f"Base RMSE (Close) on test set: {base_rmse_close:.4f}")

# ---------------------------
# 1) Integrated Gradients (local) - multiple samples
# ---------------------------
print("\n-> Integrated Gradients (local samples)")

class WrapperClose(torch.nn.Module):
    def __init__(self, model, target_idx=1):
        super().__init__()
        self.model = model
        self.target_idx = target_idx
    def forward(self, x):
        preds, _ = self.model(x)
        # return the scalar prediction for requested target (batch,)
        return preds[:, self.target_idx]

ig_wrapper = WrapperClose(model, target_idx=1).to(DEVICE)
ig = IntegratedGradients(ig_wrapper)

IG_SAMPLES = min(6, len(test_inputs))
ig_records = []
for si in range(IG_SAMPLES):
    inp_np = test_inputs[si:si+1].astype(np.float32)
    inp = torch.from_numpy(inp_np).to(DEVICE)
    baseline = torch.zeros_like(inp).to(DEVICE)            # zero baseline (simple)
    # compute IG (may take some seconds)
    attributions, delta = ig.attribute(inputs=inp, baselines=baseline, n_steps=50, return_convergence_delta=True)
    at_np = attributions.squeeze(0).cpu().numpy()         # (seq_len, n_features)
    # per-feature summary: sum abs across time
    feat_scores = np.sum(np.abs(at_np), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'ig_score': feat_scores})
    df['ig_pct'] = 100 * df['ig_score'] / (df['ig_score'].sum() + 1e-12)
    df = df.sort_values('ig_score', ascending=False).reset_index(drop=True)
    print(f"\nIG top-10 features for test sample {si} (date {test_dates[si].date()}):")
    display(df.head(10))
    # save heatmap of attributions (feature x time)
    plt.figure(figsize=(10,6))
    sns.heatmap(at_np.T, cmap='bwr', center=0, yticklabels=FEATURE_COLS)
    plt.title(f"IG heatmap (sample {si} - {test_dates[si].date()})")
    plt.xlabel("time index (older -> newer)")
    plt.tight_layout(); plt.show()
    # store topk
    topk = df.head(12).copy(); topk['sample_idx'] = si; topk['date'] = str(test_dates[si].date())
    ig_records.append(topk)

ig_df = pd.concat(ig_records, ignore_index=True)
ig_df.to_csv("tcn_ig_local.csv", index=False)
print("Saved tcn_ig_local.csv")

# ---------------------------
# 2) Saliency (Grad × Input) - local
# ---------------------------
print("\n-> Saliency (gradient × input)")

saliency = Saliency(ig_wrapper)   # wrapper returns scalar per sample
sal_records = []
for si in range(IG_SAMPLES):
    inp_np = test_inputs[si:si+1].astype(np.float32)
    inp = torch.from_numpy(inp_np).to(DEVICE)
    inp.requires_grad = True
    grads = saliency.attribute(inp)            # (1, seq_len, n_features)
    grads_np = grads.squeeze(0).cpu().numpy()
    # grad * input (elementwise), using original scaled input values
    grad_x_input = grads_np * test_inputs[si]
    feat_scores = np.sum(np.abs(grad_x_input), axis=0)
    df = pd.DataFrame({'feature': FEATURE_COLS, 'saliency_score': feat_scores})
    df['saliency_pct'] = 100 * df['saliency_score'] / (df['saliency_score'].sum() + 1e-12)
    df = df.sort_values('saliency_score', ascending=False).reset_index(drop=True)
    print(f"\nSaliency top-10 for sample {si} (date {test_dates[si].date()}):")
    display(df.head(10))
    plt.figure(figsize=(10,6))
    sns.heatmap(grad_x_input.T, cmap='bwr', center=0, yticklabels=FEATURE_COLS)
    plt.title(f"Saliency heatmap (sample {si} - {test_dates[si].date()})")
    plt.xlabel("time index (older -> newer)")
    plt.tight_layout(); plt.show()
    topk = df.head(12).copy(); topk['sample_idx'] = si; topk['date'] = str(test_dates[si].date())
    sal_records.append(topk)

sal_df = pd.concat(sal_records, ignore_index=True)
sal_df.to_csv("tcn_saliency_local.csv", index=False)
print("Saved tcn_saliency_local.csv")

# ---------------------------
# 3) Permutation Feature Importance (global)
# ---------------------------
print("\n-> Permutation feature importance (global). This will evaluate effect on RMSE (Close).")

def permutation_importance_sequences(X_seq, y_unscaled, features, target_idx=1, n_repeats=3):
    base_preds = predict_unscaled_from_seq_np(X_seq)
    base_rmse = math.sqrt(mean_squared_error(y_unscaled[:, target_idx], base_preds[:, target_idx]))
    print(f"Base RMSE (Close) = {base_rmse:.4f}")
    results = []
    N = X_seq.shape[0]
    for f_i, f_name in enumerate(features):
        diffs = []
        for r in range(n_repeats):
            Xp = X_seq.copy()
            perm = np.random.permutation(N)
            # replace entire time-series for feature f_i by that from shuffled samples
            Xp[:, :, f_i] = Xp[perm, :, f_i]
            preds_perm = predict_unscaled_from_seq_np(Xp)
            rmse = math.sqrt(mean_squared_error(y_unscaled[:, target_idx], preds_perm[:, target_idx]))
            diffs.append(rmse - base_rmse)
        results.append((f_name, np.mean(diffs), np.std(diffs)))
    df = pd.DataFrame(results, columns=['feature','delta_rmse_mean','delta_rmse_std']).sort_values('delta_rmse_mean', ascending=False)
    return df

perm_df = permutation_importance_sequences(test_inputs.copy(), targets_inv.copy(), FEATURE_COLS, target_idx=1, n_repeats=3)
display(perm_df.head(12))
perm_df.to_csv("tcn_permutation_importance.csv", index=False)
print("Saved tcn_permutation_importance.csv")

plt.figure(figsize=(8,6))
sns.barplot(x='delta_rmse_mean', y='feature', data=perm_df.head(12))
plt.title("Permutation importance (delta RMSE) - Close")
plt.xlabel("Increase in RMSE when feature permuted")
plt.tight_layout(); plt.show()

# ---------------------------
# 4) SHAP (KernelExplainer) - per-sample flattened representation (mean across sequence)
# ---------------------------
print("\n-> SHAP (KernelExplainer). We collapse each sequence to per-feature mean to create 2D inputs for KernelExplainer.")
# X_flat shape (N, n_features) where each row is mean(feature over seq)
X_flat = test_inputs.mean(axis=1)   # (N, n_features)
y_true_unscaled = targets_inv.copy()  # unscaled ground-truth

SHAP_BG = min(40, len(X_flat))
SHAP_N = min(6, len(X_flat))
SHAP_NSAMPLES = 200   # adjust for runtime/precision tradeoff

bg_idx = np.random.choice(len(X_flat), SHAP_BG, replace=False)
background = X_flat[bg_idx]

def shap_predict_flat(x2d):
    # x2d: (m, n_features) per-sample feature-summary
    # convert back to sequence-shaped input expected by model by repeating summary across SEQ_LEN
    seq_like = np.repeat(x2d[:, np.newaxis, :], SEQ_LEN, axis=1).astype(np.float32)
    preds = predict_unscaled_from_seq_np(seq_like)   # (m, 2)
    return preds[:, 1]   # return Close predictions only

print("Initializing KernelExplainer (may take a moment)...")
explainer = shap.KernelExplainer(shap_predict_flat, background)
to_explain = X_flat[:SHAP_N]
shap_vals = explainer.shap_values(to_explain, nsamples=SHAP_NSAMPLES)

# shap_vals format varies by SHAP version: either array (m, n_features) or list
shap_records = []
for i in range(len(to_explain)):
    vals = shap_vals[i] if isinstance(shap_vals, list) else shap_vals[i]
    df_sh = pd.DataFrame({'feature': FEATURE_COLS, 'shap': np.array(vals)})
    df_sh['abs_shap'] = df_sh['shap'].abs()
    df_sh = df_sh.sort_values('abs_shap', ascending=False).reset_index(drop=True)
    print(f"\nSHAP top features for sample {i} (date {test_dates[i].date()}):")
    display(df_sh.head(10))
    plt.figure(figsize=(8,4))
    sns.barplot(x='shap', y='feature', data=df_sh.head(10))
    plt.title(f"SHAP contributions (Close) - sample {i}")
    plt.tight_layout(); plt.show()
    topk = df_sh.head(12).copy(); topk['sample_idx'] = i; topk['date'] = str(test_dates[i].date())
    shap_records.append(topk)

shap_out = pd.concat(shap_records, ignore_index=True)
shap_out.to_csv("tcn_shap_local.csv", index=False)
print("Saved tcn_shap_local.csv")

# ---------------------------
# Aggregate summary + save
# ---------------------------
print("\nBuilding aggregate summary...")

top_ig = ig_df.groupby('feature')['ig_pct'].mean().sort_values(ascending=False)
top_sal = sal_df.groupby('feature')['saliency_pct'].mean().sort_values(ascending=False)
top_perm = perm_df.set_index('feature')['delta_rmse_mean']
top_shap = shap_out.groupby('feature')['abs_shap'].mean().sort_values(ascending=False)

agg = pd.DataFrame({
    'IG_pct': top_ig,
    'Saliency_pct': top_sal,
    'Perm_delta_RMSE': top_perm,
    'SHAP_abs_mean': top_shap
}).fillna(0).sort_values(by=['IG_pct','Saliency_pct'], ascending=False)

print("\nTop aggregated features (snippet):")
display(agg.head(15))

agg.to_csv("tcn_xai_aggregate_summary.csv")
print("Saved tcn_xai_aggregate_summary.csv")

print("\nXAI run complete. Files produced:")
print("- tcn_ig_local.csv")
print("- tcn_saliency_local.csv")
print("- tcn_permutation_importance.csv")
print("- tcn_shap_local.csv")
print("- tcn_xai_aggregate_summary.csv")
```

```python
# gan_data_pipeline_full_fixed.py
"""
Corrected end-to-end pipeline:
- Data download (yfinance) + feature engineering (pandas_ta)
- Sequence creation
- Autoencoder pretraining (EncoderRNN + GeneratorRNN)
- WGAN-GP adversarial training with two critics (RNN global + Patch CNN local)
Fixes:
- gradient_penalty uses .reshape and temporarily disables cuDNN for critic forward pass
- DataLoader uses num_workers=2 and pin_memory=True
"""

import os
import random
import math
from tqdm import tqdm
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import yfinance as yf
import pandas_ta as ta
from statsmodels.tsa.stattools import acf

# -------------------------
# Config
# -------------------------
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

TICKERS = {
    "reliance": "RELIANCE.NS",
    "nifty": "^NSEI",
    "usdinr": "USDINR=X",
    "brent": "BZ=F"
}
START_DATE = "2016-01-01"
END_DATE = None

SEQ_LEN = 48
STEP = 1
BATCH_SIZE = 64
WORKERS = 2  # set to 2 for safety in shared environments

HIDDEN_DIM = 64
LATENT_DIM = 32
CRITIC_ITERS = 5
EPOCHS_PRETRAIN = 30
EPOCHS_JOINT = 200
LR_G = 1e-4
LR_D = 1e-4
LAMBDA_GP = 10.0
FM_WEIGHT = 0.1
SPECTRAL_NORM = False
SAVE_DIR = "gan_outputs"
os.makedirs(SAVE_DIR, exist_ok=True)
PRINT_EVERY = 1

# -------------------------
# Helpers
# -------------------------
def seed_everything(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

seed_everything()

def snorm(m):
    return nn.utils.spectral_norm(m) if SPECTRAL_NORM else m

# -------------------------
# 1) Download data
# -------------------------
def download_data(tickers_dict, start=START_DATE, end=END_DATE):
    frames = {}
    for name, ticker in tickers_dict.items():
        print("Downloading:", ticker)
        t = yf.Ticker(ticker)
        df = t.history(start=start, end=end, auto_adjust=False)
        if df.empty:
            raise ValueError(f"No data returned for {ticker}")
        df = df[["Open","High","Low","Close","Volume"]].copy()
        df.columns = pd.MultiIndex.from_product([[name], df.columns])
        frames[name] = df
    all_df = None
    for name, df in frames.items():
        if all_df is None:
            all_df = df
        else:
            all_df = all_df.join(df, how="outer")
    all_df.columns = [f"{col[0]}_{col[1]}" for col in all_df.columns]
    all_df.sort_index(inplace=True)
    all_df = all_df.ffill().bfill()
    return all_df

raw = download_data(TICKERS)

# -------------------------
# 2) Feature engineering
# -------------------------
def add_features(df):
    df = df.copy()
    # primary reliance fields (flattened earlier)
    df["close"] = df["reliance_Close"]
    df["open"] = df["reliance_Open"]
    df["high"] = df["reliance_High"]
    df["low"] = df["reliance_Low"]
    df["volume"] = df["reliance_Volume"]

    df["logclose"] = np.log(df["close"])
    df["ret_1"] = df["logclose"].diff()
    df["ret_1_pct"] = df["close"].pct_change()
    df["ret_3"] = df["ret_1"].rolling(3).sum()
    df["ret_7"] = df["ret_1"].rolling(7).sum()
    df["rv_7"] = df["ret_1"].rolling(7).std()
    df["rv_14"] = df["ret_1"].rolling(14).std()
    df["sma_20"] = df["close"].rolling(20).mean()
    df["ema_20"] = df["close"].ewm(span=20, adjust=False).mean()

    # pandas_ta indicators if available
    try:
        df["rsi_14"] = ta.rsi(df["close"], length=14)
        bb = ta.bbands(df["close"], length=20, std=2)
        for col in bb.columns:
            df[col] = bb[col]
        df["atr_14"] = ta.atr(df["high"], df["low"], df["close"], length=14)
        df["obv"] = ta.obv(df["close"], df["volume"])
        macd = ta.macd(df["close"])
        for col in macd.columns:
            df[col] = macd[col]
    except Exception as e:
        print("pandas_ta failed:", e)

    # exogenous returns
    if "nifty_Close" in df.columns:
        df["nifty_close"] = df["nifty_Close"]
        df["n_logclose"] = np.log(df["nifty_close"])
        df["n_ret_1"] = df["n_logclose"].diff()
        df["n_rv_7"] = df["n_ret_1"].rolling(7).std()
    if "usdinr_Close" in df.columns:
        df["fx_close"] = df["usdinr_Close"]
        df["fx_ret_1"] = np.log(df["fx_close"]).diff()
    if "brent_Close" in df.columns:
        df["brent_close"] = df["brent_Close"]
        df["brent_ret_1"] = np.log(df["brent_close"]).diff()

    df = df.dropna().copy()
    return df

fe = add_features(raw)
print("Rows after fe:", len(fe))

# -------------------------
# 3) Prepare sequences & scaler
# -------------------------
FEATURES_FOR_GAN = [
    "open","high","low","close","volume",
    "ret_1","ret_3","ret_7","rv_7","rv_14",
    "sma_20","ema_20","rsi_14","atr_14","obv",
    "MACD_12_26_9","MACDh_12_26_9","MACDs_12_26_9",
    "n_ret_1","fx_ret_1","brent_ret_1"
]
FEATURES_FOR_GAN = [c for c in FEATURES_FOR_GAN if c in fe.columns]
print("Using features:", FEATURES_FOR_GAN)

scaler = StandardScaler()
scaler.fit(fe[FEATURES_FOR_GAN].values)

def make_sequences(df_features, seq_len=SEQ_LEN, step=STEP):
    arr = df_features.values
    N = len(arr)
    seqs = []
    for i in range(0, N - seq_len, step):
        seqs.append(arr[i:i+seq_len])
    return np.stack(seqs)

fe_scaled = pd.DataFrame(scaler.transform(fe[FEATURES_FOR_GAN]), index=fe.index, columns=FEATURES_FOR_GAN)
seqs = make_sequences(fe_scaled, seq_len=SEQ_LEN, step=STEP)
print("Sequences shape:", seqs.shape)

num_seqs = len(seqs)
test_ratio = 0.10
val_ratio = 0.10
n_test = int(num_seqs * test_ratio)
n_val = int(num_seqs * val_ratio)
n_train = num_seqs - n_val - n_test
train_seqs = seqs[:n_train]
val_seqs = seqs[n_train:n_train+n_val]
test_seqs = seqs[n_train+n_val:]
print("Train/Val/Test sequences:", train_seqs.shape, val_seqs.shape, test_seqs.shape)

class SeqDataset(Dataset):
    def __init__(self, arr):
        self.arr = arr.astype(np.float32)
    def __len__(self):
        return len(self.arr)
    def __getitem__(self, idx):
        return self.arr[idx]

train_loader = DataLoader(SeqDataset(train_seqs), batch_size=BATCH_SIZE, shuffle=True, drop_last=True, num_workers=WORKERS, pin_memory=True)
val_loader = DataLoader(SeqDataset(val_seqs), batch_size=BATCH_SIZE, shuffle=False, num_workers=WORKERS, pin_memory=True)
test_loader = DataLoader(SeqDataset(test_seqs), batch_size=BATCH_SIZE, shuffle=False, num_workers=WORKERS, pin_memory=True)

n_features = train_seqs.shape[-1]

# -------------------------
# 4) Models
# -------------------------
class EncoderRNN(nn.Module):
    def __init__(self, in_dim, hid_dim):
        super().__init__()
        self.gru = nn.GRU(in_dim, hid_dim, batch_first=True)
        self.proj = nn.Linear(hid_dim, LATENT_DIM)
    def forward(self, x):
        h, _ = self.gru(x)
        z = self.proj(h)
        return z

class GeneratorRNN(nn.Module):
    def __init__(self, latent_dim, hid_dim, out_dim):
        super().__init__()
        self.gru = nn.GRU(latent_dim, hid_dim, batch_first=True)
        self.out = nn.Linear(hid_dim, out_dim)
    def forward(self, z):
        h, _ = self.gru(z)
        out = self.out(h)
        return out

class RNNCritic(nn.Module):
    def __init__(self, in_dim, hid_dim):
        super().__init__()
        self.gru = snorm(nn.GRU(in_dim, hid_dim, batch_first=True))
        self.fc = snorm(nn.Sequential(nn.Linear(hid_dim, hid_dim//2), nn.ReLU(), nn.Linear(hid_dim//2,1)))
    def forward(self, x):
        h, _ = self.gru(x)
        last = h[:, -1, :]
        out = self.fc(last).squeeze(-1)
        return out

class PatchCriticCNN(nn.Module):
    def __init__(self, in_channels, channels=[32,64], kernel=5):
        super().__init__()
        layers = []
        ch = in_channels
        for h in channels:
            layers.append(snorm(nn.Conv1d(ch, h, kernel_size=kernel, padding=kernel//2)))
            layers.append(nn.LeakyReLU(0.2))
            layers.append(nn.AvgPool1d(2))
            ch = h
        self.cnn = nn.Sequential(*layers)
        # compute flattened length for head
        reduced_len = SEQ_LEN // (2 ** len(channels))
        self.head = snorm(nn.Sequential(nn.Flatten(), nn.Linear(ch * reduced_len, 1)))
    def forward(self, x):
        x = x.permute(0,2,1)
        h = self.cnn(x)
        out = self.head(h)
        return out.squeeze(-1)

encoder = EncoderRNN(n_features, HIDDEN_DIM).to(DEVICE)
generator = GeneratorRNN(LATENT_DIM, HIDDEN_DIM, n_features).to(DEVICE)
d_rnn = RNNCritic(n_features, HIDDEN_DIM).to(DEVICE)
d_patch = PatchCriticCNN(n_features, channels=[32,64]).to(DEVICE)

opt_EG = torch.optim.Adam(list(encoder.parameters()) + list(generator.parameters()), lr=LR_G, betas=(0.5,0.9))
opt_D_rnn = torch.optim.Adam(d_rnn.parameters(), lr=LR_D, betas=(0.5,0.9))
opt_D_patch = torch.optim.Adam(d_patch.parameters(), lr=LR_D, betas=(0.5,0.9))

mse = nn.MSELoss()

# -------------------------
# 5) AE pretraining
# -------------------------
def pretrain_autoencoder(epochs=EPOCHS_PRETRAIN):
    print("Pretraining autoencoder (encoder+generator)...")
    for ep in range(1, epochs+1):
        enc_loss_acc = []
        for batch in train_loader:
            x = batch.to(DEVICE)
            z = encoder(x)
            x_rec = generator(z)
            loss_recon = mse(x_rec, x)
            opt_EG.zero_grad()
            loss_recon.backward()
            opt_EG.step()
            enc_loss_acc.append(loss_recon.item())
        if ep % 5 == 0 or ep == 1:
            print(f"Pretrain epoch {ep}/{epochs} | Recon loss {np.mean(enc_loss_acc):.6f}")
    torch.save({
        "encoder": encoder.state_dict(),
        "generator": generator.state_dict()
    }, os.path.join(SAVE_DIR, "ae_pretrained.pth"))
    print("Autoencoder pretrained and saved.")

pretrain_autoencoder()

# -------------------------
# Utility: feature matching
# -------------------------
def feature_matching(real, fake):
    return torch.mean((real.mean(dim=(0,1)) - fake.mean(dim=(0,1)))**2)

# -------------------------
# 6) Gradient penalty (fixed)
# -------------------------
def gradient_penalty(critic, real, fake, device=DEVICE):
    """
    WGAN-GP gradient penalty that temporarily disables CuDNN for the critic forward pass.
    Uses .reshape to handle non-contiguous grads.
    """
    batch_size = real.size(0)
    alpha = torch.rand(batch_size, 1, 1, device=device)
    alpha = alpha.expand_as(real)

    interp = (alpha * real + (1 - alpha) * fake).requires_grad_(True)

    # disable cuDNN for critic forward to allow double-backward on RNNs
    with torch.backends.cudnn.flags(enabled=False):
        out = critic(interp)

    grads = torch.autograd.grad(
        outputs=out,
        inputs=interp,
        grad_outputs=torch.ones_like(out, device=device),
        create_graph=True,
        retain_graph=True,
        only_inputs=True
    )[0]  # (batch, seq_len, features)

    grads = grads.reshape(batch_size, -1)   # safe flatten even if non-contiguous
    grad_norm = grads.norm(2, dim=1)
    gp = ((grad_norm - 1) ** 2).mean()
    return gp

def sample_noise(batch_size, seq_len=SEQ_LEN, latent_dim=LATENT_DIM, device=DEVICE):
    return torch.randn(batch_size, seq_len, latent_dim, device=device)

# -------------------------
# 7) WGAN-GP adversarial training (two critics)
# -------------------------
def train_gan(epochs=EPOCHS_JOINT):
    print("Starting adversarial training...")
    step = 0
    for ep in range(1, epochs+1):
        d_losses = []
        g_losses = []
        pbar = tqdm(train_loader, desc=f"Epoch {ep}/{epochs}", leave=False)
        for batch in pbar:
            real = batch.to(DEVICE)
            bsz = real.size(0)

            # Train critics
            for _ in range(CRITIC_ITERS):
                z_noise = sample_noise(bsz)
                fake = generator(z_noise)

                # RNN critic
                opt_D_rnn.zero_grad()
                d_real = d_rnn(real)
                d_fake = d_rnn(fake.detach())
                loss_d_rnn = d_fake.mean() - d_real.mean()
                gp_rnn = gradient_penalty(d_rnn, real, fake.detach())
                loss_d_rnn_total = loss_d_rnn + LAMBDA_GP * gp_rnn
                loss_d_rnn_total.backward()
                opt_D_rnn.step()

                # Patch critic
                opt_D_patch.zero_grad()
                d_real_p = d_patch(real)
                d_fake_p = d_patch(fake.detach())
                loss_d_patch = d_fake_p.mean() - d_real_p.mean()
                gp_patch = gradient_penalty(d_patch, real, fake.detach())
                loss_d_patch_total = loss_d_patch + LAMBDA_GP * gp_patch
                loss_d_patch_total.backward()
                opt_D_patch.step()

            # Generator update
            z_noise = sample_noise(bsz)
            fake = generator(z_noise)
            g_adv = - d_rnn(fake).mean() - d_patch(fake).mean()
            fm = feature_matching(real, fake)
            g_total = g_adv + FM_WEIGHT * fm
            opt_EG.zero_grad()
            g_total.backward()
            opt_EG.step()

            d_losses.append((loss_d_rnn_total.item(), loss_d_patch_total.item()))
            g_losses.append(g_total.item())
            step += 1
            if step % 100 == 0:
                pbar.set_postfix({"d_rnn": np.mean([dl[0] for dl in d_losses]), "d_patch": np.mean([dl[1] for dl in d_losses]), "g": np.mean(g_losses)})

        if ep % PRINT_EVERY == 0:
            print(f"Epoch {ep} | D_rnn {np.mean([dl[0] for dl in d_losses]):.6f} | D_patch {np.mean([dl[1] for dl in d_losses]):.6f} | G {np.mean(g_losses):.6f}")

        if ep % 20 == 0:
            torch.save({
                "encoder": encoder.state_dict(),
                "generator": generator.state_dict(),
                "d_rnn": d_rnn.state_dict(),
                "d_patch": d_patch.state_dict()
            }, os.path.join(SAVE_DIR, f"gan_epoch_{ep}.pth"))

    torch.save({
        "encoder": encoder.state_dict(),
        "generator": generator.state_dict(),
        "d_rnn": d_rnn.state_dict(),
        "d_patch": d_patch.state_dict()
    }, os.path.join(SAVE_DIR, "gan_final.pth"))
    print("Adversarial training finished and models saved.")

train_gan()

# -------------------------
# 8) Generate synthetic sequences
# -------------------------
def generate_synthetic(n_samples, batch_size=BATCH_SIZE):
    generator.eval()
    results = []
    with torch.no_grad():
        for _ in range(math.ceil(n_samples / batch_size)):
            cur = min(batch_size, n_samples - len(results))
            z = torch.randn(cur, SEQ_LEN, LATENT_DIM, device=DEVICE)
            fake = generator(z).cpu().numpy()
            results.append(fake)
    arr = np.vstack(results)
    return arr

synth = generate_synthetic(1000)
print("Synthetic shape:", synth.shape)

# inverse scale (StandardScaler)
synth_flat = synth.reshape(-1, n_features)
synth_inv_flat = scaler.inverse_transform(synth_flat)
synth_inv = synth_inv_flat.reshape(synth.shape)

# plot example
if "close" in FEATURES_FOR_GAN:
    close_idx = FEATURES_FOR_GAN.index("close")
    plt.figure(figsize=(10,4))
    plt.plot(synth_inv[0,:,close_idx], label="synthetic close (sample 0)")
    plt.plot(scaler.inverse_transform(train_seqs[0])[:,close_idx], label="real close (train sample 0)", alpha=0.7)
    plt.legend(); plt.title("Example synthetic vs real (Close)"); plt.show()

# -------------------------
# 9) Evaluation: distribution & ACF comparisons
# -------------------------
def compare_distributions(real_seqs, synth_seqs, feature_idx=0, nplot=2000):
    real_vals = scaler.inverse_transform(real_seqs.reshape(-1, n_features))[:, feature_idx][:nplot]
    synth_vals = synth_seqs.reshape(-1, n_features)[:, feature_idx][:nplot]
    plt.figure(figsize=(12,4))
    plt.subplot(1,2,1)
    sns.kdeplot(real_vals, label="real")
    sns.kdeplot(synth_vals, label="synth")
    plt.title("Value KDE"); plt.legend()
    plt.subplot(1,2,2)
    def series_to_ret(arr):
        arr_flat = arr.reshape(-1, n_features)[:, feature_idx]
        lr = np.diff(np.log(arr_flat + 1e-9))
        return lr
    real_ret = series_to_ret(real_seqs)
    synth_ret = series_to_ret(synth_seqs)
    lags = 20
    real_acf = acf(real_ret, nlags=lags, fft=True)
    synth_acf = acf(synth_ret, nlags=lags, fft=True)
    plt.plot(real_acf, label="real ACF returns")
    plt.plot(synth_acf, label="synth ACF returns")
    plt.title("ACF returns comparison"); plt.legend(); plt.show()

if "close" in FEATURES_FOR_GAN:
    compare_distributions(train_seqs[:100], synth[:100], feature_idx=FEATURES_FOR_GAN.index("close"))

# -------------------------
# 10) Save outputs
# -------------------------
import joblib
np.save(os.path.join(SAVE_DIR, "synth_sequences.npy"), synth)
joblib.dump(scaler, os.path.join(SAVE_DIR, "scaler.pkl"))
print("Saved synthetic sequences and scaler to", SAVE_DIR)
print("Done.")
```
